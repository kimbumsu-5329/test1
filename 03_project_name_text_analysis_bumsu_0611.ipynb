```python
# 03-1. 라이브러리 불러오기 및 경로 설정

# ------------------------------------------------------------
# 0) 기본 라이브러리 불러오기
# ------------------------------------------------------------

import os
import re
from pathlib import Path
from datetime import datetime

import pandas as pd
import numpy as np

from google.colab import drive

# Google Drive 마운트
drive.mount('/content/drive', force_remount=True)

# ------------------------------------------------------------
# 1) MyDrive에서 파공장(PA Factory) 폴더 찾기
# ------------------------------------------------------------

MYDRIVE = Path('/content/drive/MyDrive')

pa_candidates = [
    p for p in MYDRIVE.iterdir()
    if p.is_dir() and 'PA Factory' in p.name
]

print("파공장 후보 폴더 수:", len(pa_candidates))

for i, p in enumerate(pa_candidates, start=1):
    print(f"{i}. {p}")
    print("   폴더명 repr:", repr(p.name))

if len(pa_candidates) != 1:
    raise FileNotFoundError(
        "파공장(PA Factory) 폴더를 정확히 1개 찾지 못했습니다. "
        "Google Drive 바로가기 추가 여부와 폴더명을 확인하세요."
    )

PA_FACTORY_DIR = pa_candidates[0]

# ------------------------------------------------------------
# 2) 03번 입력 파일 찾기
# ------------------------------------------------------------

target_file_name = 'all_clean_data_v1_20260523_0054_KST.parquet'

input_candidates = list(PA_FACTORY_DIR.rglob(target_file_name))

print("\n입력 파일 후보 수:", len(input_candidates))

for i, p in enumerate(input_candidates, start=1):
    print(f"{i}. {p}")

if len(input_candidates) != 1:
    raise FileNotFoundError(
        f"{target_file_name} 파일을 정확히 1개 찾지 못했습니다. "
        "파일명 또는 Drive 위치를 확인하세요."
    )

INPUT_FILE = input_candidates[0]
INPUT_DIR = INPUT_FILE.parent

# 입력 파일 구조:
# 03_중간결과표 / 01_v1_분석데이터 / all_clean_data_v1...
# 따라서 03_중간결과표 폴더는 INPUT_DIR의 부모 폴더입니다.
PROJECT_OUTPUT_DIR = INPUT_DIR.parent

# ------------------------------------------------------------
# 3) 03번 산출물 폴더 설정
# ------------------------------------------------------------

OUTPUT_DIR = PROJECT_OUTPUT_DIR / '03_v1_사업명텍스트분석'
TABLE_DIR = OUTPUT_DIR / 'tables'
FIGURE_DIR = OUTPUT_DIR / 'figures'

# 입력 파일이 확인된 뒤에만 출력 폴더 생성
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
TABLE_DIR.mkdir(parents=True, exist_ok=True)
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 4) 경로 확인 출력
# ------------------------------------------------------------

print("\n경로 설정 완료")
print("PA_FACTORY_DIR:", PA_FACTORY_DIR)
print("PROJECT_OUTPUT_DIR:", PROJECT_OUTPUT_DIR)
print("INPUT_DIR:", INPUT_DIR)
print("INPUT_FILE:", INPUT_FILE)
print("OUTPUT_DIR:", OUTPUT_DIR)
print("TABLE_DIR:", TABLE_DIR)
print("FIGURE_DIR:", FIGURE_DIR)

print("\n입력 파일 존재 여부:", INPUT_FILE.exists())
print("03번 산출물 폴더 존재 여부:", OUTPUT_DIR.exists())
```

03-2 코드 실행 단계

```python
INPUT_FILE
```

생성할 변수 또는 출력
| 변수명               | 의미                  |
| ----------------- | ------------------- |
| `df`              | V1 원본 분석 데이터        |
| `required_for_03` | 03번 분석에 필요한 핵심 열 목록 |
| `missing_for_03`  | 03번 핵심 열 중 누락된 열 목록 |
이 단계에서는 원본 데이터프레임 df만 생성하고, 아직 새로운 분석 열은 만들지 않습니다.


```python
# 03-2. V1 데이터 불러오기 및 기본 구조 확인

# ------------------------------------------------------------
# 1) V1 parquet 데이터 불러오기
# ------------------------------------------------------------

df = pd.read_parquet(INPUT_FILE)

# ------------------------------------------------------------
# 2) 전체 구조 확인
# ------------------------------------------------------------

print("V1 데이터 불러오기 완료")
print("- 행 수:", df.shape[0])
print("- 열 수:", df.shape[1])

print("\n열 이름 목록:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i:02d}. {col}")

# ------------------------------------------------------------
# 3) 03번 사업명 텍스트 분석에 필요한 핵심 열 확인
# ------------------------------------------------------------

required_for_03 = ['사업명']
missing_for_03 = [col for col in required_for_03 if col not in df.columns]

print("\n03번 핵심 열 확인")
print("- 필요한 열:", required_for_03)
print("- 누락된 열:", missing_for_03)

if missing_for_03:
    raise ValueError(f"03번 분석에 필요한 핵심 열이 없습니다: {missing_for_03}")

# ------------------------------------------------------------
# 4) 사업명 열의 기본 상태 확인
# ------------------------------------------------------------

total_rows = len(df)
missing_project_name = df['사업명'].isna().sum()
blank_project_name = df['사업명'].astype(str).str.strip().eq('').sum()

print("\n사업명 열 기본 상태")
print("- 전체 행 수:", total_rows)
print("- 사업명 결측 행 수:", missing_project_name)
print("- 사업명 공백 문자열 행 수:", blank_project_name)
print("- 사업명 결측 비율:", round(missing_project_name / total_rows * 100, 2), "%")
print("- 사업명 공백 문자열 비율:", round(blank_project_name / total_rows * 100, 2), "%")

# ------------------------------------------------------------
# 5) 참고 열 결측 상태 확인
# ------------------------------------------------------------
# 주의:
# 03번에서는 평가종류, 사업비합계, 보조금, 사업자부담 결측을 이유로 행을 제외하지 않습니다.
# 다만 V1 데이터 상태 이해를 위해 참고용으로만 확인합니다.

reference_cols = ['평가종류', '사업비합계', '보조금', '사업자부담']
existing_reference_cols = [col for col in reference_cols if col in df.columns]

print("\n참고 열 결측 상태")
print("※ 아래 열의 결측은 03번 사업명 텍스트 분석에서 행 제외 기준으로 사용하지 않습니다.")

if existing_reference_cols:
    reference_missing_summary = (
        df[existing_reference_cols]
        .isna()
        .sum()
        .reset_index()
    )
    reference_missing_summary.columns = ['열 이름', '결측 행 수']
    reference_missing_summary['전체 행 수'] = total_rows
    reference_missing_summary['결측 비율(%)'] = (
        reference_missing_summary['결측 행 수'] / total_rows * 100
    ).round(2)

    display(reference_missing_summary)
else:
    print("참고 열이 현재 데이터에 없습니다:", reference_cols)

# ------------------------------------------------------------
# 6) 사업명 샘플 확인
# ------------------------------------------------------------

print("\n사업명 샘플 20개")
display(
    df[['사업명']]
    .head(20)
)
```

03-3. 사업명 원자료 기본 점검
1. 이 단계의 목적

이 단계는 원본 사업명 열을 수정하지 않고, 텍스트 분석 전에 확인해야 할 기초 문제를 점검하는 단계입니다.

주로 다음을 확인합니다.

사업명 결측 14행이 어떤 행인지
사업명 길이가 지나치게 짧거나 긴 사례가 있는지
줄바꿈, 탭, 중복 공백, 앞뒤 공백이 있는지
합계, 소계, 총계 같은 합계행 의심 사례가 있는지
동일한 사업명이 얼마나 반복되는지

아직 사업명_clean은 만들지 않습니다.

```python
# 03-3. 사업명 원자료 기본 점검

# ------------------------------------------------------------
# 1) 원본 사업명 열을 확인용 변수로 분리
# ------------------------------------------------------------
# 주의:
# project_name_raw는 원본 df['사업명']을 참조하는 변수입니다.
# 이 단계에서는 df['사업명'] 값을 절대 수정하지 않습니다.

project_name_raw = df['사업명']

# 결측값은 확인 편의를 위해 임시로 빈 문자열로 바꿉니다.
# 이 변수는 진단용이며, 원본 df에는 영향을 주지 않습니다.
project_name_text = project_name_raw.fillna('').astype(str)

# 앞뒤 공백을 제거한 임시 텍스트
project_name_stripped = project_name_text.str.strip()

total_rows = len(df)

print("03-3. 사업명 원자료 기본 점검")
print("- 전체 행 수:", total_rows)
print("- 원본 사업명 열 수정 여부: 수정하지 않음")


# ------------------------------------------------------------
# 2) 결측 및 공백 사업명 행 확인
# ------------------------------------------------------------

missing_mask = project_name_raw.isna()
blank_mask = project_name_raw.notna() & project_name_stripped.eq('')

missing_or_blank_mask = missing_mask | blank_mask

# 확인할 때 같이 보면 좋은 열들만 선택합니다.
# 데이터에 없는 열은 자동으로 제외됩니다.
context_cols = [
    '시도', '시군구', '지자체명', '지자체유형',
    '원본파일명', 'source_file', 'file_name',
    '사업명', '보조사업자', '평가결과', '평가종류',
    '사업비합계', '보조금', '사업자부담'
]

existing_context_cols = [col for col in context_cols if col in df.columns]

missing_project_name_rows = df.loc[missing_or_blank_mask, existing_context_cols].copy()

print("\n[1] 사업명 결측/공백 행")
print("- 사업명 결측 행 수:", int(missing_mask.sum()))
print("- 사업명 공백 문자열 행 수:", int(blank_mask.sum()))
print("- 결측 또는 공백 전체 행 수:", len(missing_project_name_rows))

display(missing_project_name_rows.head(30))


# ------------------------------------------------------------
# 3) 사업명 길이 분포 확인
# ------------------------------------------------------------

# 결측/공백이 아닌 사업명만 대상으로 길이 계산
valid_project_name = project_name_stripped[~missing_or_blank_mask]
project_name_length = valid_project_name.str.len()

print("\n[2] 사업명 길이 분포")
display(project_name_length.describe().to_frame(name='사업명_글자수'))

# 너무 짧은 사업명 후보
# 기준은 우선 3글자 이하로 잡고, 실제 사례를 본 뒤 조정할 수 있습니다.
short_mask = (~missing_or_blank_mask) & (project_name_stripped.str.len() <= 3)
short_project_name_rows = df.loc[short_mask, existing_context_cols].copy()

print("\n[3] 너무 짧은 사업명 후보")
print("- 3글자 이하 사업명 수:", len(short_project_name_rows))
display(short_project_name_rows.head(30))

# 너무 긴 사업명 후보
# 기준은 우선 80글자 이상으로 잡고, 실제 사례를 본 뒤 조정할 수 있습니다.
long_mask = (~missing_or_blank_mask) & (project_name_stripped.str.len() >= 80)
long_project_name_rows = df.loc[long_mask, existing_context_cols].copy()

print("\n[4] 너무 긴 사업명 후보")
print("- 80글자 이상 사업명 수:", len(long_project_name_rows))
display(long_project_name_rows.head(30))


# ------------------------------------------------------------
# 4) 공백·줄바꿈·탭 등 기본 형식 문제 확인
# ------------------------------------------------------------

newline_mask = project_name_text.str.contains(r'\n|\r', regex=True, na=False)
tab_mask = project_name_text.str.contains(r'\t', regex=True, na=False)
multi_space_mask = project_name_text.str.contains(r' {2,}', regex=True, na=False)
leading_trailing_space_mask = project_name_raw.notna() & (project_name_text != project_name_stripped)

# 특수 따옴표나 이상한 공백 후보
# 예: ‘ ’ “ ”, non-breaking space(\u00A0)
special_quote_mask = project_name_text.str.contains(r'[‘’“”]', regex=True, na=False)
nbsp_mask = project_name_text.str.contains('\u00A0', regex=False, na=False)

issue_summary = pd.DataFrame({
    '점검항목': [
        '사업명 결측',
        '사업명 공백 문자열',
        '줄바꿈 포함',
        '탭 포함',
        '연속 공백 2칸 이상 포함',
        '앞뒤 공백 포함',
        '특수 따옴표 포함',
        'NBSP 특수공백 포함'
    ],
    '행 수': [
        int(missing_mask.sum()),
        int(blank_mask.sum()),
        int(newline_mask.sum()),
        int(tab_mask.sum()),
        int(multi_space_mask.sum()),
        int(leading_trailing_space_mask.sum()),
        int(special_quote_mask.sum()),
        int(nbsp_mask.sum())
    ]
})

issue_summary['전체 대비 비율(%)'] = (issue_summary['행 수'] / total_rows * 100).round(3)

print("\n[5] 사업명 기본 형식 문제 요약")
display(issue_summary)


# ------------------------------------------------------------
# 5) 형식 문제별 샘플 확인
# ------------------------------------------------------------

def show_issue_sample(mask, title, n=10):
    """문제 후보 행을 보기 좋게 출력하는 보조 함수"""
    sample = df.loc[mask, existing_context_cols].head(n).copy()
    print(f"\n{title}")
    print("- 후보 행 수:", int(mask.sum()))
    display(sample)

show_issue_sample(newline_mask, "[샘플] 줄바꿈 포함 사업명")
show_issue_sample(multi_space_mask, "[샘플] 연속 공백 2칸 이상 포함 사업명")
show_issue_sample(leading_trailing_space_mask, "[샘플] 앞뒤 공백 포함 사업명")
show_issue_sample(special_quote_mask, "[샘플] 특수 따옴표 포함 사업명")
show_issue_sample(nbsp_mask, "[샘플] NBSP 특수공백 포함 사업명")


# ------------------------------------------------------------
# 6) 합계행/소계행/총계행 의심 사례 확인
# ------------------------------------------------------------
# 주의:
# 이 단계에서는 실제로 제외하지 않습니다.
# '합계', '소계', '총계'라는 단어가 포함된 사업명이 실제 사업명일 수도 있으므로
# 반드시 사람이 확인해야 합니다.

total_keyword_mask = project_name_text.str.contains(r'합계|소계|총계', regex=True, na=False)
suspected_total_rows = df.loc[total_keyword_mask, existing_context_cols].copy()

print("\n[6] 합계행/소계행/총계행 의심 사례")
print("- 후보 행 수:", len(suspected_total_rows))
display(suspected_total_rows.head(30))


# ------------------------------------------------------------
# 7) 반복 등장 사업명 확인
# ------------------------------------------------------------
# 동일한 사업명이 반복 등장하는 것은 오류일 수도 있고,
# 여러 지자체에서 같은 이름의 사업을 운영한 정상 사례일 수도 있습니다.
# 여기서는 상위 반복 사업명만 확인합니다.

duplicate_project_names = (
    project_name_stripped[~missing_or_blank_mask]
    .value_counts()
    .reset_index()
)

duplicate_project_names.columns = ['사업명', '반복횟수']
duplicate_project_names = duplicate_project_names[duplicate_project_names['반복횟수'] >= 2]

print("\n[7] 반복 등장 사업명 상위 30개")
print("- 2회 이상 반복 등장한 사업명 종류 수:", len(duplicate_project_names))
display(duplicate_project_names.head(30))


# ------------------------------------------------------------
# 8) 다음 단계 판단을 위한 요약 출력
# ------------------------------------------------------------

print("\n03-3 점검 완료")
print("다음 단계로 넘어가기 전에 아래를 확인하세요.")
print("1) 사업명 결측 14행이 실제로 분석 제외 후보인지")
print("2) 줄바꿈/연속공백/앞뒤공백이 사업명_clean에서 정리 가능한 수준인지")
print("3) 합계/소계/총계 후보가 실제 합계행인지")
print("4) 반복 등장 사업명이 오류가 아니라 정상 반복인지")
print("5) 특수 따옴표·특수공백이 정제 규칙에 포함되어야 하는지")
```

03-4. 사업명_clean 생성 및 정제 전후 비교
1. 이 단계의 목적

이 단계에서는 원본 사업명 열을 절대 수정하지 않고, 분석용 새 열인 사업명_clean을 생성합니다.

이번 정제는 의미를 바꾸는 작업이 아니라, 텍스트 분석 전에 필요한 최소한의 형식 정리입니다.

정리 대상은 다음입니다.

줄바꿈 \n, \r → 공백으로 변환
탭 \t → 공백으로 변환
특수공백 → 일반 공백으로 변환
연속 공백 2칸 이상 → 1칸으로 축소
앞뒤 공백 제거
특수 따옴표 일부 정규화
한글 유니코드 정규화

아직 글자 간 공백 오류를 고치지는 않습니다.
예를 들어 도 내 행 사장 같은 유형은 별도 진단 단계에서 확인합니다.

2. 입력 데이터

이전 단계까지 사용한 원본 데이터프레임입니다.

df

필수 열은 다음입니다.

사업명

3. 생성할 변수 또는 출력
| 생성 변수                 | 의미                  |
| --------------------- | ------------------- |
| `사업명_clean`           | 분석용으로 최소 정제한 사업명    |
| `사업명_key`             | 중복·비교용 사업명 키        |
| `사업명_is_missing`      | 원본 사업명 결측 여부        |
| `사업명_clean_is_empty`  | 정제 후 빈 문자열 여부       |
| `사업명_clean_changed`   | 원본과 정제본이 달라졌는지 여부   |
| `사업명_clean_length`    | 정제 후 글자 수           |
| `clean_change_sample` | 정제 전후가 달라진 사례 확인용 표 |

이 셀에서는 아직 파일 저장을 하지 않습니다.
먼저 정제 결과가 안전한지 화면에서 확인합니다.

```python
# 03-4. 사업명_clean 생성 및 정제 전후 비교

# ------------------------------------------------------------
# 0) 유니코드 정규화를 위한 라이브러리
# ------------------------------------------------------------

import unicodedata

# ------------------------------------------------------------
# 1) 사업명 정제 함수 정의
# ------------------------------------------------------------
# 주의:
# 이 함수는 원본 df['사업명']을 직접 수정하지 않습니다.
# 새 열 df['사업명_clean']을 만들 때만 사용합니다.

def clean_project_name(value):
    """
    사업명 텍스트를 분석용으로 최소 정제하는 함수입니다.

    정제 원칙:
    - 원본 의미를 바꾸지 않는다.
    - 모든 공백을 제거하지 않는다.
    - 줄바꿈, 탭, 특수공백, 중복공백, 앞뒤공백만 정리한다.
    - 글자 간 공백 오류는 이 단계에서 고치지 않는다.
    """

    # 결측값은 그대로 pd.NA로 둡니다.
    if pd.isna(value):
        return pd.NA

    # 문자열로 변환
    text = str(value)

    # 한글 유니코드 정규화
    # Colab/Drive에서 한글이 분해형으로 보이는 경우가 있어 NFC로 통일합니다.
    text = unicodedata.normalize('NFC', text)

    # 특수공백을 일반 공백으로 변환
    text = text.replace('\u00A0', ' ')  # NBSP
    text = text.replace('\u200B', '')   # zero-width space

    # 줄바꿈과 탭을 공백으로 변환
    text = re.sub(r'[\r\n\t]+', ' ', text)

    # 특수 따옴표를 일반 따옴표로 통일
    # 의미를 제거하지 않고 형태만 통일합니다.
    text = text.replace('‘', "'").replace('’', "'")
    text = text.replace('“', '"').replace('”', '"')

    # 연속 공백을 하나의 공백으로 축소
    text = re.sub(r'\s+', ' ', text)

    # 앞뒤 공백 제거
    text = text.strip()

    # 정제 후 빈 문자열이면 pd.NA 처리
    if text == '':
        return pd.NA

    return text


# ------------------------------------------------------------
# 2) 원본 보존 상태에서 분석용 새 열 생성
# ------------------------------------------------------------

# 원본 사업명 결측 여부
df['사업명_is_missing'] = df['사업명'].isna()

# 분석용 정제 사업명
df['사업명_clean'] = df['사업명'].apply(clean_project_name)

# 비교·중복 확인용 key
# 현재는 사업명_clean과 같게 두되, 나중에 괄호/연도 제거 버전이 필요하면 별도 key를 만들 수 있습니다.
df['사업명_key'] = df['사업명_clean']

# 정제 후 빈 값 여부
df['사업명_clean_is_empty'] = df['사업명_clean'].isna()

# 정제 후 글자 수
df['사업명_clean_length'] = df['사업명_clean'].astype('string').str.len()

# 원본과 정제본이 달라졌는지 확인
# 결측 비교 오류를 피하기 위해 문자열 임시값으로 비교합니다.
original_for_compare = df['사업명'].fillna('<NA>').astype(str)
clean_for_compare = df['사업명_clean'].fillna('<NA>').astype(str)

df['사업명_clean_changed'] = original_for_compare != clean_for_compare


# ------------------------------------------------------------
# 3) 정제 결과 요약
# ------------------------------------------------------------

clean_summary = pd.DataFrame({
    '항목': [
        '전체 행 수',
        '원본 사업명 결측 행 수',
        '정제 후 사업명_clean 결측 행 수',
        '정제 전후 변경 행 수',
        '정제 후 3글자 이하 행 수',
        '정제 후 80글자 이상 행 수'
    ],
    '값': [
        len(df),
        int(df['사업명_is_missing'].sum()),
        int(df['사업명_clean_is_empty'].sum()),
        int(df['사업명_clean_changed'].sum()),
        int((df['사업명_clean_length'] <= 3).sum()),
        int((df['사업명_clean_length'] >= 80).sum())
    ]
})

print("03-4. 사업명_clean 생성 완료")
display(clean_summary)


# ------------------------------------------------------------
# 4) 정제 전후 변경 사례 확인
# ------------------------------------------------------------
# repr()을 사용하면 눈에 잘 보이지 않는 앞뒤 공백, 줄바꿈, 따옴표 차이를 확인할 수 있습니다.

clean_change_sample = (
    df.loc[df['사업명_clean_changed'], ['사업명', '사업명_clean']]
    .head(30)
    .copy()
)

clean_change_sample['사업명_repr'] = clean_change_sample['사업명'].apply(lambda x: repr(x))
clean_change_sample['사업명_clean_repr'] = clean_change_sample['사업명_clean'].apply(lambda x: repr(x))

print("\n[1] 사업명 정제 전후 변경 사례 30개")
display(clean_change_sample[['사업명_repr', '사업명_clean_repr']])


# ------------------------------------------------------------
# 5) 사업명_clean 결측 행 확인
# ------------------------------------------------------------
# 원본 사업명이 결측이면 사업명_clean도 결측이어야 정상입니다.

context_cols = [
    '시도', '시군구', '지자체명', '지자체유형',
    '원본파일명', 'source_file', 'file_name',
    '사업명', '사업명_clean',
    '보조사업자', '평가결과', '평가종류',
    '사업비합계', '보조금', '사업자부담'
]

existing_context_cols = [col for col in context_cols if col in df.columns]

print("\n[2] 사업명_clean 결측 행")
display(
    df.loc[df['사업명_clean_is_empty'], existing_context_cols]
    .head(30)
)


# ------------------------------------------------------------
# 6) 정제 후에도 남아 있는 기본 형식 문제 재점검
# ------------------------------------------------------------

clean_text = df['사업명_clean'].fillna('').astype(str)

clean_newline_count = clean_text.str.contains(r'\n|\r', regex=True).sum()
clean_tab_count = clean_text.str.contains(r'\t', regex=True).sum()
clean_multi_space_count = clean_text.str.contains(r' {2,}', regex=True).sum()
clean_leading_trailing_count = (clean_text != clean_text.str.strip()).sum()
clean_special_quote_count = clean_text.str.contains(r'[‘’“”]', regex=True).sum()
clean_nbsp_count = clean_text.str.contains('\u00A0', regex=False).sum()

clean_issue_summary = pd.DataFrame({
    '점검항목': [
        '줄바꿈 포함',
        '탭 포함',
        '연속 공백 2칸 이상 포함',
        '앞뒤 공백 포함',
        '특수 따옴표 포함',
        'NBSP 특수공백 포함'
    ],
    '정제 후 행 수': [
        int(clean_newline_count),
        int(clean_tab_count),
        int(clean_multi_space_count),
        int(clean_leading_trailing_count),
        int(clean_special_quote_count),
        int(clean_nbsp_count)
    ]
})

print("\n[3] 사업명_clean 기본 형식 문제 재점검")
display(clean_issue_summary)


# ------------------------------------------------------------
# 7) 다음 단계용 텍스트 분석 대상 행 수 확인
# ------------------------------------------------------------
# 03번 텍스트 분석에서는 사업명_clean이 있는 행을 기본 대상으로 사용합니다.
# 원본 df에서는 행을 삭제하지 않고, 분석용 df_text를 따로 만듭니다.

df_text = df.loc[~df['사업명_clean_is_empty']].copy()

print("\n[4] 다음 단계 텍스트 분석 대상")
print("- 원본 df 행 수:", len(df))
print("- 사업명_clean이 있는 df_text 행 수:", len(df_text))
print("- 사업명_clean 결측으로 제외되는 행 수:", len(df) - len(df_text))

print("\n03-4 완료")
print("다음 단계에서는 df_text를 기준으로 글자 간 공백 오류 의심 사례를 진단합니다.")
```

03-5. 글자 간 공백 오류 의심 사례 진단
1. 이 단계의 목적

이 단계는 사업명_clean에서 글자 사이가 비정상적으로 띄어진 사업명 후보를 찾는 단계입니다.

예를 들면 아래와 같은 사례를 의심합니다.

도 내 행 사장
문 화 예 술
지 역 사 회

다만 정상적인 띄어쓰기와 구분이 어렵기 때문에, 이 단계에서는 자동 수정하지 않습니다.
후보를 뽑고 사람이 직접 확인한 뒤, 반복되는 패턴이 있을 때만 다음 단계에서 규칙화합니다.

2. 입력 데이터

이전 단계에서 만든 텍스트 분석용 데이터프레임입니다
df_text

필수 열은 다음입니다.
사업명_clean

3. 생성할 변수 또는 출력
| 변수명                      | 의미                |
| ------------------------ | ----------------- |
| `char_space_issue_rows`  | 글자 간 공백 오류 의심 사례  |
| `char_space_summary`     | 의심 사례 수 요약        |
| `char_space_run_text`    | 의심되는 한글 1글자 연속 구간 |
| `char_space_run_count`   | 의심 구간 개수          |
| `char_space_max_run_len` | 가장 긴 1글자 연속 구간 길이 |


```python
# 03-5. 글자 간 공백 오류 의심 사례 진단

# ------------------------------------------------------------
# 1) 글자 간 공백 의심 구간을 찾는 함수 정의
# ------------------------------------------------------------

def find_single_hangul_runs(text, min_run_len=3):
    """
    한글 1글자 토큰이 공백을 사이에 두고 연속으로 등장하는 구간을 찾습니다.

    예:
    - '도 내 행 사장' -> ['도 내 행']
    - '문 화 예 술 행사' -> ['문 화 예 술']

    주의:
    - 이 함수는 오류를 확정하지 않습니다.
    - 정상적인 표현도 후보로 잡힐 수 있으므로 사람이 직접 확인해야 합니다.
    """

    if pd.isna(text):
        return []

    tokens = str(text).split()

    runs = []
    current_run = []

    for tok in tokens:
        # 한글 1글자 토큰인지 확인
        if re.fullmatch(r'[가-힣]', tok):
            current_run.append(tok)
        else:
            # 지금까지 쌓인 1글자 토큰 연속 구간이 기준 이상이면 저장
            if len(current_run) >= min_run_len:
                runs.append(' '.join(current_run))
            current_run = []

    # 마지막 구간 처리
    if len(current_run) >= min_run_len:
        runs.append(' '.join(current_run))

    return runs


# ------------------------------------------------------------
# 2) df_text에 의심 구간 정보 생성
# ------------------------------------------------------------
# 주의:
# 원본 df가 아니라 df_text에만 진단용 열을 추가합니다.

df_text['char_space_runs'] = df_text['사업명_clean'].apply(find_single_hangul_runs)

df_text['char_space_run_count'] = df_text['char_space_runs'].apply(len)

df_text['char_space_max_run_len'] = df_text['char_space_runs'].apply(
    lambda runs: max([len(run.split()) for run in runs], default=0)
)

df_text['char_space_issue_flag'] = df_text['char_space_run_count'] > 0

# 사람이 보기 쉽게 의심 구간을 문자열로 합칩니다.
df_text['char_space_run_text'] = df_text['char_space_runs'].apply(
    lambda runs: ' | '.join(runs) if len(runs) > 0 else ''
)


# ------------------------------------------------------------
# 3) 의심 사례 요약
# ------------------------------------------------------------

char_space_issue_rows = df_text.loc[df_text['char_space_issue_flag']].copy()

char_space_summary = pd.DataFrame({
    '항목': [
        '텍스트 분석 대상 행 수',
        '글자 간 공백 의심 행 수',
        '글자 간 공백 의심 비율(%)',
        '최대 1글자 연속 구간 길이'
    ],
    '값': [
        len(df_text),
        len(char_space_issue_rows),
        round(len(char_space_issue_rows) / len(df_text) * 100, 3),
        int(df_text['char_space_max_run_len'].max())
    ]
})

print("03-5. 글자 간 공백 오류 의심 사례 진단")
display(char_space_summary)


# ------------------------------------------------------------
# 4) 의심 사례 샘플 확인
# ------------------------------------------------------------

context_cols = [
    '시도', '시군구', '지자체명', '지자체유형',
    '원본파일명', 'source_file', 'file_name',
    '사업명', '사업명_clean',
    'char_space_run_text',
    'char_space_run_count',
    'char_space_max_run_len',
    '보조사업자', '평가결과', '평가종류'
]

existing_context_cols = [col for col in context_cols if col in df_text.columns]

print("\n[1] 글자 간 공백 의심 사례 상위 50개")
display(
    char_space_issue_rows
    .sort_values(['char_space_max_run_len', 'char_space_run_count'], ascending=False)
    [existing_context_cols]
    .head(50)
)


# ------------------------------------------------------------
# 5) 의심 구간 빈도 확인
# ------------------------------------------------------------
# 같은 의심 구간이 반복되면 실제 오류일 가능성이 상대적으로 높습니다.

char_space_run_freq = (
    char_space_issue_rows['char_space_runs']
    .explode()
    .dropna()
    .value_counts()
    .reset_index()
)

char_space_run_freq.columns = ['의심구간', '빈도']

print("\n[2] 반복되는 글자 간 공백 의심 구간 상위 30개")
display(char_space_run_freq.head(30))


# ------------------------------------------------------------
# 6) 기준 완화/강화 참고용: 1글자 연속 길이별 분포
# ------------------------------------------------------------

run_len_distribution = (
    df_text.loc[df_text['char_space_max_run_len'] > 0, 'char_space_max_run_len']
    .value_counts()
    .sort_index()
    .reset_index()
)

run_len_distribution.columns = ['가장 긴 1글자 연속 구간 길이', '행 수']

print("\n[3] 1글자 연속 구간 길이별 분포")
display(run_len_distribution)


# ------------------------------------------------------------
# 7) 다음 단계 안내
# ------------------------------------------------------------

print("\n03-5 완료")
print("이 단계에서는 후보만 찾았습니다.")
print("다음으로 확인할 것:")
print("1) 의심 사례 상위 50개가 실제 오류인지")
print("2) 반복 의심 구간이 특정 지자체나 파일에 몰려 있는지")
print("3) 정상 표현이 많이 잡힌다면 기준을 더 엄격하게 조정할지")
print("4) 실제 오류 패턴이 확인되면 자동 수정이 아니라 규칙 후보로 기록할지")
```

03-6. 글자 간 공백 의심 사례 검토표 저장
1. 이 단계의 목적

이 단계는 03-5에서 탐지한 글자 간 공백 의심 사례 27건을 별도 검토표로 저장하는 단계입니다.

아직 자동 수정하지 않습니다.
대신 다음과 같은 검토용 열을 추가합니다.

검토판단
수정필요여부
수정후_사업명_clean
검토메모

이 표를 보고 사람이 직접 판단한 뒤, 이후 반복 가능한 보정 규칙을 만들지 결정합니다.

2. 입력 데이터

이전 단계에서 생성된 데이터입니다.

df_text
char_space_issue_rows

필수 열은 다음입니다.

사업명
사업명_clean
char_space_run_text
char_space_run_count
char_space_max_run_len

3. 생성할 변수 또는 출력
| 변수 또는 파일                                               | 의미                    |
| ------------------------------------------------------ | --------------------- |
| `char_space_review`                                    | 글자 간 공백 의심 사례 검토표     |
| `03_char_space_issue_candidates_v1_YYYYMMDD_HHMM.xlsx` | Drive에 저장되는 검토용 Excel |
| `03_char_space_issue_candidates_v1_YYYYMMDD_HHMM.csv`  | Drive에 저장되는 검토용 CSV   |

저장 위치는 03번 산출물 폴더의 tables입니다.

03_v1_사업명텍스트분석/tables

```python
# 03-6. 글자 간 공백 의심 사례 검토표 저장

# ------------------------------------------------------------
# 1) 검토표에 포함할 열 정리
# ------------------------------------------------------------

review_cols = [
    '시도', '시군구', '지자체명', '지자체유형',
    '원본파일명', 'source_file', 'file_name',
    '사업명',
    '사업명_clean',
    'char_space_run_text',
    'char_space_run_count',
    'char_space_max_run_len',
    '보조사업자',
    '평가결과',
    '평가종류',
    '사업비합계',
    '보조금',
    '사업자부담'
]

existing_review_cols = [col for col in review_cols if col in char_space_issue_rows.columns]

# ------------------------------------------------------------
# 2) 글자 간 공백 의심 사례 검토표 생성
# ------------------------------------------------------------

char_space_review = (
    char_space_issue_rows
    .sort_values(
        ['char_space_max_run_len', 'char_space_run_count'],
        ascending=False
    )
    [existing_review_cols]
    .copy()
)

# 원래 데이터의 index를 남겨두면 나중에 df_text와 연결하기 쉽습니다.
char_space_review = char_space_review.reset_index().rename(columns={'index': '원본_index'})

# ------------------------------------------------------------
# 3) 수동 검토용 빈 열 추가
# ------------------------------------------------------------
# 이 열들은 사람이 직접 채우기 위한 열입니다.
# 코드가 임의로 수정값을 만들지 않습니다.

char_space_review['검토판단'] = ''
char_space_review['수정필요여부'] = ''
char_space_review['수정후_사업명_clean'] = ''
char_space_review['검토메모'] = ''

# 보기 편하도록 검토용 열을 앞쪽에 배치합니다.
front_cols = [
    '원본_index',
    '시도',
    '지자체명',
    '사업명',
    '사업명_clean',
    'char_space_run_text',
    '검토판단',
    '수정필요여부',
    '수정후_사업명_clean',
    '검토메모'
]

front_cols = [col for col in front_cols if col in char_space_review.columns]
other_cols = [col for col in char_space_review.columns if col not in front_cols]

char_space_review = char_space_review[front_cols + other_cols]

# ------------------------------------------------------------
# 4) 저장 파일명 설정
# ------------------------------------------------------------

timestamp = datetime.now().strftime('%Y%m%d_%H%M')

char_space_review_xlsx = TABLE_DIR / f'03_char_space_issue_candidates_v1_{timestamp}.xlsx'
char_space_review_csv = TABLE_DIR / f'03_char_space_issue_candidates_v1_{timestamp}.csv'

# ------------------------------------------------------------
# 5) 검토표 저장
# ------------------------------------------------------------

char_space_review.to_excel(char_space_review_xlsx, index=False)
char_space_review.to_csv(char_space_review_csv, index=False, encoding='utf-8-sig')

# ------------------------------------------------------------
# 6) 저장 결과 및 검토표 확인
# ------------------------------------------------------------

print("03-6. 글자 간 공백 의심 사례 검토표 저장 완료")
print("- 검토 대상 행 수:", len(char_space_review))
print("- Excel 저장 경로:", char_space_review_xlsx)
print("- CSV 저장 경로:", char_space_review_csv)

print("\n저장 파일 존재 여부")
print("- Excel:", char_space_review_xlsx.exists())
print("- CSV:", char_space_review_csv.exists())

print("\n[검토표 미리보기]")
display(char_space_review.head(30))
```

03-7. 형태소 분석기 설치 및 테스트
1. 이 단계의 목적

이 단계는 사업명 토큰화를 하기 전에 한국어 형태소 분석기인 kiwipiepy를 설치하고, 실제 사업명 몇 개에 대해 형태소 분석이 제대로 작동하는지 확인하는 단계입니다.

아직 전체 데이터에 토큰화를 적용하지 않습니다.
먼저 샘플 몇 개만 분석해서 명사·고유명사·외래어 등이 어떻게 분리되는지 확인합니다.

2. 입력 데이터

이전 단계에서 만든 텍스트 분석용 데이터프레임입니다.

df_text

필수 열은 다음입니다.

사업명_clean

3. 생성할 변수 또는 출력
| 변수명                | 의미                 |
| ------------------ | ------------------ |
| `kiwi`             | Kiwi 형태소 분석기 객체    |
| `sample_names`     | 형태소 분석 테스트용 사업명 샘플 |
| `kiwi_test_result` | 샘플 사업명의 형태소 분석 결과표 |


```python
# 03-7. 형태소 분석기 설치 및 테스트

# ------------------------------------------------------------
# 1) kiwipiepy 설치
# ------------------------------------------------------------
# Colab 런타임에서 처음 실행하는 경우 설치가 필요합니다.
# 이미 설치되어 있으면 빠르게 넘어갑니다.

!pip install -q kiwipiepy

# ------------------------------------------------------------
# 2) 라이브러리 불러오기
# ------------------------------------------------------------

from kiwipiepy import Kiwi

# 형태소 분석기 객체 생성
kiwi = Kiwi()

print("kiwipiepy 설치 및 Kiwi 객체 생성 완료")


# ------------------------------------------------------------
# 3) 테스트용 사업명 샘플 구성
# ------------------------------------------------------------
# 다양한 유형의 사업명이 잘 분석되는지 보기 위해
# 일반 사업명, 복지, 문화, 농업, 괄호 포함, 글자 간 공백 의심 사례를 섞어 확인합니다.

sample_names = [
    '마늘재배 영농지원단 지원',
    '장애인직업재활 자활교육',
    '문화학교 운영',
    '수산물 상생할인 지원사업',
    '농촌신활력 플러스사업 (추진단사무국상근직원인건비)',
    '자연환경 보전활동 사업 (자연보호정화활동및군수기경진대회)',
    '2023년 음식점 위생환경 개선사업 (시설환경개선및입식테이블지원)',
    '서 부 학 대 피 해 노 인 전 용 쉼 터 지원',
    '장 애 인 일 자 리 종 합 지 원 센 터 운영'
]


# ------------------------------------------------------------
# 4) 형태소 분석 결과를 표로 정리
# ------------------------------------------------------------

kiwi_test_rows = []

for name in sample_names:
    analyzed = kiwi.tokenize(name)

    for token in analyzed:
        kiwi_test_rows.append({
            '사업명_clean': name,
            '형태소': token.form,
            '품사': token.tag,
            '시작위치': token.start,
            '길이': token.len
        })

kiwi_test_result = pd.DataFrame(kiwi_test_rows)

print("\n[1] 형태소 분석 테스트 결과")
display(kiwi_test_result)


# ------------------------------------------------------------
# 5) 명사 계열만 따로 보기
# ------------------------------------------------------------
# NNG: 일반명사
# NNP: 고유명사
# SL : 외국어/영문
# SN : 숫자
# 필요 여부는 다음 단계에서 조정합니다.

noun_like_tags = ['NNG', 'NNP', 'SL', 'SN']

kiwi_test_nouns = kiwi_test_result[
    kiwi_test_result['품사'].isin(noun_like_tags)
].copy()

print("\n[2] 명사 계열 형태소만 확인")
display(kiwi_test_nouns)


# ------------------------------------------------------------
# 6) 사업명별 명사 계열 토큰 묶어서 확인
# ------------------------------------------------------------

sample_token_summary = (
    kiwi_test_nouns
    .groupby('사업명_clean')['형태소']
    .apply(list)
    .reset_index()
    .rename(columns={'형태소': '명사계열_토큰'})
)

print("\n[3] 사업명별 명사 계열 토큰 요약")
display(sample_token_summary)


print("\n03-7 완료")
print("다음 단계에서는 전체 df_text에 tokens_raw 열을 생성합니다.")
print("단, 샘플 분석 결과에서 복합어 분해가 심하거나 글자 간 공백 오류가 문제되면 먼저 확인합니다.")
```

03-8. 글자 간 공백 명확 오류 보정 열 생성
1. 이 단계의 목적

이 단계는 03-5에서 찾은 글자 간 공백 의심 사례 중 명확한 오류만 보정하여, 토큰화에 사용할 새 열 사업명_clean_spacefix를 만드는 단계입니다.

중요한 점은 다음입니다.

원본 사업명은 수정하지 않음
기존 사업명_clean도 덮어쓰지 않음
새 열 사업명_clean_spacefix를 만듦
정상 표현으로 보이는 내 집 앞, 군 및 도, 춤 그 후 같은 사례는 보정하지 않음
보정 여부를 spacefix_applied 열로 남김

```python
# 03-8. 글자 간 공백 명확 오류 보정 열 생성

# ------------------------------------------------------------
# 1) 원본 보존 원칙
# ------------------------------------------------------------
# df_text['사업명_clean']은 덮어쓰지 않습니다.
# 글자 간 공백이 명확히 잘못된 사례만 새 열에서 보정합니다.

df_text['사업명_clean_spacefix'] = df_text['사업명_clean'].copy()

# ------------------------------------------------------------
# 2) 명확한 글자 간 공백 오류 보정 사전
# ------------------------------------------------------------
# key   : 원본_index
# value : 보정 후 사업명
#
# 주의:
# 아래는 03-5 출력에서 명확히 오류로 보이는 사례만 우선 반영한 초안입니다.
# 애매한 사례는 넣지 않습니다.
# 예: '내 집 앞', '군 및 도', '춤 그 후'는 정상 표현 가능성이 있으므로 제외합니다.

spacefix_map = {
    71055: '서부 학대 피해 노인 전용 쉼터 지원',
    71080: '장애인 일자리 종합지원센터 운영',
    71158: '한국수산업경영인연합회 전국대회 지원',
    71382: '전라남도 자율방범연합회 운영비 지원',
    71195: '민주평화통일자문회의 전남지역회의 운영',
    71391: '전남 농업마이스터대학 교육과정장 지원',
    71351: '남도 음식문화 큰잔치 종가음식전시 지원',
    70996: '상이군경복지회관 기능보강사업',
    71311: '문화혁신대회 장애인청소년예술제',
    71231: '소비자 권익 보호사업(공모사업)',
    71300: '유림지도자 교양강좌(전남유교대학)',
    71407: '임업후계자 전국대회참가·교육지원',
    71383: '전라남도 지역특화작목발전협회 지원',
    52824: '재향경우회 보이스피싱 예방 및 홍보활동',
    53062: '찾아가는 에너지의 날 행사 보조'
}

# ------------------------------------------------------------
# 3) 보정 적용
# ------------------------------------------------------------

# 보정 여부 초기값
df_text['spacefix_applied'] = False

# 원본 index 기준으로 보정 적용
for idx, fixed_name in spacefix_map.items():
    if idx in df_text.index:
        df_text.loc[idx, '사업명_clean_spacefix'] = fixed_name
        df_text.loc[idx, 'spacefix_applied'] = True
    else:
        print(f"주의: index {idx}가 df_text에 없습니다.")

# ------------------------------------------------------------
# 4) 보정 결과 확인표 생성
# ------------------------------------------------------------

spacefix_review_cols = [
    '시도', '지자체명',
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'char_space_run_text',
    'spacefix_applied',
    '보조사업자',
    '평가결과',
    '평가종류'
]

existing_spacefix_review_cols = [
    col for col in spacefix_review_cols
    if col in df_text.columns
]

spacefix_review = (
    df_text.loc[df_text['char_space_issue_flag'], existing_spacefix_review_cols]
    .copy()
)

print("03-8. 글자 간 공백 명확 오류 보정 열 생성 완료")
print("- 글자 간 공백 의심 행 수:", int(df_text['char_space_issue_flag'].sum()))
print("- 보정 적용 행 수:", int(df_text['spacefix_applied'].sum()))
print("- 미보정 의심 행 수:", int(df_text['char_space_issue_flag'].sum() - df_text['spacefix_applied'].sum()))

print("\n[1] 글자 간 공백 의심 사례 보정 전후 확인")
display(spacefix_review)

# ------------------------------------------------------------
# 5) 보정 후에도 1글자 연속 구간이 남아 있는지 재진단
# ------------------------------------------------------------

df_text['char_space_runs_after_fix'] = df_text['사업명_clean_spacefix'].apply(find_single_hangul_runs)
df_text['char_space_run_count_after_fix'] = df_text['char_space_runs_after_fix'].apply(len)
df_text['char_space_issue_after_fix'] = df_text['char_space_run_count_after_fix'] > 0

after_fix_issue_rows = df_text.loc[df_text['char_space_issue_after_fix']].copy()

print("\n[2] 보정 후 글자 간 공백 의심 행 수")
print("- 보정 전 의심 행 수:", int(df_text['char_space_issue_flag'].sum()))
print("- 보정 후 의심 행 수:", int(df_text['char_space_issue_after_fix'].sum()))

print("\n[3] 보정 후에도 남은 의심 사례")
display(
    after_fix_issue_rows[
        [col for col in [
            '시도', '지자체명',
            '사업명_clean_spacefix',
            'char_space_runs_after_fix',
            '보조사업자',
            '평가결과'
        ] if col in after_fix_issue_rows.columns]
    ].head(30)
)

print("\n03-8 완료")
print("다음 단계에서는 사업명_clean_spacefix를 기준으로 전체 tokens_raw를 생성합니다.")
```

03-8B. 남은 글자 간 공백 오류 추가 보정
1. 이 단계의 목적

03-8 이후에도 남아 있는 글자 간 공백 의심 사례 12개 중, 실제 오류로 보이는 9개만 추가 보정합니다.

원본 사업명, 기존 사업명_clean은 그대로 두고, 사업명_clean_spacefix만 수정합니다.

```python
# 03-8B. 남은 글자 간 공백 오류 추가 보정

# ------------------------------------------------------------
# 1) 보정 메모 열이 없으면 생성
# ------------------------------------------------------------

if 'spacefix_note' not in df_text.columns:
    df_text['spacefix_note'] = ''

# ------------------------------------------------------------
# 2) 추가 보정 사전
# ------------------------------------------------------------
# 원본 index 기준으로 명확한 오류만 보정합니다.
# 정상 표현으로 판단한 26155, 29885, 35489는 제외합니다.

additional_spacefix_map = {
    71048: '노인돌봄 광역지원기관 운영 지원 (노인돌봄 광역지원기관 종사자 처우개선)',
    71156: '수산단체 (수산업경영인, 여성어업인) 국내 선진지 현장견학',
    71294: '유림지도자 교육 및 도덕성 회복사업 (성균관유도회 전라남도본부)',
    71295: '도내 초중고생 효행실천수기집 발간 (성균관유도회 전라남도본부)',
    71296: '전남 유림신문 발간 (성균관유도회 전라남도본부)',
    71297: '전라남도 선비문화포럼 (성균관청년유도회 전라남도본부)',
    71298: '영호남 유교문화 교류 학술회 (성균관청년유도회 전라남도본부)',
    71299: '제11회 대한민국 청년 유림대회 (성균관청년유도회 전라남도본부)',
    71302: '제15회 맹모삼천지교 포럼 (성균관여성유도회 전라남도본부)'
}

# ------------------------------------------------------------
# 3) 추가 보정 적용
# ------------------------------------------------------------

for idx, fixed_name in additional_spacefix_map.items():
    if idx in df_text.index:
        df_text.loc[idx, '사업명_clean_spacefix'] = fixed_name
        df_text.loc[idx, 'spacefix_applied'] = True
        df_text.loc[idx, 'spacefix_note'] = '03-8B 추가 보정: 글자 간 공백 오류'
    else:
        print(f"주의: index {idx}가 df_text에 없습니다.")

# ------------------------------------------------------------
# 4) 추가 보정 전후 확인표
# ------------------------------------------------------------

additional_spacefix_review_cols = [
    '시도', '지자체명',
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'char_space_run_text',
    'spacefix_applied',
    'spacefix_note',
    '보조사업자',
    '평가결과',
    '평가종류'
]

existing_additional_spacefix_review_cols = [
    col for col in additional_spacefix_review_cols
    if col in df_text.columns
]

additional_spacefix_review = (
    df_text.loc[list(additional_spacefix_map.keys()), existing_additional_spacefix_review_cols]
    .copy()
)

print("03-8B. 남은 글자 간 공백 오류 추가 보정 완료")
print("- 추가 보정 적용 행 수:", len(additional_spacefix_map))

print("\n[1] 추가 보정 전후 확인")
display(additional_spacefix_review)

# ------------------------------------------------------------
# 5) 추가 보정 후 글자 간 공백 의심 사례 재진단
# ------------------------------------------------------------

df_text['char_space_runs_after_fix2'] = df_text['사업명_clean_spacefix'].apply(find_single_hangul_runs)
df_text['char_space_run_count_after_fix2'] = df_text['char_space_runs_after_fix2'].apply(len)
df_text['char_space_issue_after_fix2'] = df_text['char_space_run_count_after_fix2'] > 0

after_fix2_issue_rows = df_text.loc[df_text['char_space_issue_after_fix2']].copy()

print("\n[2] 추가 보정 후 글자 간 공백 의심 행 수")
print("- 03-8 이후 남은 의심 행 수:", int(df_text['char_space_issue_after_fix'].sum()))
print("- 03-8B 이후 남은 의심 행 수:", int(df_text['char_space_issue_after_fix2'].sum()))

print("\n[3] 추가 보정 후에도 남은 의심 사례")
display(
    after_fix2_issue_rows[
        [col for col in [
            '시도', '지자체명',
            '사업명_clean_spacefix',
            'char_space_runs_after_fix2',
            '보조사업자',
            '평가결과'
        ] if col in after_fix2_issue_rows.columns]
    ].head(30)
)

print("\n03-8B 완료")
print("남은 의심 사례가 '군 및 도', '춤 그 후', '내 집 앞' 정도라면 정상 표현으로 보고 다음 토큰화 단계로 넘어가면 됩니다.")
```

```python
# 03-8C. 글자 간 공백 보정 최종 확인용 열 생성

# 보정 후 남은 의심 구간을 사람이 보기 쉬운 문자열로 변환
df_text['char_space_run_text_after_fix2'] = df_text['char_space_runs_after_fix2'].apply(
    lambda runs: ' | '.join(runs) if len(runs) > 0 else ''
)

# 최종 확인표
final_spacefix_check = df_text.loc[
    df_text['char_space_issue_after_fix2'],
    [
        '시도',
        '지자체명',
        '사업명',
        '사업명_clean',
        '사업명_clean_spacefix',
        'char_space_run_text',
        'char_space_run_text_after_fix2',
        'spacefix_applied',
        'spacefix_note',
        '보조사업자',
        '평가결과'
    ]
].copy()

print("03-8C. 글자 간 공백 보정 최종 확인")
print("- 보정 후 남은 의심 행 수:", len(final_spacefix_check))

display(final_spacefix_check)
```

03-9. 전체 tokens_raw 생성
1. 이 단계의 목적

이 단계는 사업명_clean_spacefix를 기준으로 전체 사업명에 형태소 분석을 적용하여, 분석용 원시 토큰 열인 tokens_raw를 생성하는 단계입니다.

중요한 점은 다음입니다.

원본 사업명은 수정하지 않습니다.
사업명_clean도 수정하지 않습니다.
글자 간 공백 보정이 반영된 사업명_clean_spacefix를 기준으로 토큰화합니다.
이 단계에서는 아직 불용어를 제거하지 않습니다.
지원, 사업, 운영, 숫자, 지역명 등도 일단 tokens_raw에는 남깁니다.
불용어 제거는 다음 단계에서 빈도표와 실제 사례를 본 뒤 진행합니다.


2. 입력 데이터

이전 단계까지 만든 데이터프레임입니다.

df_text


필수 열은 다음입니다.

사업명_clean_spacefix


그리고 03-7에서 생성한 형태소 분석기 객체가 필요합니다.

kiwi

| 변수명                    | 의미                      |
| ---------------------- | ----------------------- |
| `noun_like_tags`       | 추출할 품사 목록               |
| `extract_tokens_raw()` | 사업명에서 명사 계열 토큰을 추출하는 함수 |
| `tokens_raw`           | 형태소 분석 후 추출한 원시 토큰 리스트  |
| `tokens_raw_count`     | 행별 원시 토큰 개수             |
| `tokens_raw_empty`     | 토큰이 비어 있는지 여부           |
| `token_check_sample`   | 토큰화 결과 확인용 샘플           |


```python
# 03-9. 전체 tokens_raw 생성

# ------------------------------------------------------------
# 1) tqdm 불러오기
# ------------------------------------------------------------
# tqdm은 전체 79,262행 토큰화 진행률을 보기 위한 도구입니다.

from tqdm.auto import tqdm

tqdm.pandas()


# ------------------------------------------------------------
# 2) 형태소 분석기 객체 확인
# ------------------------------------------------------------
# 03-7을 실행했다면 kiwi 객체가 이미 있어야 합니다.
# 만약 런타임이 초기화되어 kiwi가 없다면 다시 생성합니다.

try:
    kiwi
    print("기존 kiwi 객체를 사용합니다.")
except NameError:
    from kiwipiepy import Kiwi
    kiwi = Kiwi()
    print("kiwi 객체가 없어 새로 생성했습니다.")


# ------------------------------------------------------------
# 3) 추출할 품사 기준 설정
# ------------------------------------------------------------
# NNG: 일반명사
# NNP: 고유명사
# SL : 외국어/영문
# SN : 숫자
#
# 주의:
# 숫자(SN)는 일단 tokens_raw에는 남겨둡니다.
# 이후 tokens_filtered 단계에서 연도/숫자 토큰 제거 여부를 판단합니다.

noun_like_tags = ['NNG', 'NNP', 'SL', 'SN']


# ------------------------------------------------------------
# 4) 원시 토큰 추출 함수 정의
# ------------------------------------------------------------

def extract_tokens_raw(text):
    """
    사업명 텍스트에서 명사 계열 토큰을 추출합니다.

    입력:
    - text: 사업명_clean_spacefix

    출력:
    - 명사 계열 토큰 리스트

    주의:
    - 이 함수는 불용어를 제거하지 않습니다.
    - 숫자, 지역명, 행정 일반어도 일단 보존합니다.
    - 불용어 제거는 다음 단계에서 별도로 수행합니다.
    """

    if pd.isna(text):
        return []

    text = str(text).strip()

    if text == '':
        return []

    analyzed = kiwi.tokenize(text)

    tokens = []

    for token in analyzed:
        token_text = token.form.strip()
        token_tag = token.tag

        # 명사 계열 품사만 우선 추출
        if token_tag in noun_like_tags and token_text != '':
            tokens.append(token_text)

    return tokens


# ------------------------------------------------------------
# 5) 전체 df_text에 tokens_raw 생성
# ------------------------------------------------------------
# 기준 열은 사업명_clean_spacefix입니다.
# 글자 간 공백 보정이 반영된 열을 사용해야 합니다.

df_text['tokens_raw'] = df_text['사업명_clean_spacefix'].progress_apply(extract_tokens_raw)

# 행별 토큰 개수
df_text['tokens_raw_count'] = df_text['tokens_raw'].apply(len)

# 토큰이 비어 있는지 여부
df_text['tokens_raw_empty'] = df_text['tokens_raw_count'].eq(0)


# ------------------------------------------------------------
# 6) tokens_raw 생성 결과 요약
# ------------------------------------------------------------

tokens_raw_summary = pd.DataFrame({
    '항목': [
        '텍스트 분석 대상 행 수',
        'tokens_raw 생성 행 수',
        'tokens_raw 빈 행 수',
        'tokens_raw 빈 행 비율(%)',
        '행별 평균 토큰 수',
        '행별 최소 토큰 수',
        '행별 최대 토큰 수'
    ],
    '값': [
        len(df_text),
        df_text['tokens_raw'].notna().sum(),
        int(df_text['tokens_raw_empty'].sum()),
        round(df_text['tokens_raw_empty'].mean() * 100, 3),
        round(df_text['tokens_raw_count'].mean(), 3),
        int(df_text['tokens_raw_count'].min()),
        int(df_text['tokens_raw_count'].max())
    ]
})

print("03-9. 전체 tokens_raw 생성 완료")
display(tokens_raw_summary)


# ------------------------------------------------------------
# 7) 토큰화 결과 샘플 확인
# ------------------------------------------------------------
# 일반 샘플과 글자 간 공백 보정 적용 사례를 함께 확인합니다.

sample_cols = [
    '시도',
    '지자체명',
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'spacefix_applied',
    'tokens_raw',
    'tokens_raw_count',
    '보조사업자',
    '평가결과',
    '평가종류'
]

existing_sample_cols = [col for col in sample_cols if col in df_text.columns]

print("\n[1] 전체 토큰화 결과 샘플 30개")
display(
    df_text[existing_sample_cols]
    .head(30)
)


print("\n[2] 글자 간 공백 보정 적용 행의 토큰화 결과")
display(
    df_text.loc[
        df_text['spacefix_applied'],
        existing_sample_cols
    ].head(30)
)


# ------------------------------------------------------------
# 8) tokens_raw가 비어 있는 행 확인
# ------------------------------------------------------------

empty_token_rows = df_text.loc[df_text['tokens_raw_empty'], existing_sample_cols].copy()

print("\n[3] tokens_raw가 비어 있는 행")
print("- 빈 토큰 행 수:", len(empty_token_rows))
display(empty_token_rows.head(30))


# ------------------------------------------------------------
# 9) 토큰 개수가 많은 행 확인
# ------------------------------------------------------------
# 너무 많은 토큰이 생성된 행은 괄호 안 설명이 길거나 여러 사업명이 붙어 있을 수 있습니다.

print("\n[4] tokens_raw_count 상위 30개")
display(
    df_text
    .sort_values('tokens_raw_count', ascending=False)
    [existing_sample_cols]
    .head(30)
)


print("\n03-9 완료")
print("다음 단계에서는 tokens_raw 전체 빈도표를 만들고, 불용어 후보를 검토합니다.")
```

03-10. tokens_raw 전체 빈도표 생성 및 불용어 후보 1차 확인
1. 이 단계의 목적

이 단계는 tokens_raw를 전체 행 기준으로 펼쳐서 단어 빈도표를 만드는 단계입니다.

아직 불용어를 제거하지 않습니다.
먼저 어떤 단어가 많이 등장하는지 확인하고, 다음 단계에서 제거할 단어와 보존할 단어를 판단합니다.

특히 아래 항목을 확인합니다.

지원, 사업, 운영 같은 행정 일반어
2023, 1, 13 같은 숫자·연도 토큰
경남, 전남, 서울, 남해 같은 지역명
복지, 문화, 체육, 농업, 교육, 환경, 장애인, 노인, 아동 같은 정책분야 단어
1글자 토큰

```python
# 03-10. tokens_raw 전체 빈도표 생성 및 불용어 후보 1차 확인

# ------------------------------------------------------------
# 1) tokens_raw를 긴 형태로 펼치기
# ------------------------------------------------------------
# tokens_raw는 각 행마다 리스트 형태입니다.
# 예: ['마늘', '재배', '영농', '지원단', '지원']
#
# explode()를 사용하면 토큰 하나가 한 행이 되는 형태로 바뀝니다.

token_raw_long = (
    df_text[['tokens_raw']]
    .explode('tokens_raw')
    .rename(columns={'tokens_raw': 'token'})
    .dropna()
    .copy()
)

# 혹시 빈 문자열 토큰이 있으면 제거합니다.
token_raw_long['token'] = token_raw_long['token'].astype(str).str.strip()
token_raw_long = token_raw_long[token_raw_long['token'] != '']

print("03-10. tokens_raw 긴 형태 변환 완료")
print("- 원본 df_text 행 수:", len(df_text))
print("- 전체 토큰 수:", len(token_raw_long))
print("- 고유 토큰 수:", token_raw_long['token'].nunique())


# ------------------------------------------------------------
# 2) 전체 원시 토큰 빈도표 생성
# ------------------------------------------------------------

token_freq_raw = (
    token_raw_long['token']
    .value_counts()
    .reset_index()
)

token_freq_raw.columns = ['token', 'freq']

token_freq_raw['ratio_percent'] = (
    token_freq_raw['freq'] / token_freq_raw['freq'].sum() * 100
).round(4)

print("\n[1] tokens_raw 전체 빈도 상위 100개")
display(token_freq_raw.head(100))


# ------------------------------------------------------------
# 3) 1글자 토큰 빈도 확인
# ------------------------------------------------------------
# 1글자 토큰은 의미 있는 경우도 있지만, 형태소 분석 오류나 지역명 분해일 수도 있습니다.
# 예: 동, 구, 군, 회, 장, 비 등

one_char_token_freq = token_freq_raw[
    token_freq_raw['token'].str.len() == 1
].copy()

print("\n[2] 1글자 토큰 빈도 상위 50개")
display(one_char_token_freq.head(50))


# ------------------------------------------------------------
# 4) 숫자 토큰 빈도 확인
# ------------------------------------------------------------
# 연도, 회차, 숫자 등은 정책분야 분석에서는 대체로 제거 후보입니다.
# 다만 제1회, 3·1운동처럼 의미가 있을 수 있으므로 일단 후보로만 봅니다.

number_token_freq = token_freq_raw[
    token_freq_raw['token'].str.fullmatch(r'\d+')
].copy()

print("\n[3] 숫자 토큰 빈도 상위 50개")
display(number_token_freq.head(50))


# ------------------------------------------------------------
# 5) 자주 등장할 가능성이 높은 행정 일반어 후보 확인
# ------------------------------------------------------------
# 이 목록은 확정 불용어가 아닙니다.
# 빈도와 실제 사례를 보고 다음 단계에서 제거 여부를 판단합니다.

admin_general_terms = [
    '지원', '사업', '운영', '추진', '활동', '보조', '관리',
    '개선', '조성', '육성', '활성', '활성화', '행사', '대회',
    '교육', '프로그램', '센터', '시설', '운영비', '인건비',
    '지원사업', '보조금', '공모', '참가', '개최'
]

admin_general_check = token_freq_raw[
    token_freq_raw['token'].isin(admin_general_terms)
].copy()

print("\n[4] 행정 일반어 후보 빈도")
display(admin_general_check)


# ------------------------------------------------------------
# 6) 정책분야 핵심어 후보 확인
# ------------------------------------------------------------
# 이 단어들은 불용어로 제거하면 안 될 가능성이 높습니다.
# 04번 정책분야 분류의 중요한 단서가 될 수 있습니다.

policy_keyword_candidates = [
    '복지', '보건', '의료', '건강',
    '문화', '예술', '체육', '관광',
    '농업', '농촌', '수산', '어업', '축산', '임업',
    '교육', '청소년', '아동', '어린이', '청년',
    '노인', '장애인', '여성', '다문화',
    '환경', '안전', '방범', '재난',
    '일자리', '고용', '창업', '기업', '산업',
    '마을', '공동체', '도시', '주거', '교통'
]

policy_keyword_check = token_freq_raw[
    token_freq_raw['token'].isin(policy_keyword_candidates)
].copy()

print("\n[5] 정책분야 핵심어 후보 빈도")
display(policy_keyword_check)


# ------------------------------------------------------------
# 7) 지역명 후보 1차 확인
# ------------------------------------------------------------
# 전국 데이터에서는 지역명이 상위에 많이 나올 수 있습니다.
# 지역명은 정책분야 단어가 아니므로 제거 후보일 수 있지만,
# 지역 기반 사업명 해석에 필요할 수도 있어 보존/제거 버전 비교가 필요합니다.

region_keyword_candidates = [
    # 광역
    '서울', '부산', '대구', '인천', '광주', '대전', '울산', '세종',
    '경기', '강원', '충북', '충남', '전북', '전남', '경북', '경남', '제주',
    '서울시', '부산시', '대구시', '인천시', '광주시', '대전시', '울산시',
    '경기도', '강원도', '충청북도', '충청남도', '전라북도', '전라남도',
    '경상북도', '경상남도', '제주도',

    # 일부 샘플에서 이미 보인 지역명
    '남해', '진주', '창원', '김해', '전주', '청주', '부여',
    '안양', '계양', '영주', '속초', '밀양', '양산', '통영',
    '사천', '창녕', '하남', '제천', '연수'
]

region_keyword_check = token_freq_raw[
    token_freq_raw['token'].isin(region_keyword_candidates)
].copy()

print("\n[6] 지역명 후보 빈도")
display(region_keyword_check)


# ------------------------------------------------------------
# 8) 불용어 후보 검토용 표 생성
# ------------------------------------------------------------
# 상위 빈도 토큰을 중심으로, 사람이 제거 여부를 판단할 수 있는 표를 만듭니다.
# 지금은 자동으로 제거하지 않습니다.

candidate_stopword_check = token_freq_raw.head(300).copy()

# 기본 분류 태그를 붙입니다.
# 이 태그는 검토 편의를 위한 초안이며, 확정 판단이 아닙니다.

def classify_candidate_token(token):
    if re.fullmatch(r'\d+', token):
        return '숫자/연도 후보'
    if token in admin_general_terms:
        return '행정 일반어 후보'
    if token in region_keyword_candidates:
        return '지역명 후보'
    if token in policy_keyword_candidates:
        return '정책분야 핵심어 후보'
    if len(token) == 1:
        return '1글자 토큰 후보'
    return '검토필요'

candidate_stopword_check['검토분류_초안'] = candidate_stopword_check['token'].apply(classify_candidate_token)

# 사람이 직접 채울 열
candidate_stopword_check['제거여부'] = ''
candidate_stopword_check['검토메모'] = ''

print("\n[7] 불용어 후보 검토표 상위 100개")
display(candidate_stopword_check.head(100))


# ------------------------------------------------------------
# 9) 결과 저장
# ------------------------------------------------------------

timestamp = datetime.now().strftime('%Y%m%d_%H%M')

token_freq_raw_csv = TABLE_DIR / f'03_token_freq_raw_v1_{timestamp}.csv'
candidate_stopword_xlsx = TABLE_DIR / f'03_candidate_stopword_check_v1_{timestamp}.xlsx'
candidate_stopword_csv = TABLE_DIR / f'03_candidate_stopword_check_v1_{timestamp}.csv'

token_freq_raw.to_csv(token_freq_raw_csv, index=False, encoding='utf-8-sig')
candidate_stopword_check.to_excel(candidate_stopword_xlsx, index=False)
candidate_stopword_check.to_csv(candidate_stopword_csv, index=False, encoding='utf-8-sig')

print("\n03-10 저장 완료")
print("- 원시 토큰 빈도표 CSV:", token_freq_raw_csv)
print("- 불용어 후보 검토표 Excel:", candidate_stopword_xlsx)
print("- 불용어 후보 검토표 CSV:", candidate_stopword_csv)

print("\n저장 파일 존재 여부")
print("- token_freq_raw_csv:", token_freq_raw_csv.exists())
print("- candidate_stopword_xlsx:", candidate_stopword_xlsx.exists())
print("- candidate_stopword_csv:", candidate_stopword_csv.exists())


print("\n03-10 완료")
print("다음 단계에서는 상위 빈도 단어와 후보 검토표를 보고 1차 불용어 목록을 만들고, tokens_filtered를 생성합니다.")
```

03-11. 1차 불용어 적용 및 tokens_filtered 생성
1. 이 단계의 목적

이 단계는 tokens_raw에서 분석에 방해가 될 가능성이 큰 토큰을 1차로 제거하여, 두 가지 필터링 토큰 열을 만드는 단계입니다.

이번에는 두 버전을 동시에 만듭니다.

| 열 이름                          | 의미                                 |
| ----------------------------- | ---------------------------------- |
| `tokens_filtered_keep_region` | 숫자·행정 일반어·일부 1글자 잡음은 제거하되, 지역명은 보존 |
| `tokens_filtered_no_region`   | 위 기준에 더해 지역명 후보까지 제거               |


지역명을 바로 완전히 제거하지 않고 두 버전을 만드는 이유는, 이후 빈도 분석과 06A 네트워크 분석에서 지역명이 결과를 얼마나 좌우하는지 비교하기 위해서입니다.

```python
# 03-11. 1차 불용어 적용 및 tokens_filtered 생성

# ------------------------------------------------------------
# 1) 1차 기본 불용어 목록 설정
# ------------------------------------------------------------
# 주의:
# 이 목록은 확정 불용어가 아니라 1차 분석용 초안입니다.
# 정책분야를 구분하는 단어는 최대한 보존합니다.
#
# 예를 들어 '교육', '복지', '문화', '체육', '농업', '환경',
# '장애인', '노인', '아동', '청소년' 등은 제거하지 않습니다.

basic_stopwords = {
    # 매우 일반적인 행정·보조사업 표현
    '지원', '사업', '운영', '보조', '추진',
    '지원사업', '보조금',

    # 비용·운영 관련 일반 표현
    '운영비', '인건비',

    # 행정 처리·절차 성격이 강한 표현
    '관리', '개최', '참가'
}

# ------------------------------------------------------------
# 2) 1글자 잡음 후보 설정
# ------------------------------------------------------------
# 모든 1글자 토큰을 제거하지는 않습니다.
# '쌀', '벼', '집', '길'처럼 의미가 있는 1글자 단어도 있기 때문입니다.
# 여기서는 빈도표에서 잡음 가능성이 높은 1글자만 보수적으로 제거합니다.

one_char_stopwords = {
    '회', '기', '도', '군', '동', '시', '리', '지', '인',
    '원', '장', '면', '대', '구', '단', '종', '전', '사',
    '관', '정', '상', '급', '료', '중'
}

# 보존할 정책분야 핵심어
# 혹시 basic_stopwords나 one_char_stopwords에 들어가더라도 이 목록은 최종 제거하지 않도록 보호합니다.
policy_preserve_terms = {
    '장애인', '체육', '문화', '교육', '마을', '환경', '복지',
    '노인', '아동', '청소년', '여성', '농업', '예술', '안전',
    '청년', '농촌', '공동체', '건강', '산업', '기업', '다문화',
    '관광', '도시', '어린이', '축산', '창업', '교통', '일자리',
    '방범', '의료', '재난', '주거', '임업', '어업', '수산', '보건',
    '고용'
}

# 보호 목록에 있는 단어는 불용어에서 제거
basic_stopwords = basic_stopwords - policy_preserve_terms
one_char_stopwords = one_char_stopwords - policy_preserve_terms


# ------------------------------------------------------------
# 3) 지역명 후보 목록 설정
# ------------------------------------------------------------
# 03-10에서 region_keyword_candidates 변수가 있으면 그대로 사용합니다.
# 없을 경우를 대비해 기본 목록을 다시 정의합니다.

try:
    region_keyword_candidates
except NameError:
    region_keyword_candidates = [
        '서울', '부산', '대구', '인천', '광주', '대전', '울산', '세종',
        '경기', '강원', '충북', '충남', '전북', '전남', '경북', '경남', '제주',
        '서울시', '부산시', '대구시', '인천시', '광주시', '대전시', '울산시',
        '경기도', '강원도', '충청북도', '충청남도', '전라북도', '전라남도',
        '경상북도', '경상남도', '제주도',
        '남해', '진주', '창원', '김해', '전주', '청주', '부여',
        '안양', '계양', '영주', '속초', '밀양', '양산', '통영',
        '사천', '창녕', '하남', '제천', '연수'
    ]

region_stopwords = set(region_keyword_candidates)

# 정책분야 핵심어와 겹치면 제거하지 않도록 보호
region_stopwords = region_stopwords - policy_preserve_terms


# ------------------------------------------------------------
# 4) 토큰 필터링 함수 정의
# ------------------------------------------------------------

def is_number_token(token):
    """숫자 토큰인지 확인합니다."""
    return bool(re.fullmatch(r'\d+', str(token)))


def filter_tokens(tokens, remove_region=False):
    """
    tokens_raw에서 1차 불용어를 제거합니다.

    remove_region=False:
    - 지역명은 보존합니다.

    remove_region=True:
    - 지역명 후보까지 제거합니다.
    """

    if not isinstance(tokens, list):
        return []

    filtered = []

    for token in tokens:
        token = str(token).strip()

        if token == '':
            continue

        # 숫자 토큰 제거
        if is_number_token(token):
            continue

        # 기본 불용어 제거
        if token in basic_stopwords:
            continue

        # 1글자 잡음 후보 제거
        if token in one_char_stopwords:
            continue

        # 지역명 제거 버전일 때만 지역명 제거
        if remove_region and token in region_stopwords:
            continue

        filtered.append(token)

    return filtered


# ------------------------------------------------------------
# 5) 두 가지 tokens_filtered 열 생성
# ------------------------------------------------------------

df_text['tokens_filtered_keep_region'] = df_text['tokens_raw'].apply(
    lambda tokens: filter_tokens(tokens, remove_region=False)
)

df_text['tokens_filtered_no_region'] = df_text['tokens_raw'].apply(
    lambda tokens: filter_tokens(tokens, remove_region=True)
)

df_text['tokens_filtered_keep_region_count'] = df_text['tokens_filtered_keep_region'].apply(len)
df_text['tokens_filtered_no_region_count'] = df_text['tokens_filtered_no_region'].apply(len)

df_text['tokens_filtered_keep_region_empty'] = df_text['tokens_filtered_keep_region_count'].eq(0)
df_text['tokens_filtered_no_region_empty'] = df_text['tokens_filtered_no_region_count'].eq(0)


# ------------------------------------------------------------
# 6) 필터링 결과 요약
# ------------------------------------------------------------

filter_summary = pd.DataFrame({
    '항목': [
        '분석 대상 행 수',
        'tokens_raw 전체 토큰 수',
        '지역명 보존 버전 전체 토큰 수',
        '지역명 제거 버전 전체 토큰 수',
        '지역명 보존 버전 빈 행 수',
        '지역명 제거 버전 빈 행 수',
        '지역명 보존 버전 평균 토큰 수',
        '지역명 제거 버전 평균 토큰 수'
    ],
    '값': [
        len(df_text),
        int(df_text['tokens_raw_count'].sum()),
        int(df_text['tokens_filtered_keep_region_count'].sum()),
        int(df_text['tokens_filtered_no_region_count'].sum()),
        int(df_text['tokens_filtered_keep_region_empty'].sum()),
        int(df_text['tokens_filtered_no_region_empty'].sum()),
        round(df_text['tokens_filtered_keep_region_count'].mean(), 3),
        round(df_text['tokens_filtered_no_region_count'].mean(), 3)
    ]
})

print("03-11. 1차 불용어 적용 및 tokens_filtered 생성 완료")
display(filter_summary)


# ------------------------------------------------------------
# 7) 필터링 전후 샘플 확인
# ------------------------------------------------------------

sample_cols = [
    '시도', '지자체명',
    '사업명',
    '사업명_clean_spacefix',
    'tokens_raw',
    'tokens_filtered_keep_region',
    'tokens_filtered_no_region',
    'tokens_raw_count',
    'tokens_filtered_keep_region_count',
    'tokens_filtered_no_region_count',
    '평가결과',
    '평가종류'
]

existing_sample_cols = [col for col in sample_cols if col in df_text.columns]

print("\n[1] 필터링 전후 샘플 30개")
display(df_text[existing_sample_cols].head(30))


# ------------------------------------------------------------
# 8) 필터링 후 빈 토큰 행 확인
# ------------------------------------------------------------

print("\n[2] 지역명 보존 버전에서 빈 토큰이 된 행")
display(
    df_text.loc[
        df_text['tokens_filtered_keep_region_empty'],
        existing_sample_cols
    ].head(30)
)

print("\n[3] 지역명 제거 버전에서 빈 토큰이 된 행")
display(
    df_text.loc[
        df_text['tokens_filtered_no_region_empty'],
        existing_sample_cols
    ].head(30)
)


# ------------------------------------------------------------
# 9) 필터링 버전별 빈도표 생성 함수
# ------------------------------------------------------------

def make_token_freq(df, token_col):
    """리스트 형태 토큰 열을 펼쳐 빈도표를 생성합니다."""

    token_long = (
        df[[token_col]]
        .explode(token_col)
        .rename(columns={token_col: 'token'})
        .dropna()
        .copy()
    )

    token_long['token'] = token_long['token'].astype(str).str.strip()
    token_long = token_long[token_long['token'] != '']

    token_freq = (
        token_long['token']
        .value_counts()
        .reset_index()
    )

    token_freq.columns = ['token', 'freq']
    token_freq['ratio_percent'] = (
        token_freq['freq'] / token_freq['freq'].sum() * 100
    ).round(4)

    return token_freq


token_freq_keep_region = make_token_freq(df_text, 'tokens_filtered_keep_region')
token_freq_no_region = make_token_freq(df_text, 'tokens_filtered_no_region')


print("\n[4] 지역명 보존 버전 빈도 상위 100개")
display(token_freq_keep_region.head(100))

print("\n[5] 지역명 제거 버전 빈도 상위 100개")
display(token_freq_no_region.head(100))


# ------------------------------------------------------------
# 10) 지역명 제거 전후 차이 확인
# ------------------------------------------------------------

region_removed_check = token_freq_keep_region[
    token_freq_keep_region['token'].isin(region_stopwords)
].copy()

print("\n[6] 지역명 보존 버전에 남아 있는 지역명 후보")
display(region_removed_check.head(100))


# ------------------------------------------------------------
# 11) 결과 저장
# ------------------------------------------------------------

timestamp = datetime.now().strftime('%Y%m%d_%H%M')

token_freq_keep_region_csv = TABLE_DIR / f'03_token_freq_filtered_keep_region_v1_{timestamp}.csv'
token_freq_no_region_csv = TABLE_DIR / f'03_token_freq_filtered_no_region_v1_{timestamp}.csv'

filter_summary_csv = TABLE_DIR / f'03_token_filter_summary_v1_{timestamp}.csv'

token_freq_keep_region.to_csv(token_freq_keep_region_csv, index=False, encoding='utf-8-sig')
token_freq_no_region.to_csv(token_freq_no_region_csv, index=False, encoding='utf-8-sig')
filter_summary.to_csv(filter_summary_csv, index=False, encoding='utf-8-sig')

print("\n03-11 저장 완료")
print("- 지역명 보존 빈도표:", token_freq_keep_region_csv)
print("- 지역명 제거 빈도표:", token_freq_no_region_csv)
print("- 필터링 요약표:", filter_summary_csv)

print("\n저장 파일 존재 여부")
print("- token_freq_keep_region_csv:", token_freq_keep_region_csv.exists())
print("- token_freq_no_region_csv:", token_freq_no_region_csv.exists())
print("- filter_summary_csv:", filter_summary_csv.exists())

print("\n03-11 완료")
print("다음 단계에서는 두 빈도표를 비교하고, 불용어 목록을 조정할지 판단합니다.")
```

03-12. 필터링 결과 비교 및 불용어 조정 후보 점검
1. 이 단계의 목적

이 단계는 03-11에서 만든 두 가지 토큰 버전을 비교하는 단계입니다.

비교 대상은 다음입니다.

열	의미
tokens_filtered_keep_region	지역명 보존 버전
tokens_filtered_no_region	지역명 제거 버전

이번 단계에서는 아직 불용어 목록을 수정하지 않습니다.
대신 다음을 확인합니다.

지역명 제거 전후 상위 단어 차이
아직 많이 남아 있는 행정 일반어 후보
정책분야 핵심어가 잘 보존됐는지
빈 토큰 행이 어떤 유형인지
추가 불용어 후보를 만들 필요가 있는지

```python
# 03-12. 필터링 결과 비교 및 불용어 조정 후보 점검

# ------------------------------------------------------------
# 1) 지역명 보존/제거 버전 빈도 비교표 생성
# ------------------------------------------------------------

freq_compare = token_freq_keep_region.rename(
    columns={
        'freq': 'freq_keep_region',
        'ratio_percent': 'ratio_keep_region'
    }
).merge(
    token_freq_no_region.rename(
        columns={
            'freq': 'freq_no_region',
            'ratio_percent': 'ratio_no_region'
        }
    ),
    on='token',
    how='outer'
)

# 결측은 0으로 채웁니다.
freq_compare[['freq_keep_region', 'ratio_keep_region', 'freq_no_region', 'ratio_no_region']] = (
    freq_compare[['freq_keep_region', 'ratio_keep_region', 'freq_no_region', 'ratio_no_region']]
    .fillna(0)
)

# 빈도 차이 계산
freq_compare['freq_diff_keep_minus_no'] = (
    freq_compare['freq_keep_region'] - freq_compare['freq_no_region']
)

freq_compare['ratio_diff_keep_minus_no'] = (
    freq_compare['ratio_keep_region'] - freq_compare['ratio_no_region']
).round(4)

# 보기 편하게 정렬
freq_compare = freq_compare.sort_values(
    ['freq_keep_region', 'freq_no_region'],
    ascending=False
).reset_index(drop=True)

print("03-12. 필터링 결과 비교")
print("\n[1] 지역명 보존/제거 버전 빈도 비교 상위 100개")
display(freq_compare.head(100))


# ------------------------------------------------------------
# 2) 지역명 제거로 실제 사라진 토큰 확인
# ------------------------------------------------------------
# keep_region에는 있는데 no_region에서는 0인 토큰을 확인합니다.
# 대부분 지역명 후보여야 정상입니다.

removed_by_region_filter = freq_compare[
    (freq_compare['freq_keep_region'] > 0) &
    (freq_compare['freq_no_region'] == 0)
].copy()

print("\n[2] 지역명 제거 버전에서 사라진 토큰 상위 100개")
display(removed_by_region_filter.head(100))


# ------------------------------------------------------------
# 3) 아직 남아 있는 행정 일반어 후보 확인
# ------------------------------------------------------------
# 03-11에서 일부러 보존한 단어들 중,
# 너무 상위에 남아 있는 단어를 검토합니다.
#
# 주의:
# 여기서 나온다고 바로 제거하지 않습니다.
# '교육', '시설', '센터', '대회', '행사'는 정책분야/사업유형 해석에 필요할 수 있습니다.

remaining_admin_terms = [
    '대회', '행사', '시설', '센터', '활동',
    '개선', '조성', '육성', '프로그램',
    '공모', '개발', '교류', '기념', '홍보',
    '보급', '구입', '제작', '정비', '설치',
    '상담', '서비스', '강화', '증진', '확대',
    '참여', '추가'
]

remaining_admin_check = token_freq_no_region[
    token_freq_no_region['token'].isin(remaining_admin_terms)
].copy()

print("\n[3] 지역명 제거 버전에 아직 남아 있는 행정/사업유형 후보")
display(remaining_admin_check)


# ------------------------------------------------------------
# 4) 정책분야 핵심어 보존 여부 확인
# ------------------------------------------------------------
# 정책분야 분류에 중요한 단어들이 필터링 후에도 살아 있는지 확인합니다.

policy_core_terms = [
    '장애인', '체육', '문화', '교육', '마을', '환경', '복지',
    '노인', '아동', '청소년', '여성', '농업', '예술', '안전',
    '청년', '농촌', '공동체', '건강', '산업', '기업', '다문화',
    '관광', '도시', '어린이', '축산', '창업', '교통', '일자리',
    '방범', '의료', '재난', '주거', '임업', '어업', '수산', '보건',
    '고용'
]

preserved_policy_check = token_freq_no_region[
    token_freq_no_region['token'].isin(policy_core_terms)
].copy()

print("\n[4] 정책분야 핵심어 보존 확인")
display(preserved_policy_check)


# ------------------------------------------------------------
# 5) 필터링 후 빈 토큰 행 검토
# ------------------------------------------------------------
# 빈 토큰 행이 많지 않은지, 어떤 유형인지 확인합니다.

empty_token_review_cols = [
    '시도', '지자체명',
    '사업명',
    '사업명_clean_spacefix',
    'tokens_raw',
    'tokens_filtered_keep_region',
    'tokens_filtered_no_region',
    'tokens_raw_count',
    'tokens_filtered_keep_region_count',
    'tokens_filtered_no_region_count',
    '평가결과',
    '평가종류'
]

existing_empty_token_review_cols = [
    col for col in empty_token_review_cols
    if col in df_text.columns
]

empty_token_review = df_text.loc[
    df_text['tokens_filtered_no_region_empty'],
    existing_empty_token_review_cols
].copy()

print("\n[5] 지역명 제거 버전 빈 토큰 행 검토")
print("- 지역명 제거 버전 빈 토큰 행 수:", len(empty_token_review))
display(empty_token_review.head(100))


# ------------------------------------------------------------
# 6) 추가 불용어 후보표 생성
# ------------------------------------------------------------
# 상위 빈도 중 아직 검토가 필요한 단어를 모아둡니다.
# 여기서는 후보만 만들고, 실제 제거는 다음 셀에서 결정합니다.

additional_stopword_candidates = token_freq_no_region.head(200).copy()

def classify_additional_stopword_candidate(token):
    if token in policy_core_terms:
        return '정책분야 핵심어_보존권장'
    if token in remaining_admin_terms:
        return '행정/사업유형 후보_검토필요'
    if len(str(token)) == 1:
        return '1글자 토큰_검토필요'
    return '검토필요'

additional_stopword_candidates['검토분류_초안'] = additional_stopword_candidates['token'].apply(
    classify_additional_stopword_candidate
)

additional_stopword_candidates['추가제거여부'] = ''
additional_stopword_candidates['검토메모'] = ''

print("\n[6] 추가 불용어 후보 검토표 상위 100개")
display(additional_stopword_candidates.head(100))


# ------------------------------------------------------------
# 7) 결과 저장
# ------------------------------------------------------------

timestamp = datetime.now().strftime('%Y%m%d_%H%M')

filter_compare_xlsx = TABLE_DIR / f'03_filter_compare_v1_{timestamp}.xlsx'

with pd.ExcelWriter(filter_compare_xlsx, engine='openpyxl') as writer:
    freq_compare.to_excel(writer, sheet_name='freq_compare', index=False)
    removed_by_region_filter.to_excel(writer, sheet_name='removed_by_region', index=False)
    remaining_admin_check.to_excel(writer, sheet_name='remaining_admin', index=False)
    preserved_policy_check.to_excel(writer, sheet_name='policy_core_check', index=False)
    empty_token_review.to_excel(writer, sheet_name='empty_token_review', index=False)
    additional_stopword_candidates.to_excel(writer, sheet_name='additional_candidates', index=False)

print("\n03-12 저장 완료")
print("- 필터링 비교 Excel:", filter_compare_xlsx)
print("- 저장 파일 존재 여부:", filter_compare_xlsx.exists())

print("\n03-12 완료")
print("다음 단계에서는 추가 불용어를 실제로 적용할지 판단합니다.")
```

03-13. 애매 지역명 후보 보정 및 보수적 지역명 제거 버전 생성
1. 이 단계의 목적

이 단계는 03-11에서 만든 tokens_filtered_no_region을 바로 확정하지 않고, 지역명으로 보기 애매한 토큰을 다시 보존하는 보수적 지역명 제거 버전을 만드는 단계입니다.

특히 다음 단어는 자동 지역명 제거에서 제외합니다.

토큰	이유
대전	대전광역시일 수도 있지만, 난 대전, 미술대전처럼 전시·공모 의미일 수 있음
연수	인천 연수구일 수도 있지만, 교육·훈련 의미의 연수일 수 있음
경기	경기도일 수도 있지만, 스포츠 경기·경제 경기 의미일 수 있음

원본 열과 기존 필터링 열은 덮어쓰지 않고, 새 열을 만듭니다.

```python
# 03-13. 애매 지역명 후보 보정 및 보수적 지역명 제거 버전 생성

# ------------------------------------------------------------
# 1) 애매 지역명 후보 지정
# ------------------------------------------------------------
# 아래 토큰은 지역명일 수도 있지만, 일반명사/사업명 의미일 가능성도 큽니다.
# 따라서 보수적 지역명 제거 버전에서는 제거하지 않고 보존합니다.

ambiguous_region_keep_terms = {
    '대전',  # 대전광역시 vs 대전(大展), 미술대전, 난 대전 등
    '연수',  # 인천 연수구 vs 교육/훈련 연수
    '경기'   # 경기도 vs 스포츠 경기/경제 경기
}

# 기존 region_stopwords에서 애매 토큰 제외
region_stopwords_strict = set(region_stopwords) - ambiguous_region_keep_terms

print("03-13. 애매 지역명 후보 보정")
print("- 기존 지역명 제거 후보 수:", len(region_stopwords))
print("- 애매 지역명 보존 토큰:", sorted(ambiguous_region_keep_terms))
print("- 보수적 지역명 제거 후보 수:", len(region_stopwords_strict))


# ------------------------------------------------------------
# 2) 보수적 지역명 제거 함수 정의
# ------------------------------------------------------------

def filter_tokens_no_region_strict(tokens):
    """
    tokens_raw에서 기본 불용어, 1글자 잡음, 숫자, 보수적 지역명 후보를 제거합니다.

    단, '대전', '연수', '경기'처럼 의미가 애매한 토큰은 지역명으로 자동 제거하지 않습니다.
    """

    if not isinstance(tokens, list):
        return []

    filtered = []

    for token in tokens:
        token = str(token).strip()

        if token == '':
            continue

        # 숫자 토큰 제거
        if is_number_token(token):
            continue

        # 기본 불용어 제거
        if token in basic_stopwords:
            continue

        # 1글자 잡음 후보 제거
        if token in one_char_stopwords:
            continue

        # 보수적 지역명 제거
        if token in region_stopwords_strict:
            continue

        filtered.append(token)

    return filtered


# ------------------------------------------------------------
# 3) 새 열 생성
# ------------------------------------------------------------
# 기존 tokens_filtered_no_region은 덮어쓰지 않습니다.

df_text['tokens_filtered_no_region_strict'] = df_text['tokens_raw'].apply(
    filter_tokens_no_region_strict
)

df_text['tokens_filtered_no_region_strict_count'] = (
    df_text['tokens_filtered_no_region_strict'].apply(len)
)

df_text['tokens_filtered_no_region_strict_empty'] = (
    df_text['tokens_filtered_no_region_strict_count'].eq(0)
)


# ------------------------------------------------------------
# 4) 기존 no_region 버전과 보수적 버전 비교 요약
# ------------------------------------------------------------

strict_region_summary = pd.DataFrame({
    '항목': [
        '분석 대상 행 수',
        '기존 지역명 제거 버전 전체 토큰 수',
        '보수적 지역명 제거 버전 전체 토큰 수',
        '기존 지역명 제거 버전 빈 행 수',
        '보수적 지역명 제거 버전 빈 행 수',
        '복원된 토큰 수',
        '빈 행에서 회복된 행 수'
    ],
    '값': [
        len(df_text),
        int(df_text['tokens_filtered_no_region_count'].sum()),
        int(df_text['tokens_filtered_no_region_strict_count'].sum()),
        int(df_text['tokens_filtered_no_region_empty'].sum()),
        int(df_text['tokens_filtered_no_region_strict_empty'].sum()),
        int(
            df_text['tokens_filtered_no_region_strict_count'].sum()
            - df_text['tokens_filtered_no_region_count'].sum()
        ),
        int(
            (
                df_text['tokens_filtered_no_region_empty']
                & ~df_text['tokens_filtered_no_region_strict_empty']
            ).sum()
        )
    ]
})

print("\n[1] 기존 지역명 제거 버전 vs 보수적 지역명 제거 버전 요약")
display(strict_region_summary)


# ------------------------------------------------------------
# 5) 애매 지역명 토큰이 복원된 행 확인
# ------------------------------------------------------------

def list_recovered_ambiguous_tokens(tokens):
    """tokens_raw 안에 애매 지역명 보존 토큰이 있는지 확인합니다."""
    if not isinstance(tokens, list):
        return []
    return [t for t in tokens if t in ambiguous_region_keep_terms]

df_text['ambiguous_region_tokens_in_raw'] = df_text['tokens_raw'].apply(
    list_recovered_ambiguous_tokens
)

df_text['ambiguous_region_token_count'] = df_text['ambiguous_region_tokens_in_raw'].apply(len)

ambiguous_region_recovery_review = df_text.loc[
    df_text['ambiguous_region_token_count'] > 0,
    [
        '시도',
        '지자체명',
        '사업명',
        '사업명_clean_spacefix',
        'tokens_raw',
        'ambiguous_region_tokens_in_raw',
        'tokens_filtered_no_region',
        'tokens_filtered_no_region_strict',
        'tokens_filtered_no_region_count',
        'tokens_filtered_no_region_strict_count',
        '평가결과',
        '평가종류'
    ]
].copy()

print("\n[2] 애매 지역명 후보 포함 행 검토")
print("- 애매 토큰 포함 행 수:", len(ambiguous_region_recovery_review))
display(ambiguous_region_recovery_review.head(100))


# ------------------------------------------------------------
# 6) 기존 no_region에서는 비었지만 strict 버전에서는 살아난 행 확인
# ------------------------------------------------------------

recovered_empty_rows = df_text.loc[
    df_text['tokens_filtered_no_region_empty']
    & ~df_text['tokens_filtered_no_region_strict_empty'],
    [
        '시도',
        '지자체명',
        '사업명',
        '사업명_clean_spacefix',
        'tokens_raw',
        'tokens_filtered_no_region',
        'tokens_filtered_no_region_strict',
        '평가결과',
        '평가종류'
    ]
].copy()

print("\n[3] 기존 no_region에서는 빈 토큰이었지만 strict 버전에서 회복된 행")
print("- 회복된 행 수:", len(recovered_empty_rows))
display(recovered_empty_rows.head(100))


# ------------------------------------------------------------
# 7) 보수적 지역명 제거 버전 빈도표 생성
# ------------------------------------------------------------

token_freq_no_region_strict = make_token_freq(
    df_text,
    'tokens_filtered_no_region_strict'
)

print("\n[4] 보수적 지역명 제거 버전 빈도 상위 100개")
display(token_freq_no_region_strict.head(100))


# ------------------------------------------------------------
# 8) 아직 남아 있는 지역명 후보 확인
# ------------------------------------------------------------
# 여기에는 대전/연수/경기처럼 의도적으로 보존한 애매 토큰이 포함될 수 있습니다.

remaining_region_in_strict = token_freq_no_region_strict[
    token_freq_no_region_strict['token'].isin(region_stopwords)
].copy()

print("\n[5] 보수적 지역명 제거 버전에 남아 있는 기존 지역명 후보")
display(remaining_region_in_strict.head(100))


# ------------------------------------------------------------
# 9) 결과 저장
# ------------------------------------------------------------

timestamp = datetime.now().strftime('%Y%m%d_%H%M')

token_freq_no_region_strict_csv = (
    TABLE_DIR / f'03_token_freq_filtered_no_region_strict_v1_{timestamp}.csv'
)

strict_region_review_xlsx = (
    TABLE_DIR / f'03_region_filter_strict_review_v1_{timestamp}.xlsx'
)

token_freq_no_region_strict.to_csv(
    token_freq_no_region_strict_csv,
    index=False,
    encoding='utf-8-sig'
)

with pd.ExcelWriter(strict_region_review_xlsx, engine='openpyxl') as writer:
    strict_region_summary.to_excel(writer, sheet_name='summary', index=False)
    ambiguous_region_recovery_review.to_excel(writer, sheet_name='ambiguous_token_rows', index=False)
    recovered_empty_rows.to_excel(writer, sheet_name='recovered_empty_rows', index=False)
    token_freq_no_region_strict.to_excel(writer, sheet_name='freq_no_region_strict', index=False)
    remaining_region_in_strict.to_excel(writer, sheet_name='remaining_region_terms', index=False)

print("\n03-13 저장 완료")
print("- 보수적 지역명 제거 빈도표:", token_freq_no_region_strict_csv)
print("- 보수적 지역명 검토 Excel:", strict_region_review_xlsx)

print("\n저장 파일 존재 여부")
print("- token_freq_no_region_strict_csv:", token_freq_no_region_strict_csv.exists())
print("- strict_region_review_xlsx:", strict_region_review_xlsx.exists())

print("\n03-13 완료")
print("다음 단계에서는 보수적 지역명 제거 버전을 기준으로 2차 불용어 적용 여부를 판단합니다.")
```

03-14. 최종 분석용 토큰 기준 설정 및 2차 불용어 후보 비교
1. 이 단계의 목적

이 단계에서는 03번의 기본 분석용 토큰 열을 정합니다.

현재까지 만든 주요 토큰 열은 다음과 같습니다.
| 열                                  | 의미                        |
| ---------------------------------- | ------------------------- |
| `tokens_raw`                       | 형태소 분석 직후 원시 토큰           |
| `tokens_filtered_keep_region`      | 지역명 보존 1차 필터링             |
| `tokens_filtered_no_region`        | 강한 지역명 제거 1차 필터링          |
| `tokens_filtered_no_region_strict` | 애매 지역명은 보존한 보수적 지역명 제거 버전 |

현재 흐름상 기본 분석용 토큰은 tokens_filtered_no_region_strict로 두는 것이 가장 안전합니다.

다만 대회, 행사, 시설, 센터, 활동, 개선, 조성 같은 단어가 여전히 상위에 남아 있으므로, 이들을 추가로 제거했을 때 결과가 얼마나 달라지는지 비교용 버전도 만듭니다

```python
# 03-14. 최종 분석용 토큰 기준 설정 및 2차 불용어 후보 비교

# ------------------------------------------------------------
# 1) 기본 분석용 토큰 열 설정
# ------------------------------------------------------------
# 현재 기본 분석 후보는 보수적 지역명 제거 버전입니다.
# 기존 열은 보존하고, 분석용 별칭 열을 새로 만듭니다.

df_text['tokens_analysis_primary'] = df_text['tokens_filtered_no_region_strict'].copy()
df_text['tokens_analysis_primary_count'] = df_text['tokens_analysis_primary'].apply(len)
df_text['tokens_analysis_primary_empty'] = df_text['tokens_analysis_primary_count'].eq(0)

print("03-14. 기본 분석용 토큰 설정")
print("- 기본 분석용 토큰 열: tokens_analysis_primary")
print("- 기준 열: tokens_filtered_no_region_strict")


# ------------------------------------------------------------
# 2) 2차 불용어 후보 설정
# ------------------------------------------------------------
# 주의:
# 아래 단어들은 바로 최종 제거 확정이 아닙니다.
# 제거했을 때 결과가 얼마나 달라지는지 비교하기 위한 후보입니다.
#
# '대회', '행사', '시설', '센터', '프로그램'은 일부 정책분야 단서가 될 수 있으므로
# 이번 2차 후보에서는 일단 제외합니다.
#
# 비교적 일반 행정동사/사업수단에 가까운 단어만 보수적으로 후보에 넣습니다.

second_stopword_candidates = {
    '개선', '조성', '육성', '강화',
    '설치', '구입', '보급', '정비',
    '참여', '추가', '제작', '확대'
}

# 정책분야 핵심어 보호
second_stopword_candidates = second_stopword_candidates - policy_preserve_terms

print("\n[1] 2차 불용어 후보")
print(sorted(second_stopword_candidates))


# ------------------------------------------------------------
# 3) 2차 불용어 후보 적용 함수
# ------------------------------------------------------------

def apply_second_filter(tokens):
    """
    기본 분석용 토큰에서 2차 불용어 후보를 제거합니다.

    이 결과는 비교용입니다.
    최종 확정은 빈도 비교 후 결정합니다.
    """
    if not isinstance(tokens, list):
        return []

    return [
        token for token in tokens
        if str(token).strip() != ''
        and token not in second_stopword_candidates
    ]


df_text['tokens_analysis_second_filter'] = df_text['tokens_analysis_primary'].apply(
    apply_second_filter
)

df_text['tokens_analysis_second_filter_count'] = (
    df_text['tokens_analysis_second_filter'].apply(len)
)

df_text['tokens_analysis_second_filter_empty'] = (
    df_text['tokens_analysis_second_filter_count'].eq(0)
)


# ------------------------------------------------------------
# 4) 2차 필터링 영향 요약
# ------------------------------------------------------------

second_filter_summary = pd.DataFrame({
    '항목': [
        '분석 대상 행 수',
        '기본 분석용 토큰 전체 수',
        '2차 필터링 후 토큰 전체 수',
        '2차 필터링으로 제거된 토큰 수',
        '기본 분석용 토큰 빈 행 수',
        '2차 필터링 후 빈 행 수',
        '2차 필터링으로 새로 빈 행 수',
        '기본 분석용 평균 토큰 수',
        '2차 필터링 후 평균 토큰 수'
    ],
    '값': [
        len(df_text),
        int(df_text['tokens_analysis_primary_count'].sum()),
        int(df_text['tokens_analysis_second_filter_count'].sum()),
        int(
            df_text['tokens_analysis_primary_count'].sum()
            - df_text['tokens_analysis_second_filter_count'].sum()
        ),
        int(df_text['tokens_analysis_primary_empty'].sum()),
        int(df_text['tokens_analysis_second_filter_empty'].sum()),
        int(
            (
                ~df_text['tokens_analysis_primary_empty']
                & df_text['tokens_analysis_second_filter_empty']
            ).sum()
        ),
        round(df_text['tokens_analysis_primary_count'].mean(), 3),
        round(df_text['tokens_analysis_second_filter_count'].mean(), 3)
    ]
})

print("\n[2] 2차 필터링 영향 요약")
display(second_filter_summary)


# ------------------------------------------------------------
# 5) 빈 행이 새로 생긴 경우 확인
# ------------------------------------------------------------

new_empty_after_second_filter = df_text.loc[
    ~df_text['tokens_analysis_primary_empty']
    & df_text['tokens_analysis_second_filter_empty'],
    [
        '시도',
        '지자체명',
        '사업명',
        '사업명_clean_spacefix',
        'tokens_raw',
        'tokens_analysis_primary',
        'tokens_analysis_second_filter',
        '평가결과',
        '평가종류'
    ]
].copy()

print("\n[3] 2차 필터링으로 새로 빈 토큰이 된 행")
print("- 새로 빈 행 수:", len(new_empty_after_second_filter))
display(new_empty_after_second_filter.head(100))


# ------------------------------------------------------------
# 6) 기본 버전과 2차 필터링 버전 빈도표 생성
# ------------------------------------------------------------

token_freq_analysis_primary = make_token_freq(
    df_text,
    'tokens_analysis_primary'
)

token_freq_second_filter = make_token_freq(
    df_text,
    'tokens_analysis_second_filter'
)

print("\n[4] 기본 분석용 토큰 빈도 상위 100개")
display(token_freq_analysis_primary.head(100))

print("\n[5] 2차 필터링 비교용 토큰 빈도 상위 100개")
display(token_freq_second_filter.head(100))


# ------------------------------------------------------------
# 7) 기본 버전과 2차 필터링 버전 비교표
# ------------------------------------------------------------

second_filter_compare = token_freq_analysis_primary.rename(
    columns={
        'freq': 'freq_primary',
        'ratio_percent': 'ratio_primary'
    }
).merge(
    token_freq_second_filter.rename(
        columns={
            'freq': 'freq_second_filter',
            'ratio_percent': 'ratio_second_filter'
        }
    ),
    on='token',
    how='outer'
)

second_filter_compare[
    ['freq_primary', 'ratio_primary', 'freq_second_filter', 'ratio_second_filter']
] = second_filter_compare[
    ['freq_primary', 'ratio_primary', 'freq_second_filter', 'ratio_second_filter']
].fillna(0)

second_filter_compare['freq_diff_primary_minus_second'] = (
    second_filter_compare['freq_primary']
    - second_filter_compare['freq_second_filter']
)

second_filter_compare['ratio_diff_primary_minus_second'] = (
    second_filter_compare['ratio_primary']
    - second_filter_compare['ratio_second_filter']
).round(4)

second_filter_compare = second_filter_compare.sort_values(
    ['freq_primary', 'freq_second_filter'],
    ascending=False
).reset_index(drop=True)

print("\n[6] 기본 버전 vs 2차 필터링 버전 비교 상위 100개")
display(second_filter_compare.head(100))


# ------------------------------------------------------------
# 8) 2차 제거 후보가 실제로 제거되었는지 확인
# ------------------------------------------------------------

removed_second_terms_check = second_filter_compare[
    second_filter_compare['token'].isin(second_stopword_candidates)
].copy()

print("\n[7] 2차 불용어 후보 제거 확인")
display(removed_second_terms_check)


# ------------------------------------------------------------
# 9) 결과 저장
# ------------------------------------------------------------

timestamp = datetime.now().strftime('%Y%m%d_%H%M')

token_freq_primary_csv = TABLE_DIR / f'03_token_freq_analysis_primary_v1_{timestamp}.csv'
token_freq_second_csv = TABLE_DIR / f'03_token_freq_second_filter_compare_v1_{timestamp}.csv'
second_filter_review_xlsx = TABLE_DIR / f'03_second_filter_review_v1_{timestamp}.xlsx'

token_freq_analysis_primary.to_csv(
    token_freq_primary_csv,
    index=False,
    encoding='utf-8-sig'
)

token_freq_second_filter.to_csv(
    token_freq_second_csv,
    index=False,
    encoding='utf-8-sig'
)

with pd.ExcelWriter(second_filter_review_xlsx, engine='openpyxl') as writer:
    second_filter_summary.to_excel(writer, sheet_name='summary', index=False)
    new_empty_after_second_filter.to_excel(writer, sheet_name='new_empty_rows', index=False)
    token_freq_analysis_primary.to_excel(writer, sheet_name='freq_primary', index=False)
    token_freq_second_filter.to_excel(writer, sheet_name='freq_second_filter', index=False)
    second_filter_compare.to_excel(writer, sheet_name='freq_compare', index=False)
    removed_second_terms_check.to_excel(writer, sheet_name='removed_terms', index=False)

print("\n03-14 저장 완료")
print("- 기본 분석용 토큰 빈도표:", token_freq_primary_csv)
print("- 2차 필터링 비교용 빈도표:", token_freq_second_csv)
print("- 2차 필터링 검토 Excel:", second_filter_review_xlsx)

print("\n저장 파일 존재 여부")
print("- token_freq_primary_csv:", token_freq_primary_csv.exists())
print("- token_freq_second_csv:", token_freq_second_csv.exists())
print("- second_filter_review_xlsx:", second_filter_review_xlsx.exists())

print("\n03-14 완료")
print("다음 단계에서는 2차 필터링을 최종 적용할지, 아니면 기본 분석용 토큰을 유지할지 결정합니다.")
```

```python
# 03-14B. 03번 산출물 파일 정리: final / review / archive_intermediate로 이동
# 주의: 삭제하지 않고 폴더만 나눕니다.

from pathlib import Path
import shutil

# 정리할 폴더
FINAL_DIR = TABLE_DIR / 'final'
REVIEW_DIR = TABLE_DIR / 'review'
ARCHIVE_DIR = TABLE_DIR / 'archive_intermediate'

FINAL_DIR.mkdir(parents=True, exist_ok=True)
REVIEW_DIR.mkdir(parents=True, exist_ok=True)
ARCHIVE_DIR.mkdir(parents=True, exist_ok=True)

# 현재 tables 폴더의 파일 목록
all_table_files = [p for p in TABLE_DIR.iterdir() if p.is_file()]

print("정리 전 tables 파일 수:", len(all_table_files))

# 파일명 패턴별 이동 규칙
final_patterns = [
    '03_token_freq_analysis_primary',
    '03_token_freq_filtered_no_region_strict'
]

review_patterns = [
    '03_char_space_issue_candidates',
    '03_region_filter_strict_review',
    '03_second_filter_review'
]

archive_patterns = [
    '03_token_freq_raw',
    '03_candidate_stopword_check',
    '03_token_freq_filtered_keep_region',
    '03_token_freq_filtered_no_region_v1',
    '03_token_filter_summary',
    '03_filter_compare',
    '03_token_freq_second_filter_compare'
]

def move_by_patterns(file_path, patterns, target_dir):
    """파일명이 특정 패턴을 포함하면 target_dir로 이동합니다."""
    for pattern in patterns:
        if pattern in file_path.name:
            target_path = target_dir / file_path.name

            # 이미 대상 위치에 있으면 건너뜁니다.
            if file_path.resolve() == target_path.resolve():
                return True

            shutil.move(str(file_path), str(target_path))
            return True
    return False

moved_log = []

for file_path in all_table_files:
    moved = False

    if move_by_patterns(file_path, final_patterns, FINAL_DIR):
        moved_log.append((file_path.name, 'final'))
        moved = True

    elif move_by_patterns(file_path, review_patterns, REVIEW_DIR):
        moved_log.append((file_path.name, 'review'))
        moved = True

    elif move_by_patterns(file_path, archive_patterns, ARCHIVE_DIR):
        moved_log.append((file_path.name, 'archive_intermediate'))
        moved = True

    else:
        moved_log.append((file_path.name, 'not_moved'))

# 이동 결과 확인
move_result = pd.DataFrame(moved_log, columns=['file_name', 'moved_to'])

print("\n파일 정리 결과")
display(move_result)

print("\n정리 후 폴더별 파일 수")
print("- final:", len(list(FINAL_DIR.glob('*'))))
print("- review:", len(list(REVIEW_DIR.glob('*'))))
print("- archive_intermediate:", len(list(ARCHIVE_DIR.glob('*'))))
print("- tables 루트에 남은 파일:", len([p for p in TABLE_DIR.iterdir() if p.is_file()]))

print("\n폴더 경로")
print("- FINAL_DIR:", FINAL_DIR)
print("- REVIEW_DIR:", REVIEW_DIR)
print("- ARCHIVE_DIR:", ARCHIVE_DIR)
```

```python
# 03-RESTART 수정 버전
# 기존 03번 중간 분석 parquet 파일을 찾아서 불러오기만 합니다.
# 새 파일은 생성하지 않습니다.

from pathlib import Path
from google.colab import drive
import pandas as pd
import numpy as np
import re
from datetime import datetime
import unicodedata

drive.mount('/content/drive', force_remount=True)

MYDRIVE = Path('/content/drive/MyDrive')

# ------------------------------------------------------------
# 1) 파공장(PA Factory) 폴더 찾기
# ------------------------------------------------------------

pa_candidates = [
    p for p in MYDRIVE.iterdir()
    if p.is_dir() and 'PA Factory' in p.name
]

print("파공장 후보 폴더 수:", len(pa_candidates))

for i, p in enumerate(pa_candidates, start=1):
    print(f"{i}. {p}")

if len(pa_candidates) != 1:
    raise FileNotFoundError("파공장(PA Factory) 폴더를 정확히 1개 찾지 못했습니다.")

PA_FACTORY_DIR = pa_candidates[0]


# ------------------------------------------------------------
# 2) 03번 중간 분석 parquet 파일 넓게 검색
# ------------------------------------------------------------

# 1차: 파일명 정확 패턴 검색
raw_candidates = list(
    PA_FACTORY_DIR.rglob('03_project_name_text_analysis_v1_*.parquet')
)

print("\n1차 검색 후보 수:", len(raw_candidates))

for i, p in enumerate(sorted(raw_candidates), start=1):
    print(f"{i}. {p}")

# 2차: 혹시 파일명 일부가 다를 경우 더 넓게 검색
if len(raw_candidates) == 0:
    print("\n1차 검색에서 찾지 못해 더 넓게 검색합니다.")

    raw_candidates = list(
        PA_FACTORY_DIR.rglob('*project_name_text*analysis*.parquet')
    )

    print("2차 검색 후보 수:", len(raw_candidates))

    for i, p in enumerate(sorted(raw_candidates), start=1):
        print(f"{i}. {p}")

# 3차: parquet 전체 중 03번 final 후보 확인
if len(raw_candidates) == 0:
    print("\n2차 검색에서도 찾지 못해 parquet 전체를 검색합니다.")

    all_parquet_candidates = list(PA_FACTORY_DIR.rglob('*.parquet'))

    print("PA Factory 아래 parquet 전체 수:", len(all_parquet_candidates))

    for i, p in enumerate(sorted(all_parquet_candidates), start=1):
        print(f"{i}. {p}")

    raise FileNotFoundError(
        "03_project_name_text_analysis 관련 parquet을 찾지 못했습니다. "
        "위 parquet 목록에서 실제 파일명을 확인해야 합니다."
    )


# ------------------------------------------------------------
# 3) final 폴더 안에 있는 후보를 우선 선택
# ------------------------------------------------------------

# 경로 문자열을 유니코드 NFC로 정규화해서 비교합니다.
def norm_path_text(path):
    return unicodedata.normalize('NFC', str(path))

final_candidates = [
    p for p in raw_candidates
    if 'tables/final' in norm_path_text(p).replace('\\', '/')
]

print("\nfinal 폴더 후보 수:", len(final_candidates))

for i, p in enumerate(sorted(final_candidates), start=1):
    print(f"{i}. {p}")

# final 후보가 있으면 그중 최신 파일 사용
# final 후보가 없으면 raw 후보 중 최신 파일 사용
if len(final_candidates) > 0:
    ANALYSIS_FILE = sorted(final_candidates)[-1]
else:
    print("\n주의: final 폴더 후보가 없어 전체 후보 중 최신 파일을 사용합니다.")
    ANALYSIS_FILE = sorted(raw_candidates)[-1]

print("\n사용할 03번 중간 분석 파일:")
print(ANALYSIS_FILE)


# ------------------------------------------------------------
# 4) 경로 변수 복원
# ------------------------------------------------------------

FINAL_DIR = ANALYSIS_FILE.parent
TABLE_DIR = FINAL_DIR.parent
OUTPUT_DIR = TABLE_DIR.parent
PROJECT_OUTPUT_DIR = OUTPUT_DIR.parent
FIGURE_DIR = OUTPUT_DIR / 'figures'

print("\n복원된 경로")
print("FINAL_DIR:", FINAL_DIR)
print("TABLE_DIR:", TABLE_DIR)
print("OUTPUT_DIR:", OUTPUT_DIR)
print("FIGURE_DIR:", FIGURE_DIR)


# ------------------------------------------------------------
# 5) 03번 중간 분석 데이터 불러오기
# ------------------------------------------------------------

df_03_analysis = pd.read_parquet(ANALYSIS_FILE)
df_text = df_03_analysis.copy()

print("\n03번 중간 분석 데이터 불러오기 완료")
print("- 행 수:", df_text.shape[0])
print("- 열 수:", df_text.shape[1])

print("\n주요 열 존재 여부")
key_cols = [
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'tokens_raw',
    'tokens_analysis_primary',
    'tokens_analysis_second_filter'
]

for col in key_cols:
    print(f"- {col}:", col in df_text.columns)


# ------------------------------------------------------------
# 6) 빈도표 생성 함수 복원
# ------------------------------------------------------------

def make_token_freq(df, token_col):
    """
    리스트 형태 토큰 열을 펼쳐 빈도표를 생성합니다.
    """
    token_long = (
        df[[token_col]]
        .explode(token_col)
        .rename(columns={token_col: 'token'})
        .dropna()
        .copy()
    )

    token_long['token'] = token_long['token'].astype(str).str.strip()
    token_long = token_long[token_long['token'] != '']

    token_freq = (
        token_long['token']
        .value_counts()
        .reset_index()
    )

    token_freq.columns = ['token', 'freq']
    token_freq['ratio_percent'] = (
        token_freq['freq'] / token_freq['freq'].sum() * 100
    ).round(4)

    return token_freq


# ------------------------------------------------------------
# 7) 기본 빈도표 메모리에서만 재생성
# ------------------------------------------------------------

token_freq_analysis_primary = make_token_freq(
    df_text,
    'tokens_analysis_primary'
)

token_freq_second_filter = make_token_freq(
    df_text,
    'tokens_analysis_second_filter'
)

print("\n빈도표 재생성 완료")
print("- token_freq_analysis_primary 행 수:", len(token_freq_analysis_primary))
print("- token_freq_second_filter 행 수:", len(token_freq_second_filter))

print("\n기본 분석용 토큰 빈도 상위 20개")
display(token_freq_analysis_primary.head(20))

print("\n재개 준비 완료")
```

03번 사업명 텍스트 분석에서 지금까지 처리한 내용을 요약하고, 이후 logs/work_log.md에 기록할 수 있는 문장을 정리합니다.

```python
# 03-RELOAD. 파일 생성 없이 03번 중간 분석 데이터 불러오기
# 주의: 이 셀은 기존 parquet 파일을 읽기만 하며, 새 파일을 생성하지 않습니다.

from pathlib import Path
from google.colab import drive
import pandas as pd
import numpy as np
import re
import unicodedata

# ------------------------------------------------------------
# 1) Google Drive 마운트
# ------------------------------------------------------------

drive.mount('/content/drive', force_remount=True)

MYDRIVE = Path('/content/drive/MyDrive')

# ------------------------------------------------------------
# 2) PA Factory 폴더 찾기
# ------------------------------------------------------------

pa_candidates = [
    p for p in MYDRIVE.iterdir()
    if p.is_dir() and 'PA Factory' in p.name
]

print("파공장 후보 폴더 수:", len(pa_candidates))

for i, p in enumerate(pa_candidates, start=1):
    print(f"{i}. {p}")

if len(pa_candidates) != 1:
    raise FileNotFoundError(
        "PA Factory 폴더를 정확히 1개 찾지 못했습니다. "
        "Drive 바로가기 또는 폴더명을 확인하세요."
    )

PA_FACTORY_DIR = pa_candidates[0]

# ------------------------------------------------------------
# 3) 03번 중간 분석 parquet 찾기
# ------------------------------------------------------------

analysis_candidates = list(
    PA_FACTORY_DIR.rglob('03_project_name_text_analysis_v1_*.parquet')
)

# final 폴더 안의 파일 우선
final_candidates = [
    p for p in analysis_candidates
    if 'tables/final' in str(p).replace('\\', '/')
]

print("\n03번 parquet 후보 수:", len(analysis_candidates))
print("final 폴더 후보 수:", len(final_candidates))

if len(final_candidates) > 0:
    ANALYSIS_FILE = sorted(final_candidates)[-1]
elif len(analysis_candidates) > 0:
    ANALYSIS_FILE = sorted(analysis_candidates)[-1]
else:
    raise FileNotFoundError(
        "03_project_name_text_analysis_v1_*.parquet 파일을 찾지 못했습니다."
    )

print("\n사용할 03번 중간 분석 파일:")
print(ANALYSIS_FILE)

# ------------------------------------------------------------
# 4) 경로 변수 복원
# ------------------------------------------------------------

FINAL_DIR = ANALYSIS_FILE.parent
TABLE_DIR = FINAL_DIR.parent
OUTPUT_DIR = TABLE_DIR.parent
FIGURE_DIR = OUTPUT_DIR / 'figures'

print("\n복원된 경로")
print("OUTPUT_DIR:", OUTPUT_DIR)
print("TABLE_DIR:", TABLE_DIR)
print("FINAL_DIR:", FINAL_DIR)
print("FIGURE_DIR:", FIGURE_DIR)

# ------------------------------------------------------------
# 5) 03번 데이터 불러오기
# ------------------------------------------------------------

df_03_analysis = pd.read_parquet(ANALYSIS_FILE)

# 기존 코드와 호환되도록 df_text 이름도 같이 사용
df_text = df_03_analysis.copy()

print("\n03번 중간 분석 데이터 불러오기 완료")
print("- 행 수:", df_text.shape[0])
print("- 열 수:", df_text.shape[1])

# ------------------------------------------------------------
# 6) 주요 열 확인
# ------------------------------------------------------------

key_cols = [
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'tokens_raw',
    'tokens_analysis_primary',
    'tokens_analysis_second_filter'
]

print("\n주요 열 존재 여부")
for col in key_cols:
    print(f"- {col}: {col in df_text.columns}")

# ------------------------------------------------------------
# 7) 빈도표 생성 함수 복원
# ------------------------------------------------------------

def make_token_freq(df, token_col):
    """
    리스트 형태 토큰 열을 펼쳐 빈도표를 생성합니다.
    새 파일을 저장하지 않고 메모리에서만 생성합니다.
    """
    token_long = (
        df[[token_col]]
        .explode(token_col)
        .rename(columns={token_col: 'token'})
        .dropna()
        .copy()
    )

    token_long['token'] = token_long['token'].astype(str).str.strip()
    token_long = token_long[token_long['token'] != '']

    token_freq = (
        token_long['token']
        .value_counts()
        .reset_index()
    )

    token_freq.columns = ['token', 'freq']
    token_freq['ratio_percent'] = (
        token_freq['freq'] / token_freq['freq'].sum() * 100
    ).round(4)

    return token_freq

# ------------------------------------------------------------
# 8) 기존 빈도표를 메모리에서만 재생성
# ------------------------------------------------------------

token_freq_analysis_primary = make_token_freq(
    df_text,
    'tokens_analysis_primary'
)

token_freq_second_filter = make_token_freq(
    df_text,
    'tokens_analysis_second_filter'
)

print("\n빈도표 메모리 재생성 완료")
print("- token_freq_analysis_primary 행 수:", len(token_freq_analysis_primary))
print("- token_freq_second_filter 행 수:", len(token_freq_second_filter))

print("\n기본 분석용 토큰 빈도 상위 20개")
display(token_freq_analysis_primary.head(20))

print("\n재시작 후 작업 준비 완료")
print("이 셀은 새 파일을 생성하지 않았습니다.")
```

03-SUPP-1. V1 원자료 기준 v1_row_id 및 메타열 보강 준비
1. 이 단계의 목적

현재 메모리에 복원된 기존 03번 master df_text에 대해, 원본 V1 데이터의 행 순서를 기준으로 v1_row_id를 보강합니다.

핵심은 다음입니다.

기존 토큰화 결과는 건드리지 않음
기존 사업명, 사업명_clean, tokens_analysis_primary 등은 변경하지 않음
원본 V1의 행 순서 기준으로 v1_row_id 생성
사업명 결측 14행을 제외한 79,262행과 현재 03번 master가 같은 순서인지 검증
가능하면 원파일명도 원본 V1에서 가져와 보강

```python
# 03-SUPP-1. V1 원자료 기준 v1_row_id 및 메타열 보강 준비
# 주의: 이 셀은 파일을 저장하지 않습니다.

from pathlib import Path
import pandas as pd
import numpy as np
import re
import unicodedata
from datetime import datetime
from zoneinfo import ZoneInfo

# ------------------------------------------------------------
# 1) 현재 03번 master 데이터 확인
# ------------------------------------------------------------

if 'df_text' not in globals():
    raise NameError(
        "df_text가 없습니다. 먼저 03번 master parquet을 불러오는 RELOAD 셀을 실행하세요."
    )

print("현재 df_text 상태")
print("- 행 수:", len(df_text))
print("- 열 수:", df_text.shape[1])

required_current_cols = [
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'tokens_raw',
    'tokens_analysis_primary',
    'tokens_analysis_second_filter'
]

print("\n현재 03번 master 주요 열 존재 여부")
for col in required_current_cols:
    print(f"- {col}:", col in df_text.columns)


# ------------------------------------------------------------
# 2) 원본 V1 parquet 파일 찾기
# ------------------------------------------------------------
# 기존 RELOAD 셀에서 PA_FACTORY_DIR이 있으면 사용하고,
# 없으면 /content/drive/MyDrive 아래에서 다시 찾습니다.

if 'PA_FACTORY_DIR' not in globals():
    MYDRIVE = Path('/content/drive/MyDrive')
    pa_candidates = [
        p for p in MYDRIVE.iterdir()
        if p.is_dir() and 'PA Factory' in p.name
    ]

    if len(pa_candidates) != 1:
        raise FileNotFoundError(
            "PA Factory 폴더를 정확히 1개 찾지 못했습니다. "
            "Drive 마운트와 바로가기 위치를 확인하세요."
        )

    PA_FACTORY_DIR = pa_candidates[0]

v1_candidates = list(
    PA_FACTORY_DIR.rglob('all_clean_data_v1_20260523_0054_KST.parquet')
)

print("\n원본 V1 parquet 후보 수:", len(v1_candidates))

for i, p in enumerate(v1_candidates, start=1):
    print(f"{i}. {p}")

if len(v1_candidates) == 0:
    raise FileNotFoundError(
        "all_clean_data_v1_20260523_0054_KST.parquet 파일을 찾지 못했습니다."
    )

# 이름이 정확히 같다면 보통 1개지만, 혹시 여러 개면 경로에 01_v1_분석데이터가 포함된 것을 우선 사용
preferred_v1 = [
    p for p in v1_candidates
    if '01_v1_' in str(p) or '01_v1_' in unicodedata.normalize('NFC', str(p))
]

if len(preferred_v1) > 0:
    V1_FILE = sorted(preferred_v1)[-1]
else:
    V1_FILE = sorted(v1_candidates)[-1]

print("\n사용할 원본 V1 파일:")
print(V1_FILE)


# ------------------------------------------------------------
# 3) 원본 V1 데이터 불러오기 및 v1_row_id 생성
# ------------------------------------------------------------

df_v1 = pd.read_parquet(V1_FILE).copy()

# 원본 V1 행 순서를 안정적으로 보존하는 식별자
if 'v1_row_id' not in df_v1.columns:
    df_v1['v1_row_id'] = df_v1.index.astype(int)

print("\n원본 V1 데이터")
print("- 행 수:", len(df_v1))
print("- 열 수:", df_v1.shape[1])
print("- v1_row_id 최소:", df_v1['v1_row_id'].min())
print("- v1_row_id 최대:", df_v1['v1_row_id'].max())
print("- v1_row_id 중복 수:", int(df_v1['v1_row_id'].duplicated().sum()))

print("\n원본 V1 열 목록")
for col in df_v1.columns:
    print("-", col)


# ------------------------------------------------------------
# 4) 원본 V1에서 텍스트 분석 대상 행만 분리
# ------------------------------------------------------------
# 03번에서는 사업명 결측 14행을 텍스트 분석 대상에서 제외했습니다.
# 여기서도 같은 기준으로 V1의 텍스트 분석 대상 행을 만듭니다.

if '사업명' not in df_v1.columns:
    raise KeyError("원본 V1 데이터에 '사업명' 열이 없습니다.")

v1_project_name_text = df_v1['사업명'].fillna('').astype(str).str.strip()

v1_project_name_missing_or_blank = (
    df_v1['사업명'].isna() |
    v1_project_name_text.eq('')
)

df_v1_text = df_v1.loc[~v1_project_name_missing_or_blank].copy()

print("\n원본 V1 기준 텍스트 분석 대상")
print("- V1 전체 행 수:", len(df_v1))
print("- 사업명 결측/공백 제외 행 수:", len(df_v1_text))
print("- 제외 행 수:", int(v1_project_name_missing_or_blank.sum()))

if len(df_v1_text) != len(df_text):
    raise ValueError(
        f"행 수가 맞지 않습니다. df_v1_text={len(df_v1_text)}, df_text={len(df_text)}. "
        "03번 master와 V1 텍스트 대상 행의 정렬 기준을 다시 확인해야 합니다."
    )


# ------------------------------------------------------------
# 5) df_text와 df_v1_text의 사업명 순서 일치 여부 확인
# ------------------------------------------------------------
# v1_row_id를 행 순서로 붙이려면 두 데이터의 행 순서가 같아야 합니다.
# 원본 사업명 기준으로 순서 일치 여부를 확인합니다.

def normalize_compare_text(s):
    if pd.isna(s):
        return ''
    return unicodedata.normalize('NFC', str(s)).strip()

text_03 = df_text['사업명'].apply(normalize_compare_text).reset_index(drop=True)
text_v1 = df_v1_text['사업명'].apply(normalize_compare_text).reset_index(drop=True)

alignment_match = text_03.eq(text_v1)
alignment_mismatch_count = int((~alignment_match).sum())

print("\n03번 master와 V1 텍스트 대상 사업명 순서 비교")
print("- 순서 불일치 행 수:", alignment_mismatch_count)

if alignment_mismatch_count > 0:
    mismatch_sample = pd.DataFrame({
        'pos': np.where(~alignment_match)[0],
        '사업명_03': text_03[~alignment_match].values,
        '사업명_v1': text_v1[~alignment_match].values
    }).head(20)

    print("\n순서 불일치 샘플")
    display(mismatch_sample)

    raise ValueError(
        "df_text와 df_v1_text의 사업명 순서가 일치하지 않습니다. "
        "v1_row_id를 단순 행 순서로 붙이면 위험합니다."
    )

print("사업명 순서가 일치합니다. v1_row_id를 행 순서 기준으로 보강할 수 있습니다.")


# ------------------------------------------------------------
# 6) 원파일명 후보 열 탐색
# ------------------------------------------------------------
# 프로젝트에서 파일 식별 메타열 이름이 원파일명/원본파일명/source_file/file_name 등으로 다를 수 있어 후보를 탐색합니다.

file_meta_candidates = [
    '원파일명',
    '원본파일명',
    'source_file',
    'file_name',
    '파일명',
    'source_path',
    '원본경로'
]

existing_file_meta_cols = [
    col for col in file_meta_candidates
    if col in df_v1_text.columns
]

print("\n원본 V1에서 발견된 파일 메타열 후보:")
if existing_file_meta_cols:
    for col in existing_file_meta_cols:
        print("-", col)
else:
    print("- 없음")

# 원파일명 표준 열 생성에 사용할 후보
source_file_col = existing_file_meta_cols[0] if existing_file_meta_cols else None

print("\n원파일명 보강에 사용할 열:", source_file_col)


# ------------------------------------------------------------
# 7) df_text_supp 생성 및 v1_row_id / 메타열 보강
# ------------------------------------------------------------
# 기존 df_text는 직접 덮어쓰지 않고, 보강 후보 df_text_supp를 따로 만듭니다.

df_text_supp = df_text.copy()

# v1_row_id가 이미 있으면 비교하고, 없으면 추가합니다.
if 'v1_row_id' in df_text_supp.columns:
    current_id = df_text_supp['v1_row_id'].reset_index(drop=True)
    expected_id = df_v1_text['v1_row_id'].reset_index(drop=True)

    id_mismatch_count = int((current_id != expected_id).sum())
    print("\n기존 v1_row_id 존재")
    print("- 기존 v1_row_id와 V1 기준 v1_row_id 불일치 수:", id_mismatch_count)

    if id_mismatch_count > 0:
        raise ValueError(
            "기존 v1_row_id가 V1 기준과 일치하지 않습니다. "
            "자동 보정하지 말고 원인을 확인해야 합니다."
        )
else:
    df_text_supp['v1_row_id'] = df_v1_text['v1_row_id'].to_numpy()
    print("\nv1_row_id를 새로 추가했습니다.")

# 원파일명 표준 열 보강
if '원파일명' not in df_text_supp.columns:
    if source_file_col is not None:
        df_text_supp['원파일명'] = df_v1_text[source_file_col].to_numpy()
        print(f"'원파일명' 열을 원본 V1의 '{source_file_col}' 열에서 생성했습니다.")
    else:
        df_text_supp['원파일명'] = pd.NA
        print("주의: 원본 V1에서 파일 메타열을 찾지 못해 '원파일명'을 결측으로 생성했습니다.")
else:
    print("'원파일명' 열이 이미 존재합니다.")

# 후속 병합에 필요한 핵심 메타열 중 현재 df_text_supp에 없고 V1에는 있는 열은 보강
meta_cols_to_check = [
    '시도',
    '지자체명',
    '사업명',
    '보조사업자',
    '평가결과',
    '평가종류',
    '사업비합계',
    '보조금',
    '사업자부담'
]

added_meta_cols = []

for col in meta_cols_to_check:
    if col not in df_text_supp.columns and col in df_v1_text.columns:
        df_text_supp[col] = df_v1_text[col].to_numpy()
        added_meta_cols.append(col)

print("\n추가 보강한 메타열:", added_meta_cols if added_meta_cols else "없음")


# ------------------------------------------------------------
# 8) 보강 결과 점검
# ------------------------------------------------------------

v1_row_id_missing = int(df_text_supp['v1_row_id'].isna().sum())
v1_row_id_duplicate = int(df_text_supp['v1_row_id'].duplicated().sum())
v1_row_id_min = int(df_text_supp['v1_row_id'].min())
v1_row_id_max = int(df_text_supp['v1_row_id'].max())

wonfile_exists = '원파일명' in df_text_supp.columns
wonfile_missing = int(df_text_supp['원파일명'].isna().sum()) if wonfile_exists else None

core_required_cols = [
    'v1_row_id',
    '시도',
    '지자체명',
    '원파일명',
    '사업명',
    '보조사업자',
    '평가결과',
    '평가종류'
]

core_col_status = pd.DataFrame({
    '열': core_required_cols,
    '존재여부': [col in df_text_supp.columns for col in core_required_cols],
    '결측수': [
        int(df_text_supp[col].isna().sum()) if col in df_text_supp.columns else None
        for col in core_required_cols
    ]
})

supp_check_summary = pd.DataFrame({
    '항목': [
        '보강 후보 데이터 행 수',
        '보강 후보 데이터 열 수',
        'v1_row_id 결측 수',
        'v1_row_id 중복 수',
        'v1_row_id 최소값',
        'v1_row_id 최대값',
        '원파일명 존재 여부',
        '원파일명 결측 수',
        '사업명 순서 불일치 수'
    ],
    '값': [
        len(df_text_supp),
        df_text_supp.shape[1],
        v1_row_id_missing,
        v1_row_id_duplicate,
        v1_row_id_min,
        v1_row_id_max,
        wonfile_exists,
        wonfile_missing,
        alignment_mismatch_count
    ]
})

print("\n[1] 보강 결과 요약")
display(supp_check_summary)

print("\n[2] 핵심 열 존재 및 결측 확인")
display(core_col_status)

print("\n[3] v1_row_id와 핵심 열 샘플")
sample_cols = [
    'v1_row_id',
    '시도',
    '지자체명',
    '원파일명',
    '사업명',
    '사업명_clean_spacefix',
    'tokens_analysis_primary',
    '보조사업자',
    '평가결과',
    '평가종류'
]
sample_cols = [col for col in sample_cols if col in df_text_supp.columns]
display(df_text_supp[sample_cols].head(20))


# ------------------------------------------------------------
# 9) 04번 handoff 핵심 열 목록
# ------------------------------------------------------------

core_handoff_cols = [
    'v1_row_id',
    '시도',
    '지자체명',
    '원파일명',
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'tokens_analysis_primary',
    'tokens_analysis_second_filter',
    '보조사업자',
    '평가결과',
    '평가종류'
]

existing_core_handoff_cols = [
    col for col in core_handoff_cols
    if col in df_text_supp.columns
]

print("\n[4] 04번으로 넘길 핵심 열 목록")
for col in existing_core_handoff_cols:
    print("-", col)

print("\n03-SUPP-1 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("위 결과에서 v1_row_id 결측 0, 중복 0, 사업명 순서 불일치 0이면 다음 저장 셀로 넘어가면 됩니다.")
```

03-SUPP-2. 식별자 보강 03번 master 새 파일 저장

03-SUPP-1 결과가 정상일 때만 아래 셀을 실행하면 됩니다.

1. 이 단계의 목적

df_text_supp를 새 03번 master 파일로 저장합니다.

기존 파일은 덮어쓰지 않고, 새 timestamp를 붙여 저장합니다.

```python
# 03-SUPP-2. 식별자 보강 03번 master 새 파일 저장
# 주의: 기존 파일을 덮어쓰지 않고 새 timestamp로 저장합니다.

from zoneinfo import ZoneInfo

# ------------------------------------------------------------
# 1) 저장 전 필수 객체 확인
# ------------------------------------------------------------

if 'df_text_supp' not in globals():
    raise NameError("df_text_supp가 없습니다. 먼저 03-SUPP-1 셀을 실행하세요.")

if 'FINAL_DIR' not in globals():
    raise NameError("FINAL_DIR이 없습니다. 먼저 RELOAD 셀을 실행하세요.")

FINAL_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2) 저장 전 최종 안전 점검
# ------------------------------------------------------------

required_save_cols = [
    'v1_row_id',
    '시도',
    '지자체명',
    '원파일명',
    '사업명',
    '보조사업자',
    '평가결과',
    '평가종류',
    '사업명_clean_spacefix',
    'tokens_analysis_primary',
    'tokens_analysis_second_filter'
]

missing_required_save_cols = [
    col for col in required_save_cols
    if col not in df_text_supp.columns
]

if missing_required_save_cols:
    print("주의: 저장 전 필수 후보 열 중 누락된 열이 있습니다.")
    for col in missing_required_save_cols:
        print("-", col)

# v1_row_id는 반드시 정상이어야 저장 진행
if 'v1_row_id' not in df_text_supp.columns:
    raise KeyError("v1_row_id가 없어 저장할 수 없습니다.")

if df_text_supp['v1_row_id'].isna().sum() > 0:
    raise ValueError("v1_row_id 결측이 있어 저장을 중단합니다.")

if df_text_supp['v1_row_id'].duplicated().sum() > 0:
    raise ValueError("v1_row_id 중복이 있어 저장을 중단합니다.")

# ------------------------------------------------------------
# 3) 저장 전 요약 생성
# ------------------------------------------------------------

save_check_summary_supp = pd.DataFrame({
    '항목': [
        '저장 데이터 행 수',
        '저장 데이터 열 수',
        'v1_row_id 결측 수',
        'v1_row_id 중복 수',
        'v1_row_id 최소값',
        'v1_row_id 최대값',
        '원파일명 존재 여부',
        '원파일명 결측 수',
        'tokens_raw 빈 행 수',
        'tokens_analysis_primary 빈 행 수',
        'tokens_analysis_second_filter 빈 행 수'
    ],
    '값': [
        len(df_text_supp),
        df_text_supp.shape[1],
        int(df_text_supp['v1_row_id'].isna().sum()),
        int(df_text_supp['v1_row_id'].duplicated().sum()),
        int(df_text_supp['v1_row_id'].min()),
        int(df_text_supp['v1_row_id'].max()),
        '원파일명' in df_text_supp.columns,
        int(df_text_supp['원파일명'].isna().sum()) if '원파일명' in df_text_supp.columns else None,
        int(df_text_supp['tokens_raw_empty'].sum()) if 'tokens_raw_empty' in df_text_supp.columns else None,
        int(df_text_supp['tokens_analysis_primary_empty'].sum()) if 'tokens_analysis_primary_empty' in df_text_supp.columns else None,
        int(df_text_supp['tokens_analysis_second_filter_empty'].sum()) if 'tokens_analysis_second_filter_empty' in df_text_supp.columns else None
    ]
})

print("저장 전 최종 점검")
display(save_check_summary_supp)

# ------------------------------------------------------------
# 4) 저장 파일명 설정
# ------------------------------------------------------------

timestamp = datetime.now(ZoneInfo("Asia/Seoul")).strftime('%Y%m%d_%H%M')

supp_master_parquet = FINAL_DIR / f'03_project_name_text_analysis_v1_{timestamp}.parquet'
supp_sample_xlsx = FINAL_DIR / f'03_project_name_text_analysis_sample_v1_{timestamp}.xlsx'
supp_columns_csv = FINAL_DIR / f'03_project_name_text_analysis_columns_v1_{timestamp}.csv'
supp_save_check_csv = FINAL_DIR / f'03_project_name_text_analysis_save_check_v1_{timestamp}.csv'
supp_handoff_memo_txt = FINAL_DIR / f'03_handoff_memo_for_04_v1_{timestamp}.txt'

# ------------------------------------------------------------
# 5) 파일 저장
# ------------------------------------------------------------

df_text_supp.to_parquet(supp_master_parquet, index=False)

# 샘플은 기존처럼 head(1000) 저장
# 전체 대표 샘플이 아니라 구조 확인용임
df_text_supp.head(1000).to_excel(supp_sample_xlsx, index=False)

columns_df_supp = pd.DataFrame({
    'column_order': range(1, len(df_text_supp.columns) + 1),
    'column_name': df_text_supp.columns
})

columns_df_supp.to_csv(supp_columns_csv, index=False, encoding='utf-8-sig')
save_check_summary_supp.to_csv(supp_save_check_csv, index=False, encoding='utf-8-sig')

# ------------------------------------------------------------
# 6) 04번 전달 메모 생성
# ------------------------------------------------------------

new_master_name = supp_master_parquet.name

core_handoff_cols = [
    'v1_row_id',
    '시도',
    '지자체명',
    '원파일명',
    '사업명',
    '사업명_clean',
    '사업명_clean_spacefix',
    'tokens_analysis_primary',
    'tokens_analysis_second_filter',
    '보조사업자',
    '평가결과',
    '평가종류'
]

existing_core_handoff_cols = [
    col for col in core_handoff_cols
    if col in df_text_supp.columns
]

handoff_memo = f"""
# 04번 노트북 전달 메모

## 새 03번 master 파일

04번 정책분야 분석에서는 아래 파일을 입력으로 사용한다.

`{new_master_name}`

Drive 경로:

`{FINAL_DIR}`

## 사용 이유

이 파일은 기존 03번 master에 원본 V1 행 식별자인 `v1_row_id`와 후속 병합에 필요한 메타열을 보강한 버전이다.

기존 토큰화 결과, 불용어 처리 결과, 정책어휘 빈도 결과는 임의로 변경하지 않았다.

## 행 수 기준

- 원본 V1 데이터 행 수: 79,276행
- 03번 텍스트 분석 master 행 수: 79,262행
- V1 79,276행 중 `사업명` 결측 14행은 텍스트 분석용 데이터에서 제외되었다.
- 원본 행 식별자인 `v1_row_id`는 V1 데이터의 원래 행 순서를 기준으로 한다.

## 병합 기준

- `v1_row_id`를 05번 병합의 우선 식별자로 사용할 수 있다.
- 04번에서 정책분야 변수를 만든 뒤 05번 또는 06B번으로 넘길 때도 `v1_row_id`를 함께 보존해야 한다.

## 04번에서 사용할 핵심 열

{chr(10).join(['- `' + col + '`' for col in existing_core_handoff_cols])}

## 토큰 열 기준

- `tokens_analysis_primary`: 04번 기본 참고 토큰
- `tokens_analysis_second_filter`: 2차 불용어 제거 비교용 토큰
- 두 열의 차이는 지역명 제거 방식이 아니라 2차 불용어 적용 여부이다.

## 주의사항

- 원본 `사업명`은 덮어쓰지 않는다.
- 04번 정책분야 분류는 공식 행정분류 확정값이 아니라 사업명 텍스트 기반 탐색용 분류이다.
- 애매한 사례는 `검토필요`, `기타`, `복합` 등으로 남긴다.
"""

with open(supp_handoff_memo_txt, 'w', encoding='utf-8') as f:
    f.write(handoff_memo)

# ------------------------------------------------------------
# 7) 저장 결과 출력
# ------------------------------------------------------------

print("\n03-SUPP-2 저장 완료")
print("- 새 03번 master parquet:", supp_master_parquet)
print("- 확인용 sample Excel:", supp_sample_xlsx)
print("- columns CSV:", supp_columns_csv)
print("- save_check CSV:", supp_save_check_csv)
print("- 04번 handoff memo:", supp_handoff_memo_txt)

print("\n저장 파일 존재 여부")
print("- supp_master_parquet:", supp_master_parquet.exists())
print("- supp_sample_xlsx:", supp_sample_xlsx.exists())
print("- supp_columns_csv:", supp_columns_csv.exists())
print("- supp_save_check_csv:", supp_save_check_csv.exists())
print("- supp_handoff_memo_txt:", supp_handoff_memo_txt.exists())

print("\n저장 결과 핵심값")
print("- 저장 파일명:", supp_master_parquet.name)
print("- 행 수:", len(df_text_supp))
print("- 열 수:", df_text_supp.shape[1])
print("- v1_row_id 결측 수:", int(df_text_supp['v1_row_id'].isna().sum()))
print("- v1_row_id 중복 수:", int(df_text_supp['v1_row_id'].duplicated().sum()))
print("- v1_row_id 최소값:", int(df_text_supp['v1_row_id'].min()))
print("- v1_row_id 최대값:", int(df_text_supp['v1_row_id'].max()))
print("- 원파일명 존재 여부:", '원파일명' in df_text_supp.columns)
print("- 원파일명 결측 수:", int(df_text_supp['원파일명'].isna().sum()) if '원파일명' in df_text_supp.columns else None)

print("\n04번으로 넘길 핵심 열 목록")
for col in existing_core_handoff_cols:
    print("-", col)

print("\n04번 전달 메모")
print(handoff_memo)
```

03-TOKEN-SUPP-1. 최신 03번 master 불러오기 및 기본 토큰 상태 확인
1. 이 단계의 목적

v1_row_id와 원파일명이 보강된 최신 03번 master를 불러와서, 토큰 보완 작업의 기준 데이터를 확인합니다

입력 데이터
03_project_name_text_analysis_v1_20260529_0135.parquet

```python
# 03-TOKEN-SUPP-1B. 토큰 리스트 타입 보정 및 빈도표 재생성
# 주의: 이 셀은 파일을 저장하지 않습니다.
# 목적: parquet에서 list가 numpy.ndarray로 복원된 경우까지 안전하게 처리합니다.

import pandas as pd
import numpy as np
import ast

# ------------------------------------------------------------
# 1) df_token_base 존재 확인
# ------------------------------------------------------------

if 'df_token_base' not in globals():
    raise NameError(
        "df_token_base가 없습니다. "
        "03-TOKEN-SUPP-1에서 최신 03번 master를 불러온 부분까지 다시 실행하세요."
    )

print("df_token_base 확인")
print("- 행 수:", len(df_token_base))
print("- 열 수:", df_token_base.shape[1])


# ------------------------------------------------------------
# 2) 안전한 토큰 리스트 변환 함수
# ------------------------------------------------------------
# 기존 오류 원인:
# - token 값이 list가 아니라 numpy.ndarray인 경우가 있음
# - 이 상태에서 pd.isna(x)를 호출하면 배열이 반환되어 ValueError 발생
#
# 해결:
# - list, tuple, np.ndarray를 먼저 처리
# - 그 다음에 None / NA / 문자열을 처리

def ensure_token_list_safe(x):
    """
    parquet에서 복원된 토큰 열을 안전하게 list[str] 형태로 변환합니다.
    list, tuple, numpy.ndarray, 문자열 표현 리스트, 결측값을 모두 처리합니다.
    """

    # 1) 이미 list인 경우
    if isinstance(x, list):
        return [str(t).strip() for t in x if str(t).strip() != ""]

    # 2) tuple인 경우
    if isinstance(x, tuple):
        return [str(t).strip() for t in list(x) if str(t).strip() != ""]

    # 3) numpy array인 경우
    if isinstance(x, np.ndarray):
        return [str(t).strip() for t in x.tolist() if str(t).strip() != ""]

    # 4) None인 경우
    if x is None:
        return []

    # 5) pandas NA / NaN 등 scalar 결측 처리
    try:
        if pd.isna(x):
            return []
    except Exception:
        # 배열형 등에서 pd.isna가 애매하면 결측으로 보지 않고 아래 로직으로 넘김
        pass

    # 6) 문자열인 경우
    if isinstance(x, str):
        x_strip = x.strip()

        if x_strip == "":
            return []

        # 문자열이 "['문화', '체육']" 같은 리스트 표현이면 파싱
        try:
            parsed = ast.literal_eval(x_strip)

            if isinstance(parsed, list):
                return [str(t).strip() for t in parsed if str(t).strip() != ""]

            if isinstance(parsed, tuple):
                return [str(t).strip() for t in list(parsed) if str(t).strip() != ""]

            if isinstance(parsed, np.ndarray):
                return [str(t).strip() for t in parsed.tolist() if str(t).strip() != ""]

        except Exception:
            pass

        # 일반 문자열이면 토큰 1개로 처리
        return [x_strip]

    # 7) 그 외 타입은 문자열 1개로 처리
    return [str(x).strip()] if str(x).strip() != "" else []


# ------------------------------------------------------------
# 3) 토큰 열 타입 보정
# ------------------------------------------------------------

token_cols_to_fix = [
    'tokens_analysis_primary',
    'tokens_analysis_second_filter'
]

for token_col in token_cols_to_fix:
    if token_col not in df_token_base.columns:
        raise KeyError(f"{token_col} 열이 df_token_base에 없습니다.")

    df_token_base[token_col] = df_token_base[token_col].apply(ensure_token_list_safe)

print("\n토큰 열 list 변환 완료")

print("\n토큰 열 샘플 타입 확인")
for token_col in token_cols_to_fix:
    first_value = df_token_base[token_col].dropna().iloc[0]
    print(f"- {token_col}: {type(first_value)}, 예시 = {first_value[:10]}")


# ------------------------------------------------------------
# 4) 빈도표 생성 함수
# ------------------------------------------------------------

def make_token_freq(df, token_col):
    token_long = (
        df[[token_col]]
        .explode(token_col)
        .rename(columns={token_col: 'token'})
        .dropna()
        .copy()
    )

    token_long['token'] = token_long['token'].astype(str).str.strip()
    token_long = token_long[token_long['token'] != '']

    token_freq = (
        token_long['token']
        .value_counts()
        .reset_index()
    )

    token_freq.columns = ['token', 'freq']

    token_freq['ratio_percent'] = (
        token_freq['freq'] / token_freq['freq'].sum() * 100
    ).round(4)

    token_freq['token_len'] = token_freq['token'].str.len()

    return token_freq


# ------------------------------------------------------------
# 5) 기존 토큰 빈도표 재생성
# ------------------------------------------------------------

token_freq_primary = make_token_freq(
    df_token_base,
    'tokens_analysis_primary'
)

token_freq_second = make_token_freq(
    df_token_base,
    'tokens_analysis_second_filter'
)

print("\n기존 토큰 빈도표 생성 완료")
print("- token_freq_primary 고유 토큰 수:", len(token_freq_primary))
print("- token_freq_second 고유 토큰 수:", len(token_freq_second))

print("\n[기본 토큰 상위 30개]")
display(token_freq_primary.head(30))


# ------------------------------------------------------------
# 6) 토큰 길이별 요약
# ------------------------------------------------------------

token_len_summary = (
    token_freq_primary
    .groupby('token_len')
    .agg(
        고유토큰수=('token', 'count'),
        전체등장횟수=('freq', 'sum')
    )
    .reset_index()
    .sort_values('token_len')
)

print("\n[토큰 길이별 고유 토큰 수 및 등장 횟수]")
display(token_len_summary)

print("\n03-TOKEN-SUPP-1B 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("다음으로 03-TOKEN-SUPP-2를 실행하면 됩니다.")
```

03-TOKEN-SUPP-2. 한 글자 토큰 및 저정보 후보 사례 검토
1. 이 단계의 목적

06A 네트워크에서 라벨로 노출된 비, 날, 한마음 같은 저정보 토큰 후보를 검토합니다.

특히 다음을 확인합니다.

- 한 글자 토큰 전체 목록
- 한 글자 토큰 빈도
- 한 글자 토큰이 포함된 실제 사업명 사례
- 사전에 언급된 저정보 후보의 빈도와 사례

```python
# 03-TOKEN-SUPP-2. 한 글자 토큰 및 저정보 후보 사례 검토
# 주의: 이 셀은 파일을 저장하지 않습니다.

# ------------------------------------------------------------
# 1) 한 글자 토큰 빈도표
# ------------------------------------------------------------

one_char_token_freq = (
    token_freq_primary[token_freq_primary['token_len'] == 1]
    .copy()
    .sort_values(['freq', 'token'], ascending=[False, True])
    .reset_index(drop=True)
)

print("한 글자 토큰 수:", len(one_char_token_freq))
print("한 글자 토큰 전체 등장 횟수:", int(one_char_token_freq['freq'].sum()))

print("\n[1] 한 글자 토큰 빈도 상위 100개")
display(one_char_token_freq.head(100))

# ------------------------------------------------------------
# 2) 특정 토큰이 포함된 사업명 사례 추출 함수
# ------------------------------------------------------------

def sample_rows_for_token(df, token_col, token, n=10):
    mask = df[token_col].apply(lambda toks: token in toks if isinstance(toks, list) else False)

    sample_cols = [
        'v1_row_id',
        '시도',
        '지자체명',
        '원파일명',
        '사업명',
        '사업명_clean_spacefix',
        token_col,
        '보조사업자',
        '평가결과',
        '평가종류'
    ]
    sample_cols = [col for col in sample_cols if col in df.columns]

    return df.loc[mask, sample_cols].head(n).copy()

# ------------------------------------------------------------
# 3) 요청에서 직접 언급된 저정보 후보 확인
# ------------------------------------------------------------

lowinfo_seed_tokens = [
    '비',
    '날',
    '한마음'
]

lowinfo_seed_rows = []

for tok in lowinfo_seed_tokens:
    freq_row = token_freq_primary[token_freq_primary['token'] == tok]

    if len(freq_row) == 0:
        freq = 0
        ratio = 0
    else:
        freq = int(freq_row['freq'].iloc[0])
        ratio = float(freq_row['ratio_percent'].iloc[0])

    lowinfo_seed_rows.append({
        'token': tok,
        'freq': freq,
        'ratio_percent': ratio,
        'token_len': len(tok)
    })

lowinfo_seed_review = pd.DataFrame(lowinfo_seed_rows)

print("\n[2] 요청 언급 저정보 후보 빈도")
display(lowinfo_seed_review)

print("\n[3] 요청 언급 저정보 후보별 사업명 사례")
for tok in lowinfo_seed_tokens:
    print(f"\n--- token = {tok} ---")
    sample_df = sample_rows_for_token(
        df_token_base,
        'tokens_analysis_primary',
        tok,
        n=15
    )
    display(sample_df)

# ------------------------------------------------------------
# 4) 한 글자 토큰별 사업명 사례표 생성
# ------------------------------------------------------------
# 상위 50개 한 글자 토큰에 대해 대표 사례를 3개씩 확인합니다.

one_char_sample_records = []

for tok in one_char_token_freq['token'].head(50):
    sample_df = sample_rows_for_token(
        df_token_base,
        'tokens_analysis_primary',
        tok,
        n=3
    )

    for _, row in sample_df.iterrows():
        one_char_sample_records.append({
            'token': tok,
            'token_freq': int(
                one_char_token_freq.loc[
                    one_char_token_freq['token'] == tok, 'freq'
                ].iloc[0]
            ),
            'v1_row_id': row.get('v1_row_id', None),
            '시도': row.get('시도', None),
            '지자체명': row.get('지자체명', None),
            '원파일명': row.get('원파일명', None),
            '사업명': row.get('사업명', None),
            '사업명_clean_spacefix': row.get('사업명_clean_spacefix', None),
            'tokens_analysis_primary': row.get('tokens_analysis_primary', None)
        })

one_char_sample_df = pd.DataFrame(one_char_sample_records)

print("\n[4] 한 글자 토큰 상위 50개 사례")
display(one_char_sample_df)

# ------------------------------------------------------------
# 5) 네트워크 라벨에서 문제가 될 수 있는 저정보 후보 초안
# ------------------------------------------------------------
# 이 리스트는 바로 제거 확정이 아니라, 다음 단계에서 검토할 후보입니다.
# 특히 한 글자 토큰은 무조건 제거하지 않고, 실제 사례를 보고 판단해야 합니다.

lowinfo_candidate_initial = sorted(set(
    lowinfo_seed_tokens
    + one_char_token_freq['token'].head(100).tolist()
))

print("\n[5] 저정보 후보 초안")
print(lowinfo_candidate_initial)

print("\n03-TOKEN-SUPP-2 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("다음 단계에서는 위 사례를 보고 실제 제거할 토큰 목록을 확정합니다.")
```

```python
# 03-TOKEN-SUPP-2C. 저정보 토큰 검토용 파일 저장
# 목적: 생략 없이 전체 검토표를 저장해서 공유하기 위한 파일을 만듭니다.
# 주의: 이 파일은 최종 산출물이 아니라 review용 임시 검토표입니다.

from datetime import datetime
from zoneinfo import ZoneInfo

SAVE_REVIEW_OUTPUT = True

timestamp = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M")

review_dir = TABLE_DIR / "review"
review_dir.mkdir(parents=True, exist_ok=True)

lowinfo_review_xlsx = review_dir / f"03_token_lowinfo_review_v1_{timestamp}.xlsx"
lowinfo_review_csv = review_dir / f"03_one_char_token_freq_v1_{timestamp}.csv"

if SAVE_REVIEW_OUTPUT:
    with pd.ExcelWriter(lowinfo_review_xlsx, engine="openpyxl") as writer:
        one_char_token_freq.to_excel(writer, sheet_name="one_char_freq", index=False)
        one_char_sample_df.to_excel(writer, sheet_name="one_char_samples", index=False)
        lowinfo_seed_review.to_excel(writer, sheet_name="seed_lowinfo_freq", index=False)

    one_char_token_freq.to_csv(lowinfo_review_csv, index=False, encoding="utf-8-sig")

    print("저정보 토큰 검토용 파일 저장 완료")
    print("- Excel:", lowinfo_review_xlsx)
    print("- CSV:", lowinfo_review_csv)
    print("- Excel exists:", lowinfo_review_xlsx.exists())
    print("- CSV exists:", lowinfo_review_csv.exists())
else:
    print("SAVE_REVIEW_OUTPUT = False이므로 저장하지 않았습니다.")
```

```python
# 03-TOKEN-SUPP-3. 저정보 토큰 후보 1차 판정표 생성
# 주의: 이 셀은 파일을 저장하지 않습니다.
# 목적: 06A 네트워크용 토큰에서 제거할 후보를 확정하기 전에
#       제거 확정 / 보존 후보 / 추가 검토 후보를 구분합니다.

# ------------------------------------------------------------
# 1) 1차 제거 확정 후보
# ------------------------------------------------------------
# 06A 시각화에서 라벨로 노출되었고, 실제 사례상 정책내용 해석 기여가 낮은 토큰

lowinfo_remove_confirmed = [
    "비",
    "날",
    "한마음"
]

# ------------------------------------------------------------
# 2) 한 글자이지만 정책내용 단서로 보존할 후보
# ------------------------------------------------------------
# 한 글자라고 무조건 제거하면 안 되는 토큰입니다.
# 실제 사업내용 또는 정책분야 단서가 될 수 있어 우선 보존 후보로 둡니다.

one_char_keep_candidates = [
    "쌀",
    "벼",
    "길",
    "숲",
    "물",
    "꽃",
    "집",
    "차",
    "콩",
    "논",
    "밭",
    "섬",
    "책",
    "춤",
    "빛",
    "꿈"
]

# ------------------------------------------------------------
# 3) 추가 검토 후보
# ------------------------------------------------------------
# 복합어가 잘못 쪼개졌거나, 특정 맥락에서는 의미가 있지만
# 네트워크 라벨로는 정보량이 낮을 수 있는 토큰입니다.

one_char_review_candidates = [
    "농",
    "병",
    "배",
    "H",
    "읍",
    "냉",
    "선",
    "밤",
    "홈",
    "팀",
    "달",
    "소",
    "방",
    "체",
    "포",
    "효",
    "반",
    "민",
    "후",
    "금",
    "왕",
    "팜",
    "밖",
    "부",
    "향",
    "연",
    "자",
    "봉",
    "수",
    "성",
    "김",
    "초",
    "과",
    "고",
    "남",
    "묘",
    "유",
    "육",
    "백",
    "천",
    "탑",
    "송",
    "실",
    "창",
    "교",
    "북",
    "학",
    "청",
    "터",
    "업",
    "강",
    "궁",
    "암",
    "액",
    "통"
]

# ------------------------------------------------------------
# 4) 판정표 생성
# ------------------------------------------------------------

def classify_lowinfo_token(token):
    if token in lowinfo_remove_confirmed:
        return "제거확정_1차"
    if token in one_char_keep_candidates:
        return "보존후보"
    if token in one_char_review_candidates:
        return "추가검토"
    return "미검토"

lowinfo_decision_draft = one_char_token_freq.copy()
lowinfo_decision_draft["검토판정_초안"] = lowinfo_decision_draft["token"].apply(classify_lowinfo_token)

# 한마음은 한 글자가 아니므로 별도 추가
hanmaum_row = token_freq_primary[token_freq_primary["token"] == "한마음"].copy()
if len(hanmaum_row) > 0:
    hanmaum_row["검토판정_초안"] = "제거확정_1차"
    lowinfo_decision_draft = pd.concat(
        [lowinfo_decision_draft, hanmaum_row],
        ignore_index=True
    )

# 보기 좋게 정렬
decision_order = {
    "제거확정_1차": 1,
    "추가검토": 2,
    "보존후보": 3,
    "미검토": 4
}

lowinfo_decision_draft["정렬순서"] = lowinfo_decision_draft["검토판정_초안"].map(decision_order).fillna(9)

lowinfo_decision_draft = (
    lowinfo_decision_draft
    .sort_values(["정렬순서", "freq"], ascending=[True, False])
    .drop(columns=["정렬순서"])
    .reset_index(drop=True)
)

print("[1] 저정보 토큰 후보 1차 판정표")
display(lowinfo_decision_draft.head(120))

print("\n[2] 판정별 토큰 수 및 등장 횟수")
decision_summary = (
    lowinfo_decision_draft
    .groupby("검토판정_초안")
    .agg(
        토큰수=("token", "count"),
        전체등장횟수=("freq", "sum")
    )
    .reset_index()
    .sort_values("전체등장횟수", ascending=False)
)

display(decision_summary)

# ------------------------------------------------------------
# 5) 제거확정 후보 사례 재확인
# ------------------------------------------------------------

print("\n[3] 제거확정 후보 사례 재확인")

for tok in lowinfo_remove_confirmed:
    print(f"\n--- token = {tok} ---")
    display(
        sample_rows_for_token(
            df_token_base,
            "tokens_analysis_primary",
            tok,
            n=20
        )
    )

print("\n03-TOKEN-SUPP-3 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("다음 단계에서 제거확정 후보만 먼저 반영한 tokens_analysis_network를 생성할 수 있습니다.")
```

```python
# 03-TOKEN-SUPP-4. 두 글자 이상 저정보 토큰 후보 검토
# 주의: 이 셀은 파일을 저장하지 않습니다.
# 목적: 한 글자 토큰 외에 2글자 이상 저정보 후보를 빈도와 실제 사례 기준으로 검토합니다.

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1) 필수 객체 확인
# ------------------------------------------------------------

required_objects = [
    "df_token_base",
    "token_freq_primary"
]

for obj_name in required_objects:
    if obj_name not in globals():
        raise NameError(f"{obj_name}가 없습니다. 앞 단계 셀을 먼저 실행하세요.")

required_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_primary"
]

missing_cols = [col for col in required_cols if col not in df_token_base.columns]
if missing_cols:
    raise KeyError(f"df_token_base에 필요한 열이 없습니다: {missing_cols}")

print("입력 데이터 확인")
print("- df_token_base 행 수:", len(df_token_base))
print("- token_freq_primary 고유 토큰 수:", len(token_freq_primary))


# ------------------------------------------------------------
# 2) 토큰 long table 생성
# ------------------------------------------------------------
# 한 행의 사업명에 같은 토큰이 반복될 수 있으므로,
# 토큰 빈도 freq와 문서/사업명 단위 등장 행 수를 구분해서 봅니다.

token_long_primary = (
    df_token_base[
        [
            "v1_row_id",
            "시도",
            "지자체명",
            "원파일명",
            "사업명",
            "사업명_clean_spacefix",
            "tokens_analysis_primary"
        ]
    ]
    .explode("tokens_analysis_primary")
    .rename(columns={"tokens_analysis_primary": "token"})
    .dropna()
    .copy()
)

token_long_primary["token"] = token_long_primary["token"].astype(str).str.strip()
token_long_primary = token_long_primary[token_long_primary["token"] != ""].copy()
token_long_primary["token_len"] = token_long_primary["token"].str.len()

# ------------------------------------------------------------
# 3) 토큰별 통계 생성
# ------------------------------------------------------------

token_stats_primary = (
    token_long_primary
    .groupby("token")
    .agg(
        freq=("token", "count"),
        row_count=("v1_row_id", "nunique"),
        sido_count=("시도", "nunique"),
        file_count=("원파일명", "nunique"),
        token_len=("token_len", "first")
    )
    .reset_index()
)

total_rows = df_token_base["v1_row_id"].nunique()

token_stats_primary["row_share_percent"] = (
    token_stats_primary["row_count"] / total_rows * 100
).round(4)

token_stats_primary = token_stats_primary.sort_values(
    ["freq", "row_count"],
    ascending=[False, False]
).reset_index(drop=True)

print("\n[1] 토큰별 통계 상위 50개")
display(token_stats_primary.head(50))


# ------------------------------------------------------------
# 4) 두 글자 이상 저정보 검토 후보 규칙
# ------------------------------------------------------------
# 바로 제거하지 않고, 검토 후보만 만듭니다.
# 고빈도이면서 여러 시도/파일에 넓게 퍼진 일반 표현을 우선 검토합니다.

# 요청 또는 06A 시각화에서 직접 문제로 언급된 후보
seed_lowinfo_multi = [
    "한마음"
]

# 사업 내용 자체라기보다 사업 형식·행사 형식·행정 일반 표현에 가까운 후보
# 단, 모두 제거 확정은 아니며 검토 대상입니다.
generic_review_terms = [
    "행사",
    "대회",
    "운영",
    "지원",
    "사업",
    "활동",
    "기념",
    "기념식",
    "기념행사",
    "추진",
    "화합",
    "어울림",
    "참여",
    "활성",
    "활성화",
    "발전",
    "육성",
    "증진",
    "강화",
    "개선",
    "조성",
    "보급",
    "교육",
    "프로그램",
    "체험",
    "나눔",
    "봉사",
    "교류",
    "워크숍",
    "연찬",
    "연찬회"
]

# 정책내용 단서로 볼 가능성이 커서 자동 제거하면 위험한 보존 후보
# 이 목록은 완전하지 않으며, 단순히 1차 판정 보조용입니다.
substantive_keep_terms = [
    "장애인",
    "노인",
    "아동",
    "청소년",
    "여성",
    "청년",
    "어린이",
    "어르신",
    "농업",
    "농촌",
    "농산물",
    "수산",
    "축산",
    "임업",
    "문화",
    "예술",
    "체육",
    "관광",
    "환경",
    "복지",
    "보훈",
    "의료",
    "보건",
    "안전",
    "재난",
    "주거",
    "교통",
    "마을",
    "경로당",
    "센터",
    "시설"
]

def classify_multi_lowinfo(row):
    token = row["token"]
    freq = row["freq"]
    sido_count = row["sido_count"]
    file_count = row["file_count"]
    token_len = row["token_len"]

    if token in seed_lowinfo_multi:
        return "제거검토_강"

    if token in substantive_keep_terms:
        return "보존후보_정책내용"

    if token in generic_review_terms:
        return "저정보검토_일반표현"

    # 매우 고빈도이고 여러 파일/시도에 넓게 퍼진 2글자 이상 토큰은 검토 후보
    # 기준은 너무 넓게 잡지 않기 위해 다소 보수적으로 설정
    if token_len >= 2 and freq >= 500 and sido_count >= 10 and file_count >= 30:
        return "고빈도검토"

    return "미검토"

multi_char_lowinfo_candidates = token_stats_primary[
    token_stats_primary["token_len"] >= 2
].copy()

multi_char_lowinfo_candidates["검토판정_초안"] = multi_char_lowinfo_candidates.apply(
    classify_multi_lowinfo,
    axis=1
)

# 검토 대상만 우선 보기
multi_char_lowinfo_candidates = multi_char_lowinfo_candidates[
    multi_char_lowinfo_candidates["검토판정_초안"] != "미검토"
].copy()

decision_order = {
    "제거검토_강": 1,
    "저정보검토_일반표현": 2,
    "고빈도검토": 3,
    "보존후보_정책내용": 4
}

multi_char_lowinfo_candidates["정렬순서"] = (
    multi_char_lowinfo_candidates["검토판정_초안"]
    .map(decision_order)
    .fillna(9)
)

multi_char_lowinfo_candidates = (
    multi_char_lowinfo_candidates
    .sort_values(["정렬순서", "freq"], ascending=[True, False])
    .drop(columns=["정렬순서"])
    .reset_index(drop=True)
)

print("\n[2] 두 글자 이상 저정보 후보 검토표")
display(multi_char_lowinfo_candidates.head(150))


# ------------------------------------------------------------
# 5) 판정별 요약
# ------------------------------------------------------------

multi_char_decision_summary = (
    multi_char_lowinfo_candidates
    .groupby("검토판정_초안")
    .agg(
        토큰수=("token", "count"),
        전체등장횟수=("freq", "sum"),
        평균등장행비율=("row_share_percent", "mean")
    )
    .reset_index()
    .sort_values("전체등장횟수", ascending=False)
)

print("\n[3] 두 글자 이상 후보 판정별 요약")
display(multi_char_decision_summary)


# ------------------------------------------------------------
# 6) 후보 토큰별 실제 사업명 사례
# ------------------------------------------------------------

def sample_rows_for_token_safe(df, token_col, token, n=5):
    mask = df[token_col].apply(
        lambda toks: token in toks if isinstance(toks, list) else False
    )

    sample_cols = [
        "v1_row_id",
        "시도",
        "지자체명",
        "원파일명",
        "사업명",
        "사업명_clean_spacefix",
        token_col,
        "보조사업자",
        "평가결과",
        "평가종류"
    ]
    sample_cols = [col for col in sample_cols if col in df.columns]

    return df.loc[mask, sample_cols].head(n).copy()

# 우선 제거검토_강, 저정보검토_일반표현, 고빈도검토 후보만 사례 확인
sample_target_tokens = (
    multi_char_lowinfo_candidates[
        multi_char_lowinfo_candidates["검토판정_초안"].isin(
            ["제거검토_강", "저정보검토_일반표현", "고빈도검토"]
        )
    ]
    .sort_values("freq", ascending=False)
    ["token"]
    .head(40)
    .tolist()
)

multi_char_sample_records = []

for tok in sample_target_tokens:
    stat_row = multi_char_lowinfo_candidates[
        multi_char_lowinfo_candidates["token"] == tok
    ].iloc[0]

    sample_df = sample_rows_for_token_safe(
        df_token_base,
        "tokens_analysis_primary",
        tok,
        n=5
    )

    for _, row in sample_df.iterrows():
        multi_char_sample_records.append({
            "token": tok,
            "검토판정_초안": stat_row["검토판정_초안"],
            "freq": int(stat_row["freq"]),
            "row_count": int(stat_row["row_count"]),
            "sido_count": int(stat_row["sido_count"]),
            "file_count": int(stat_row["file_count"]),
            "v1_row_id": row.get("v1_row_id", None),
            "시도": row.get("시도", None),
            "지자체명": row.get("지자체명", None),
            "원파일명": row.get("원파일명", None),
            "사업명": row.get("사업명", None),
            "사업명_clean_spacefix": row.get("사업명_clean_spacefix", None),
            "tokens_analysis_primary": row.get("tokens_analysis_primary", None)
        })

multi_char_lowinfo_sample_df = pd.DataFrame(multi_char_sample_records)

print("\n[4] 두 글자 이상 저정보 후보 사례")
display(multi_char_lowinfo_sample_df)


# ------------------------------------------------------------
# 7) 현재 단계에서의 1차 제거 제안
# ------------------------------------------------------------
# 여기서는 바로 적용하지 않고, 사람이 확인할 제안만 출력합니다.

network_remove_seed_proposal = [
    "비",
    "날",
    "한마음"
]

print("\n[5] 현재까지의 네트워크용 제거 후보 1차 제안")
print(network_remove_seed_proposal)

print("\n03-TOKEN-SUPP-4 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("위 표를 검토한 뒤, 실제 제거할 두 글자 이상 토큰을 확정합니다.")
```

```python
# 03-TOKEN-SUPP-5. 네트워크용 저정보 토큰 제거 열 생성
# 주의:
# - 기존 tokens_analysis_primary는 덮어쓰지 않습니다.
# - 06A 네트워크 시각화용 새 토큰 열 tokens_analysis_network를 생성합니다.
# - 이 셀은 파일을 저장하지 않습니다.

# ------------------------------------------------------------
# 1) 제거 확정 토큰 목록
# ------------------------------------------------------------

network_lowinfo_remove_tokens = [
    "비",
    "날",
    "한마음"
]

network_lowinfo_remove_set = set(network_lowinfo_remove_tokens)

print("네트워크용 제거 확정 토큰")
print(network_lowinfo_remove_tokens)


# ------------------------------------------------------------
# 2) 네트워크용 토큰 생성 함수
# ------------------------------------------------------------

def make_network_tokens(tokens):
    """
    tokens_analysis_primary에서 저정보 토큰만 제거한
    06A 네트워크용 토큰 리스트를 생성합니다.
    """
    if not isinstance(tokens, list):
        return []

    new_tokens = []
    removed_tokens = []

    for tok in tokens:
        tok = str(tok).strip()

        if tok == "":
            continue

        if tok in network_lowinfo_remove_set:
            removed_tokens.append(tok)
        else:
            new_tokens.append(tok)

    return new_tokens


def get_removed_lowinfo_tokens(tokens):
    """
    각 행에서 제거된 저정보 토큰을 기록합니다.
    """
    if not isinstance(tokens, list):
        return []

    removed = []

    for tok in tokens:
        tok = str(tok).strip()
        if tok in network_lowinfo_remove_set:
            removed.append(tok)

    # 중복 제거하되 원래 순서 유지
    return list(dict.fromkeys(removed))


# ------------------------------------------------------------
# 3) 새 토큰 열 생성
# ------------------------------------------------------------

df_token_supp = df_token_base.copy()

df_token_supp["tokens_analysis_network"] = df_token_supp["tokens_analysis_primary"].apply(
    make_network_tokens
)

df_token_supp["tokens_network_removed_lowinfo"] = df_token_supp["tokens_analysis_primary"].apply(
    get_removed_lowinfo_tokens
)

df_token_supp["tokens_analysis_network_count"] = df_token_supp["tokens_analysis_network"].apply(len)

df_token_supp["tokens_analysis_network_empty"] = df_token_supp["tokens_analysis_network_count"].eq(0)


# ------------------------------------------------------------
# 4) 제거 전후 요약
# ------------------------------------------------------------

before_empty_count = df_token_supp["tokens_analysis_primary"].apply(
    lambda x: len(x) == 0 if isinstance(x, list) else True
).sum()

after_empty_count = int(df_token_supp["tokens_analysis_network_empty"].sum())

removed_row_count = df_token_supp["tokens_network_removed_lowinfo"].apply(
    lambda x: len(x) > 0
).sum()

removed_token_total = df_token_supp["tokens_network_removed_lowinfo"].apply(len).sum()

network_token_supp_summary = pd.DataFrame({
    "항목": [
        "전체 행 수",
        "제거 확정 토큰 수",
        "제거 확정 토큰 목록",
        "저정보 토큰이 제거된 행 수",
        "제거된 저정보 토큰 총 등장 행 기준 수",
        "기존 primary 빈 토큰 행 수",
        "network 토큰 빈 행 수",
        "network 생성 후 추가로 빈 토큰이 된 행 수"
    ],
    "값": [
        len(df_token_supp),
        len(network_lowinfo_remove_tokens),
        ", ".join(network_lowinfo_remove_tokens),
        int(removed_row_count),
        int(removed_token_total),
        int(before_empty_count),
        int(after_empty_count),
        int(after_empty_count - before_empty_count)
    ]
})

print("[1] 네트워크용 토큰 생성 요약")
display(network_token_supp_summary)


# ------------------------------------------------------------
# 5) 제거 토큰별 제거 행 수
# ------------------------------------------------------------

removed_token_records = []

for tok in network_lowinfo_remove_tokens:
    row_count = df_token_supp["tokens_network_removed_lowinfo"].apply(
        lambda xs: tok in xs if isinstance(xs, list) else False
    ).sum()

    freq_row = token_freq_primary[token_freq_primary["token"] == tok]
    freq = int(freq_row["freq"].iloc[0]) if len(freq_row) > 0 else 0

    removed_token_records.append({
        "removed_token": tok,
        "기존_freq": freq,
        "제거발생_행수": int(row_count)
    })

removed_token_summary = pd.DataFrame(removed_token_records)

print("\n[2] 제거 토큰별 요약")
display(removed_token_summary)


# ------------------------------------------------------------
# 6) 제거 전후 사례 확인
# ------------------------------------------------------------

changed_rows = df_token_supp[
    df_token_supp["tokens_network_removed_lowinfo"].apply(lambda x: len(x) > 0)
].copy()

sample_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_primary",
    "tokens_network_removed_lowinfo",
    "tokens_analysis_network",
    "보조사업자",
    "평가결과",
    "평가종류"
]

sample_cols = [col for col in sample_cols if col in changed_rows.columns]

print("\n[3] 저정보 토큰 제거 전후 사례 50개")
display(changed_rows[sample_cols].head(50))


# ------------------------------------------------------------
# 7) network 토큰 빈도표 생성
# ------------------------------------------------------------

token_freq_network = make_token_freq(
    df_token_supp,
    "tokens_analysis_network"
)

print("\n[4] network 토큰 빈도 상위 50개")
display(token_freq_network.head(50))


# ------------------------------------------------------------
# 8) 제거 후보가 network 토큰에 남아 있는지 확인
# ------------------------------------------------------------

remaining_removed_candidates = token_freq_network[
    token_freq_network["token"].isin(network_lowinfo_remove_tokens)
].copy()

print("\n[5] 제거 후보가 network 토큰에 남아 있는지 확인")
display(remaining_removed_candidates)

if len(remaining_removed_candidates) == 0:
    print("제거 확정 토큰은 tokens_analysis_network에서 모두 제거되었습니다.")
else:
    print("주의: 제거 확정 토큰이 아직 남아 있습니다.")

print("\n03-TOKEN-SUPP-5 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("결과가 괜찮으면 다음 단계에서 새 03번 master를 저장합니다.")
```

```python
# 03-TOKEN-SUPP-5B. 한 글자 포함 복합어 재결합 후보 검토
# 주의:
# - 이 셀은 파일을 저장하지 않습니다.
# - 아직 토큰을 수정하지 않습니다.
# - 한 글자 토큰이 앞뒤 토큰과 결합되어야 하는지 검토하기 위한 후보표를 만듭니다.

from collections import Counter, defaultdict
import pandas as pd

# ------------------------------------------------------------
# 1) 입력 확인
# ------------------------------------------------------------

if "df_token_supp" not in globals():
    raise NameError("df_token_supp가 없습니다. 03-TOKEN-SUPP-5를 먼저 실행하세요.")

if "tokens_analysis_network" not in df_token_supp.columns:
    raise KeyError("tokens_analysis_network 열이 없습니다. 03-TOKEN-SUPP-5를 먼저 실행하세요.")

print("입력 데이터 확인")
print("- 행 수:", len(df_token_supp))
print("- tokens_analysis_network 존재:", "tokens_analysis_network" in df_token_supp.columns)

# ------------------------------------------------------------
# 2) 인접 토큰 pair 후보 수집
# ------------------------------------------------------------
# 한 글자 토큰이 포함된 인접 pair를 찾습니다.
# 예: [병, 해충] → 병해충 후보
# 예: [승, 압] → 승압 후보

pair_counter = Counter()
pair_examples = defaultdict(list)

for _, row in df_token_supp.iterrows():
    tokens = row["tokens_analysis_network"]

    if not isinstance(tokens, list) or len(tokens) < 2:
        continue

    for i in range(len(tokens) - 1):
        a = str(tokens[i]).strip()
        b = str(tokens[i + 1]).strip()

        if a == "" or b == "":
            continue

        # 둘 중 하나라도 한 글자인 pair만 검토
        if len(a) == 1 or len(b) == 1:
            pair = (a, b)
            pair_counter[pair] += 1

            if len(pair_examples[pair]) < 5:
                pair_examples[pair].append({
                    "v1_row_id": row.get("v1_row_id", None),
                    "시도": row.get("시도", None),
                    "지자체명": row.get("지자체명", None),
                    "사업명": row.get("사업명", None),
                    "사업명_clean_spacefix": row.get("사업명_clean_spacefix", None),
                    "tokens_analysis_network": tokens
                })

# ------------------------------------------------------------
# 3) 후보표 생성
# ------------------------------------------------------------

pair_rows = []

for pair, freq in pair_counter.items():
    a, b = pair
    joined = a + b

    examples = pair_examples[pair]
    example_names = " || ".join([str(ex["사업명_clean_spacefix"]) for ex in examples[:3]])

    # 실제 사업명 문자열 안에 결합형이 포함되는지 확인
    # 예: pair = 병 + 해충 → joined = 병해충
    joined_appears_count = sum(
        1 for ex in examples
        if joined in str(ex["사업명_clean_spacefix"])
    )

    pair_rows.append({
        "token_a": a,
        "token_b": b,
        "joined_candidate": joined,
        "pair_freq": freq,
        "sample_joined_appears_count": joined_appears_count,
        "example_names": example_names
    })

one_char_pair_candidates = pd.DataFrame(pair_rows)

one_char_pair_candidates = (
    one_char_pair_candidates
    .sort_values(["pair_freq", "joined_candidate"], ascending=[False, True])
    .reset_index(drop=True)
)

print("[1] 한 글자 포함 인접 pair 후보 상위 100개")
display(one_char_pair_candidates.head(100))

# ------------------------------------------------------------
# 4) 실제로 붙어 있을 가능성이 큰 후보만 별도 필터링
# ------------------------------------------------------------
# joined_candidate가 예시 사업명에 실제로 등장하면, 재결합 후보 가능성이 높습니다.

likely_merge_candidates = one_char_pair_candidates[
    one_char_pair_candidates["sample_joined_appears_count"] > 0
].copy()

print("\n[2] 실제 사업명에서 결합형이 확인된 후보")
display(likely_merge_candidates.head(100))

# ------------------------------------------------------------
# 5) 이미 관찰된 주요 의심 후보 직접 확인
# ------------------------------------------------------------

manual_check_pairs = [
    ("병", "해충"),
    ("승", "압"),
    ("호", "국"),
    ("백", "학"),
    ("톤", "백"),
    ("향", "군"),
    ("상", "공인"),
]

manual_rows = []

for a, b in manual_check_pairs:
    pair = (a, b)
    manual_rows.append({
        "token_a": a,
        "token_b": b,
        "joined_candidate": a + b,
        "pair_freq": pair_counter.get(pair, 0),
        "examples": " || ".join([
            str(ex["사업명_clean_spacefix"])
            for ex in pair_examples.get(pair, [])[:5]
        ])
    })

manual_merge_check = pd.DataFrame(manual_rows)

print("\n[3] 수동 확인 후보")
display(manual_merge_check)

print("\n03-TOKEN-SUPP-5B 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("다음 단계에서 재결합할 pair를 확정합니다.")
```

```python
# 03-TOKEN-SUPP-5C. 한 글자 포함 복합어 재결합 후보 전수 검토표 생성
# 주의:
# - 기존 토큰을 수정하지 않습니다.
# - 파일은 review용 Excel만 저장합니다.
# - 이 결과를 보고 실제 merge_pair_rules를 확정합니다.

from collections import Counter, defaultdict
from datetime import datetime
from zoneinfo import ZoneInfo
import pandas as pd
import re
import unicodedata

# ------------------------------------------------------------
# 1) 입력 확인
# ------------------------------------------------------------

if "df_token_supp" not in globals():
    raise NameError("df_token_supp가 없습니다. 03-TOKEN-SUPP-5를 먼저 실행하세요.")

if "tokens_analysis_network" not in df_token_supp.columns:
    raise KeyError("tokens_analysis_network 열이 없습니다. 03-TOKEN-SUPP-5를 먼저 실행하세요.")

print("입력 데이터 확인")
print("- 행 수:", len(df_token_supp))
print("- tokens_analysis_network 존재:", "tokens_analysis_network" in df_token_supp.columns)

# ------------------------------------------------------------
# 2) 텍스트 정규화 함수
# ------------------------------------------------------------
# 실제 사업명에 '냉·난방비', '냉 난방비', '냉-난방비'처럼 표기가 다를 수 있으므로
# 비교용 문자열에서는 공백과 일부 기호를 제거합니다.

def normalize_for_join_check(text):
    if pd.isna(text):
        return ""

    text = str(text)
    text = unicodedata.normalize("NFKC", text)

    # 공백 제거
    text = re.sub(r"\s+", "", text)

    # 결합 확인에 방해되는 주요 기호 제거
    text = re.sub(r"[·ㆍ\-\_/.,:;()\[\]{}<>「」『』'\"“”‘’]", "", text)

    return text

# ------------------------------------------------------------
# 3) 전체 pair 집계
# ------------------------------------------------------------

pair_counter = Counter()
pair_joined_in_name_counter = Counter()
pair_examples = defaultdict(list)

base_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_network"
]

base_cols = [col for col in base_cols if col in df_token_supp.columns]

for _, row in df_token_supp[base_cols].iterrows():
    tokens = row["tokens_analysis_network"]

    if not isinstance(tokens, list) or len(tokens) < 2:
        continue

    name_text = row.get("사업명_clean_spacefix", "")
    name_norm = normalize_for_join_check(name_text)

    for i in range(len(tokens) - 1):
        a = str(tokens[i]).strip()
        b = str(tokens[i + 1]).strip()

        if a == "" or b == "":
            continue

        # 한 글자 토큰이 포함된 인접 pair만 검토
        if len(a) == 1 or len(b) == 1:
            pair = (a, b)
            joined = a + b
            joined_norm = normalize_for_join_check(joined)

            pair_counter[pair] += 1

            # 실제 사업명에 결합형이 확인되는지 전체 데이터 기준으로 카운트
            joined_in_name = joined_norm != "" and joined_norm in name_norm

            if joined_in_name:
                pair_joined_in_name_counter[pair] += 1

            # 대표 사례는 pair당 최대 10개만 저장
            if len(pair_examples[pair]) < 10:
                pair_examples[pair].append({
                    "token_a": a,
                    "token_b": b,
                    "joined_candidate": joined,
                    "joined_in_name": joined_in_name,
                    "v1_row_id": row.get("v1_row_id", None),
                    "시도": row.get("시도", None),
                    "지자체명": row.get("지자체명", None),
                    "원파일명": row.get("원파일명", None),
                    "사업명": row.get("사업명", None),
                    "사업명_clean_spacefix": row.get("사업명_clean_spacefix", None),
                    "tokens_analysis_network": tokens
                })

# ------------------------------------------------------------
# 4) 전체 검토표 생성
# ------------------------------------------------------------

review_rows = []

for pair, freq in pair_counter.items():
    a, b = pair
    joined = a + b

    joined_row_count = pair_joined_in_name_counter.get(pair, 0)
    joined_ratio = joined_row_count / freq if freq > 0 else 0

    examples = pair_examples[pair]

    example_names = " || ".join([
        str(ex["사업명_clean_spacefix"])
        for ex in examples[:5]
    ])

    # 1차 판정 규칙
    # - 실제 결합형 확인 비율이 높고 빈도도 있으면 재결합 후보 강
    # - 빈도는 높지만 결합형 확인 비율이 낮으면 검토 필요
    # - 결합형이 거의 없으면 보류
    if freq >= 30 and joined_ratio >= 0.6:
        decision = "재결합후보_강"
    elif freq >= 10 and joined_ratio >= 0.6:
        decision = "재결합후보_중"
    elif joined_row_count > 0:
        decision = "재결합후보_약"
    else:
        decision = "보류"

    review_rows.append({
        "token_a": a,
        "token_b": b,
        "joined_candidate": joined,
        "token_a_len": len(a),
        "token_b_len": len(b),
        "pair_freq": freq,
        "joined_in_name_row_count": joined_row_count,
        "joined_in_name_ratio": round(joined_ratio, 4),
        "검토판정_초안": decision,
        "example_names": example_names
    })

one_char_pair_full_review = pd.DataFrame(review_rows)

decision_order = {
    "재결합후보_강": 1,
    "재결합후보_중": 2,
    "재결합후보_약": 3,
    "보류": 4
}

one_char_pair_full_review["정렬순서"] = (
    one_char_pair_full_review["검토판정_초안"]
    .map(decision_order)
    .fillna(9)
)

one_char_pair_full_review = (
    one_char_pair_full_review
    .sort_values(
        ["정렬순서", "pair_freq", "joined_in_name_ratio"],
        ascending=[True, False, False]
    )
    .drop(columns=["정렬순서"])
    .reset_index(drop=True)
)

print("[1] 한 글자 포함 pair 전수 검토표 상위 150개")
display(one_char_pair_full_review.head(150))

# ------------------------------------------------------------
# 5) 대표 사례 long table 생성
# ------------------------------------------------------------

example_rows = []

for pair, examples in pair_examples.items():
    a, b = pair
    joined = a + b

    freq = pair_counter[pair]
    joined_count = pair_joined_in_name_counter.get(pair, 0)
    joined_ratio = joined_count / freq if freq > 0 else 0

    for ex in examples:
        example_rows.append({
            "token_a": a,
            "token_b": b,
            "joined_candidate": joined,
            "pair_freq": freq,
            "joined_in_name_row_count": joined_count,
            "joined_in_name_ratio": round(joined_ratio, 4),
            "joined_in_name": ex["joined_in_name"],
            "v1_row_id": ex["v1_row_id"],
            "시도": ex["시도"],
            "지자체명": ex["지자체명"],
            "원파일명": ex["원파일명"],
            "사업명": ex["사업명"],
            "사업명_clean_spacefix": ex["사업명_clean_spacefix"],
            "tokens_analysis_network": ex["tokens_analysis_network"]
        })

one_char_pair_full_examples = pd.DataFrame(example_rows)

print("\n[2] 재결합 후보 판정별 요약")
merge_candidate_summary = (
    one_char_pair_full_review
    .groupby("검토판정_초안")
    .agg(
        후보수=("joined_candidate", "count"),
        pair_freq합계=("pair_freq", "sum"),
        평균결합형확인비율=("joined_in_name_ratio", "mean")
    )
    .reset_index()
    .sort_values("pair_freq합계", ascending=False)
)

display(merge_candidate_summary)

print("\n[3] 재결합후보_강 상위 50개")
display(
    one_char_pair_full_review[
        one_char_pair_full_review["검토판정_초안"] == "재결합후보_강"
    ].head(50)
)

# ------------------------------------------------------------
# 6) 검토용 파일 저장
# ------------------------------------------------------------

SAVE_REVIEW_OUTPUT = True

RUN_TS = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M")

if "TABLE_DIR" in globals():
    review_dir = TABLE_DIR / "review"
else:
    review_dir = FINAL_DIR.parent / "review"

review_dir.mkdir(parents=True, exist_ok=True)

out_review_xlsx = review_dir / f"03_token_merge_pair_full_review_v1_{RUN_TS}.xlsx"

if SAVE_REVIEW_OUTPUT:
    with pd.ExcelWriter(out_review_xlsx, engine="openpyxl") as writer:
        one_char_pair_full_review.to_excel(
            writer,
            sheet_name="pair_full_review",
            index=False
        )
        one_char_pair_full_examples.to_excel(
            writer,
            sheet_name="pair_examples",
            index=False
        )
        merge_candidate_summary.to_excel(
            writer,
            sheet_name="summary",
            index=False
        )

    print("\n검토용 Excel 저장 완료")
    print("- Excel:", out_review_xlsx)
    print("- exists:", out_review_xlsx.exists())
else:
    print("\nSAVE_REVIEW_OUTPUT = False이므로 저장하지 않았습니다.")

print("\n03-TOKEN-SUPP-5C 완료")
print("이 결과를 보고 실제 merge_pair_rules를 확정합니다.")
```

```python
# 03-TOKEN-SUPP-5D. 보수적 복합어 재결합 규칙 적용
# 주의:
# - 기존 tokens_analysis_primary, tokens_analysis_network는 덮어쓰지 않습니다.
# - 사람이 확인한 보수적 merge_pair_rules만 적용합니다.
# - 이 셀은 파일을 저장하지 않습니다.

import pandas as pd

# ------------------------------------------------------------
# 1) 보수적 재결합 규칙
# ------------------------------------------------------------

merge_pair_rules = {
    ("병", "해충"): "병해충",
    ("냉", "난방비"): "냉난방비",
    ("선", "진지"): "선진지",
    ("농", "식품"): "농식품",
    ("농", "작업"): "농작업",
    ("그룹", "홈"): "그룹홈",
    ("스마트", "팜"): "스마트팜",
    ("방과", "후"): "방과후",
    ("톤", "백"): "톤백",
    ("빨래", "방"): "빨래방",
    ("육", "묘"): "육묘",
    ("융", "복합"): "융복합",
    ("직", "매장"): "직매장",
    ("3.1", "절"): "3.1절",
}

print("적용할 재결합 규칙 수:", len(merge_pair_rules))

display(
    pd.DataFrame([
        {"token_a": k[0], "token_b": k[1], "merged_token": v}
        for k, v in merge_pair_rules.items()
    ])
)

# ------------------------------------------------------------
# 2) 재결합 함수
# ------------------------------------------------------------

def merge_token_pairs_with_log(tokens, merge_rules):
    """
    tokens 리스트에서 인접한 pair가 merge_rules에 있으면 병합합니다.
    병합된 pair 기록도 함께 반환합니다.
    """
    if not isinstance(tokens, list):
        return [], []

    merged_tokens = []
    merge_logs = []
    i = 0

    while i < len(tokens):
        if i < len(tokens) - 1:
            a = str(tokens[i]).strip()
            b = str(tokens[i + 1]).strip()
            pair = (a, b)

            if pair in merge_rules:
                merged_token = merge_rules[pair]
                merged_tokens.append(merged_token)
                merge_logs.append(f"{a}+{b}->{merged_token}")
                i += 2
                continue

        tok = str(tokens[i]).strip()
        if tok != "":
            merged_tokens.append(tok)
        i += 1

    return merged_tokens, merge_logs

# ------------------------------------------------------------
# 3) 입력 확인 및 새 final 네트워크 토큰 열 생성
# ------------------------------------------------------------

if "df_token_supp" not in globals():
    raise NameError("df_token_supp가 없습니다. 03-TOKEN-SUPP-5를 먼저 실행하세요.")

if "tokens_analysis_network" not in df_token_supp.columns:
    raise KeyError("tokens_analysis_network 열이 없습니다.")

df_token_supp = df_token_supp.copy()

merge_results = df_token_supp["tokens_analysis_network"].apply(
    lambda toks: merge_token_pairs_with_log(toks, merge_pair_rules)
)

df_token_supp["tokens_analysis_network_final"] = merge_results.apply(lambda x: x[0])
df_token_supp["tokens_network_merged_pairs"] = merge_results.apply(lambda x: x[1])
df_token_supp["tokens_analysis_network_final_count"] = df_token_supp["tokens_analysis_network_final"].apply(len)
df_token_supp["tokens_analysis_network_final_empty"] = df_token_supp["tokens_analysis_network_final_count"].eq(0)

# ------------------------------------------------------------
# 4) 재결합 결과 요약
# ------------------------------------------------------------

merged_row_count = df_token_supp["tokens_network_merged_pairs"].apply(
    lambda x: len(x) > 0 if isinstance(x, list) else False
).sum()

merged_pair_total = df_token_supp["tokens_network_merged_pairs"].apply(
    lambda x: len(x) if isinstance(x, list) else 0
).sum()

before_empty = int(df_token_supp["tokens_analysis_network_empty"].sum())
after_empty = int(df_token_supp["tokens_analysis_network_final_empty"].sum())

merge_summary = pd.DataFrame({
    "항목": [
        "전체 행 수",
        "적용한 재결합 규칙 수",
        "재결합 발생 행 수",
        "재결합 발생 총 횟수",
        "network 토큰 빈 행 수",
        "network_final 토큰 빈 행 수",
        "재결합 후 추가 빈 토큰 행 수"
    ],
    "값": [
        len(df_token_supp),
        len(merge_pair_rules),
        int(merged_row_count),
        int(merged_pair_total),
        before_empty,
        after_empty,
        after_empty - before_empty
    ]
})

print("[1] 재결합 결과 요약")
display(merge_summary)

# ------------------------------------------------------------
# 5) 재결합 규칙별 발생 횟수
# ------------------------------------------------------------

merge_log_long = (
    df_token_supp[["tokens_network_merged_pairs"]]
    .explode("tokens_network_merged_pairs")
    .dropna()
    .rename(columns={"tokens_network_merged_pairs": "merge_rule"})
)

if len(merge_log_long) > 0:
    merge_rule_summary = (
        merge_log_long["merge_rule"]
        .value_counts()
        .reset_index()
    )
    merge_rule_summary.columns = ["merge_rule", "freq"]
else:
    merge_rule_summary = pd.DataFrame(columns=["merge_rule", "freq"])

print("\n[2] 재결합 규칙별 발생 횟수")
display(merge_rule_summary)

# ------------------------------------------------------------
# 6) 재결합 전후 사례 확인
# ------------------------------------------------------------

merged_cases = df_token_supp[
    df_token_supp["tokens_network_merged_pairs"].apply(
        lambda x: len(x) > 0 if isinstance(x, list) else False
    )
].copy()

sample_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_network",
    "tokens_network_merged_pairs",
    "tokens_analysis_network_final",
    "보조사업자",
    "평가결과",
    "평가종류"
]

sample_cols = [col for col in sample_cols if col in merged_cases.columns]

print("\n[3] 재결합 전후 사례 100개")
display(merged_cases[sample_cols].head(100))

# ------------------------------------------------------------
# 7) final 네트워크 토큰 빈도 확인
# ------------------------------------------------------------

token_freq_network_final = make_token_freq(
    df_token_supp,
    "tokens_analysis_network_final"
)

print("\n[4] network_final 토큰 빈도 상위 50개")
display(token_freq_network_final.head(50))

# ------------------------------------------------------------
# 8) 재결합된 토큰들이 실제로 생겼는지 확인
# ------------------------------------------------------------

merged_token_check = token_freq_network_final[
    token_freq_network_final["token"].isin(set(merge_pair_rules.values()))
].copy()

print("\n[5] 재결합 결과 토큰 빈도")
display(merged_token_check.sort_values("freq", ascending=False))

print("\n03-TOKEN-SUPP-5D 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("결과가 괜찮으면 저장 단계에서 tokens_analysis_network_final을 06A 기본 토큰으로 넘깁니다.")
```

```python
# 03-TOKEN-SUPP-6. 네트워크용 최종 토큰 보완 master 저장
# 목적:
# - 기존 03번 master를 덮어쓰지 않고,
# - 저정보 토큰 제거 + 보수적 복합어 재결합까지 반영한
#   06A 네트워크용 최종 토큰 열을 포함한 새 master 파일을 저장합니다.
#
# 주의:
# - tokens_analysis_primary, tokens_analysis_second_filter는 그대로 보존합니다.
# - 06A에서는 tokens_analysis_network_final을 기본 토큰 열로 사용합니다.
# - 기존 04B, 05, 06B 결과를 소급 변경하는 작업이 아닙니다.

from datetime import datetime
from zoneinfo import ZoneInfo
from pathlib import Path
import pandas as pd

# ------------------------------------------------------------
# 1) 저장 여부 설정
# ------------------------------------------------------------

SAVE_OUTPUT = True

RUN_TS = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M")

print("RUN_TS:", RUN_TS)
print("SAVE_OUTPUT:", SAVE_OUTPUT)

# ------------------------------------------------------------
# 2) 저장 폴더 확인
# ------------------------------------------------------------

if "FINAL_DIR" not in globals():
    raise NameError(
        "FINAL_DIR이 없습니다. 03-TOKEN-SUPP-1에서 최신 03번 master를 불러온 뒤 다시 실행하세요."
    )

FINAL_DIR.mkdir(parents=True, exist_ok=True)

print("\n저장 폴더:")
print(FINAL_DIR)

# ------------------------------------------------------------
# 3) 저장 전 필수 객체 및 열 확인
# ------------------------------------------------------------

if "df_token_supp" not in globals():
    raise NameError("df_token_supp가 없습니다. 03-TOKEN-SUPP-5D까지 먼저 실행하세요.")

required_save_cols = [
    "v1_row_id",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_primary",
    "tokens_analysis_second_filter",
    "tokens_analysis_network",
    "tokens_network_removed_lowinfo",
    "tokens_analysis_network_count",
    "tokens_analysis_network_empty",
    "tokens_analysis_network_final",
    "tokens_network_merged_pairs",
    "tokens_analysis_network_final_count",
    "tokens_analysis_network_final_empty"
]

missing_save_cols = [col for col in required_save_cols if col not in df_token_supp.columns]

if missing_save_cols:
    raise KeyError(f"저장 전 필수 열이 없습니다: {missing_save_cols}")

print("\n저장 전 필수 열 확인 완료")

# ------------------------------------------------------------
# 4) 저장 파일 경로
# ------------------------------------------------------------

out_parquet = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_v1_{RUN_TS}.parquet"
out_sample_xlsx = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_sample_v1_{RUN_TS}.xlsx"
out_columns_csv = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_columns_v1_{RUN_TS}.csv"
out_save_check_csv = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_save_check_v1_{RUN_TS}.csv"
out_handoff_txt = FINAL_DIR / f"03_handoff_memo_for_06A_network_tokens_v1_{RUN_TS}.txt"

# ------------------------------------------------------------
# 5) 저장 전 주요 수치 계산
# ------------------------------------------------------------

primary_empty_count = int(
    df_token_supp["tokens_analysis_primary"].apply(
        lambda x: len(x) == 0 if isinstance(x, list) else True
    ).sum()
)

network_empty_count = int(df_token_supp["tokens_analysis_network_empty"].sum())
network_final_empty_count = int(df_token_supp["tokens_analysis_network_final_empty"].sum())

lowinfo_removed_row_count = int(
    df_token_supp["tokens_network_removed_lowinfo"].apply(
        lambda x: len(x) > 0 if isinstance(x, list) else False
    ).sum()
)

lowinfo_removed_total_count = int(
    df_token_supp["tokens_network_removed_lowinfo"].apply(
        lambda x: len(x) if isinstance(x, list) else 0
    ).sum()
)

merged_row_count = int(
    df_token_supp["tokens_network_merged_pairs"].apply(
        lambda x: len(x) > 0 if isinstance(x, list) else False
    ).sum()
)

merged_total_count = int(
    df_token_supp["tokens_network_merged_pairs"].apply(
        lambda x: len(x) if isinstance(x, list) else 0
    ).sum()
)

network_lowinfo_remove_tokens_str = ", ".join(network_lowinfo_remove_tokens) if "network_lowinfo_remove_tokens" in globals() else ""

merge_pair_rules_str = "; ".join([
    f"{a}+{b}->{merged}"
    for (a, b), merged in merge_pair_rules.items()
]) if "merge_pair_rules" in globals() else ""

# ------------------------------------------------------------
# 6) 저장 전 점검표 생성
# ------------------------------------------------------------

save_check = pd.DataFrame({
    "항목": [
        "저장 데이터 행 수",
        "저장 데이터 열 수",
        "v1_row_id 결측 수",
        "v1_row_id 중복 수",
        "v1_row_id 최소값",
        "v1_row_id 최대값",
        "원파일명 존재 여부",
        "원파일명 결측 수",
        "tokens_analysis_primary 존재 여부",
        "tokens_analysis_second_filter 존재 여부",
        "tokens_analysis_network 존재 여부",
        "tokens_analysis_network_final 존재 여부",
        "primary 빈 토큰 행 수",
        "network 빈 토큰 행 수",
        "network_final 빈 토큰 행 수",
        "저정보 토큰 제거 후 추가 빈 토큰 행 수",
        "재결합 후 추가 빈 토큰 행 수",
        "저정보 토큰 제거 행 수",
        "저정보 토큰 제거 총 횟수",
        "재결합 발생 행 수",
        "재결합 발생 총 횟수",
        "제거 확정 토큰 목록",
        "재결합 규칙 수",
        "재결합 규칙 목록"
    ],
    "값": [
        len(df_token_supp),
        df_token_supp.shape[1],
        int(df_token_supp["v1_row_id"].isna().sum()),
        int(df_token_supp["v1_row_id"].duplicated().sum()),
        int(df_token_supp["v1_row_id"].min()),
        int(df_token_supp["v1_row_id"].max()),
        "원파일명" in df_token_supp.columns,
        int(df_token_supp["원파일명"].isna().sum()),
        "tokens_analysis_primary" in df_token_supp.columns,
        "tokens_analysis_second_filter" in df_token_supp.columns,
        "tokens_analysis_network" in df_token_supp.columns,
        "tokens_analysis_network_final" in df_token_supp.columns,
        primary_empty_count,
        network_empty_count,
        network_final_empty_count,
        network_empty_count - primary_empty_count,
        network_final_empty_count - network_empty_count,
        lowinfo_removed_row_count,
        lowinfo_removed_total_count,
        merged_row_count,
        merged_total_count,
        network_lowinfo_remove_tokens_str,
        len(merge_pair_rules) if "merge_pair_rules" in globals() else 0,
        merge_pair_rules_str
    ]
})

print("[1] 저장 전 점검표")
display(save_check)

# ------------------------------------------------------------
# 7) 06A handoff memo 생성
# ------------------------------------------------------------

handoff_memo = f"""
# 03번 토큰 보완 결과 06A 전달 메모

## 작업 목적
06A 단어 동시출현 네트워크 HTML 시각화에서 `비`, `날`, `한마음`처럼 정책내용 해석에 크게 기여하지 않는 저정보 토큰이 라벨로 노출되어, 03번 토큰 정제 단계에서 네트워크용 보완 토큰 열을 추가했습니다.

또한 한 글자 토큰 중 일부가 복합어에서 잘못 분리된 사례가 확인되어, 06A 네트워크용 토큰에 한해 보수적 복합어 재결합 규칙을 적용했습니다.

## 기준 입력 파일
03_project_name_text_analysis_v1_20260529_0135.parquet

## 새 저장 파일
03_project_name_text_analysis_network_tokens_v1_{RUN_TS}.parquet

## 기존 토큰 열 보존
기존 토큰 열은 덮어쓰지 않았습니다.

보존한 기존 열:
- tokens_analysis_primary
- tokens_analysis_second_filter

## 새로 추가한 네트워크용 열
- tokens_analysis_network
- tokens_network_removed_lowinfo
- tokens_analysis_network_count
- tokens_analysis_network_empty
- tokens_analysis_network_final
- tokens_network_merged_pairs
- tokens_analysis_network_final_count
- tokens_analysis_network_final_empty

## 제거한 저정보 토큰
- {network_lowinfo_remove_tokens_str}

## 재결합한 복합어 규칙
{merge_pair_rules_str}

## 주요 점검 결과
- 전체 행 수: {len(df_token_supp):,}행
- primary 빈 토큰 행 수: {primary_empty_count:,}행
- network 빈 토큰 행 수: {network_empty_count:,}행
- network_final 빈 토큰 행 수: {network_final_empty_count:,}행
- 저정보 토큰 제거 후 추가 빈 토큰 행 수: {network_empty_count - primary_empty_count:,}행
- 재결합 후 추가 빈 토큰 행 수: {network_final_empty_count - network_empty_count:,}행
- 저정보 토큰 제거 행 수: {lowinfo_removed_row_count:,}행
- 저정보 토큰 제거 총 횟수: {lowinfo_removed_total_count:,}회
- 재결합 발생 행 수: {merged_row_count:,}행
- 재결합 발생 총 횟수: {merged_total_count:,}회

## 06A 사용 기준
06A 단어 동시출현 네트워크를 재생성할 때는 기존 `tokens_analysis_primary` 대신 `tokens_analysis_network_final`을 기본 토큰 열로 사용합니다.

열 구분:
- `tokens_analysis_primary`: 기존 03번 기본 분석 토큰
- `tokens_analysis_network`: 저정보 토큰 제거만 반영한 중간 열
- `tokens_analysis_network_final`: 저정보 토큰 제거와 보수적 복합어 재결합을 모두 반영한 06A 최종 네트워크용 토큰 열

## 해석 주의
이번 보완은 06A 네트워크 시각화 품질 개선을 위한 것입니다. 04B, 05, 06B에서 이미 사용한 기존 정책분야 분석 기준을 소급 변경하는 작업은 아닙니다.

`대회`, `행사`, `교육`, `활동`, `개선`, `조성` 등 일반적 토큰은 현재 단계에서 제거하지 않았습니다. 이들은 정보량이 낮아 보일 수 있으나 사업 수행 방식이나 사업 유형을 나타낼 수 있으므로, 06A 재시각화 결과를 확인한 뒤 필요 시 2차 보완 여부를 판단합니다.
""".strip()

print("\n[2] 06A handoff memo 미리보기")
print(handoff_memo)

# ------------------------------------------------------------
# 8) 저장 실행
# ------------------------------------------------------------

if SAVE_OUTPUT:
    df_token_supp.to_parquet(out_parquet, index=False)

    sample_cols = [
        "v1_row_id",
        "시도",
        "지자체명",
        "원파일명",
        "사업명",
        "사업명_clean_spacefix",
        "tokens_analysis_primary",
        "tokens_analysis_second_filter",
        "tokens_network_removed_lowinfo",
        "tokens_analysis_network",
        "tokens_network_merged_pairs",
        "tokens_analysis_network_final",
        "보조사업자",
        "평가결과",
        "평가종류"
    ]
    sample_cols = [col for col in sample_cols if col in df_token_supp.columns]

    df_token_supp[sample_cols].head(1000).to_excel(out_sample_xlsx, index=False)

    pd.DataFrame({
        "column_name": df_token_supp.columns,
        "dtype": [str(df_token_supp[col].dtype) for col in df_token_supp.columns]
    }).to_csv(out_columns_csv, index=False, encoding="utf-8-sig")

    save_check.to_csv(out_save_check_csv, index=False, encoding="utf-8-sig")

    out_handoff_txt.write_text(handoff_memo, encoding="utf-8")

    print("\n저장 완료")
    print("- parquet:", out_parquet)
    print("- sample xlsx:", out_sample_xlsx)
    print("- columns csv:", out_columns_csv)
    print("- save check csv:", out_save_check_csv)
    print("- handoff memo:", out_handoff_txt)

else:
    print("\nSAVE_OUTPUT = False이므로 저장하지 않았습니다.")

# ------------------------------------------------------------
# 9) 저장 파일 존재 여부 확인
# ------------------------------------------------------------

saved_files = [
    out_parquet,
    out_sample_xlsx,
    out_columns_csv,
    out_save_check_csv,
    out_handoff_txt
]

saved_file_check = pd.DataFrame({
    "파일": [p.name for p in saved_files],
    "경로": [str(p) for p in saved_files],
    "존재여부": [p.exists() for p in saved_files]
})

print("\n[3] 저장 파일 존재 여부")
display(saved_file_check)

print("\n03-TOKEN-SUPP-6 완료")
print("이제 06A에서는 tokens_analysis_network_final을 기본 토큰 열로 사용하면 됩니다.")
```

03-TOKEN-SUPP-6B. 저장된 네트워크 토큰 master 다시 불러오기

```python
# 03-TOKEN-SUPP-6B. 저장된 네트워크 토큰 보완 master 다시 불러오기
# 목적:
# - 03-TOKEN-SUPP-6까지 다시 실행하지 않고,
# - 저장 완료된 최신 네트워크용 03번 master를 불러옵니다.
# - 이후 03-TOKEN-SUPP-7~9에서 글자 수 무관 저정보 토큰을 검토합니다.

from pathlib import Path
import pandas as pd
import ast
import numpy as np

# ------------------------------------------------------------
# 1) 경로 설정
# ------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/파공장 (PA Factory)/2026 KAPA/[guide & output] 분석 협업 매뉴얼 및 Figure 후보")

TABLE_DIR = BASE_DIR / "03_중간결과표" / "03_v1_사업명텍스트분석" / "tables"
FINAL_DIR = TABLE_DIR / "final"
REVIEW_DIR = TABLE_DIR / "review"

NETWORK_MASTER_FILE = FINAL_DIR / "03_project_name_text_analysis_network_tokens_v1_20260607_1810.parquet"

print("불러올 파일:")
print(NETWORK_MASTER_FILE)
print("exists:", NETWORK_MASTER_FILE.exists())

if not NETWORK_MASTER_FILE.exists():
    raise FileNotFoundError(f"파일이 없습니다: {NETWORK_MASTER_FILE}")

# ------------------------------------------------------------
# 2) 파일 불러오기
# ------------------------------------------------------------

df_token_supp = pd.read_parquet(NETWORK_MASTER_FILE)

print("\n불러오기 완료")
print("- 행 수:", len(df_token_supp))
print("- 열 수:", df_token_supp.shape[1])

# ------------------------------------------------------------
# 3) 토큰 열 타입 보정 함수
# ------------------------------------------------------------
# parquet에서 리스트가 ndarray나 문자열 형태로 복원되는 경우를 방지합니다.

def ensure_token_list(x):
    if isinstance(x, list):
        return [str(t).strip() for t in x if str(t).strip() != ""]

    if isinstance(x, tuple):
        return [str(t).strip() for t in list(x) if str(t).strip() != ""]

    if isinstance(x, np.ndarray):
        return [str(t).strip() for t in x.tolist() if str(t).strip() != ""]

    if x is None:
        return []

    try:
        if pd.isna(x):
            return []
    except Exception:
        pass

    if isinstance(x, str):
        x_strip = x.strip()

        if x_strip == "":
            return []

        try:
            parsed = ast.literal_eval(x_strip)

            if isinstance(parsed, list):
                return [str(t).strip() for t in parsed if str(t).strip() != ""]

            if isinstance(parsed, tuple):
                return [str(t).strip() for t in list(parsed) if str(t).strip() != ""]

            if isinstance(parsed, np.ndarray):
                return [str(t).strip() for t in parsed.tolist() if str(t).strip() != ""]

        except Exception:
            pass

        return [x_strip]

    return [str(x).strip()] if str(x).strip() != "" else []

# ------------------------------------------------------------
# 4) 주요 토큰 열 보정
# ------------------------------------------------------------

token_cols_to_fix = [
    "tokens_analysis_primary",
    "tokens_analysis_second_filter",
    "tokens_analysis_network",
    "tokens_network_removed_lowinfo",
    "tokens_analysis_network_final",
    "tokens_network_merged_pairs"
]

for col in token_cols_to_fix:
    if col in df_token_supp.columns:
        df_token_supp[col] = df_token_supp[col].apply(ensure_token_list)

# ------------------------------------------------------------
# 5) 필수 열 확인
# ------------------------------------------------------------

required_cols = [
    "v1_row_id",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_primary",
    "tokens_analysis_network",
    "tokens_analysis_network_final"
]

missing_cols = [col for col in required_cols if col not in df_token_supp.columns]

if missing_cols:
    raise KeyError(f"필수 열이 없습니다: {missing_cols}")

# ------------------------------------------------------------
# 6) 간단 점검
# ------------------------------------------------------------

check_summary = pd.DataFrame({
    "항목": [
        "행 수",
        "열 수",
        "v1_row_id 결측 수",
        "v1_row_id 중복 수",
        "원파일명 결측 수",
        "tokens_analysis_primary 빈 토큰 행 수",
        "tokens_analysis_network 빈 토큰 행 수",
        "tokens_analysis_network_final 빈 토큰 행 수"
    ],
    "값": [
        len(df_token_supp),
        df_token_supp.shape[1],
        int(df_token_supp["v1_row_id"].isna().sum()),
        int(df_token_supp["v1_row_id"].duplicated().sum()),
        int(df_token_supp["원파일명"].isna().sum()),
        int(df_token_supp["tokens_analysis_primary"].apply(len).eq(0).sum()),
        int(df_token_supp["tokens_analysis_network"].apply(len).eq(0).sum()),
        int(df_token_supp["tokens_analysis_network_final"].apply(len).eq(0).sum())
    ]
})

display(check_summary)

print("\n03-TOKEN-SUPP-6B 완료")
print("이제 03-TOKEN-SUPP-7부터 실행하면 됩니다.")
```

```python
# 03-TOKEN-SUPP-7. 글자 수 무관 저정보 토큰 후보 점검표 생성
# 주의:
# - 이 셀은 토큰을 수정하지 않습니다.
# - tokens_analysis_network_final 기준으로 저정보 후보 검토표만 만듭니다.
# - 사람이 직접 제거 여부를 판단하기 위한 진단 셀입니다.

from collections import defaultdict
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1) 입력 확인
# ------------------------------------------------------------

if "df_token_supp" not in globals():
    raise NameError("df_token_supp가 없습니다. 03-TOKEN-SUPP-6 이전 셀들을 먼저 실행하세요.")

TOKEN_COL_FOR_REVIEW = "tokens_analysis_network_final"

if TOKEN_COL_FOR_REVIEW not in df_token_supp.columns:
    raise KeyError(f"{TOKEN_COL_FOR_REVIEW} 열이 없습니다.")

print("입력 데이터 확인")
print("- 행 수:", len(df_token_supp))
print("- 검토 기준 토큰 열:", TOKEN_COL_FOR_REVIEW)

# ------------------------------------------------------------
# 2) 토큰 long table 생성
# ------------------------------------------------------------

base_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    TOKEN_COL_FOR_REVIEW
]

base_cols = [col for col in base_cols if col in df_token_supp.columns]

token_long_final = (
    df_token_supp[base_cols]
    .explode(TOKEN_COL_FOR_REVIEW)
    .rename(columns={TOKEN_COL_FOR_REVIEW: "token"})
    .dropna()
    .copy()
)

token_long_final["token"] = token_long_final["token"].astype(str).str.strip()
token_long_final = token_long_final[token_long_final["token"] != ""].copy()
token_long_final["token_len"] = token_long_final["token"].str.len()

n_rows = df_token_supp["v1_row_id"].nunique()
n_files = df_token_supp["원파일명"].nunique() if "원파일명" in df_token_supp.columns else np.nan
n_sido = df_token_supp["시도"].nunique() if "시도" in df_token_supp.columns else np.nan
n_jichache = df_token_supp["지자체명"].nunique() if "지자체명" in df_token_supp.columns else np.nan

# ------------------------------------------------------------
# 3) 토큰별 기본 통계
# ------------------------------------------------------------

token_quality_stats = (
    token_long_final
    .groupby("token")
    .agg(
        freq=("token", "count"),
        row_count=("v1_row_id", "nunique"),
        sido_count=("시도", "nunique"),
        jichache_count=("지자체명", "nunique"),
        file_count=("원파일명", "nunique"),
        token_len=("token_len", "first")
    )
    .reset_index()
)

token_quality_stats["row_share_percent"] = (
    token_quality_stats["row_count"] / n_rows * 100
).round(4)

token_quality_stats["file_share_percent"] = (
    token_quality_stats["file_count"] / n_files * 100
).round(4)

token_quality_stats["sido_share_percent"] = (
    token_quality_stats["sido_count"] / n_sido * 100
).round(4)

token_quality_stats["jichache_share_percent"] = (
    token_quality_stats["jichache_count"] / n_jichache * 100
).round(4)

# ------------------------------------------------------------
# 4) co-occurrence breadth 계산
# ------------------------------------------------------------
# 한 토큰이 몇 종류의 다른 토큰과 함께 등장하는지 확인합니다.
# 너무 많은 토큰과 연결되는 단어는 네트워크에서 범용 허브가 될 수 있습니다.

cooc_sets = defaultdict(set)

for _, row in df_token_supp[["v1_row_id", TOKEN_COL_FOR_REVIEW]].iterrows():
    toks = row[TOKEN_COL_FOR_REVIEW]

    if not isinstance(toks, list):
        continue

    unique_toks = list(dict.fromkeys([str(t).strip() for t in toks if str(t).strip() != ""]))

    if len(unique_toks) < 2:
        continue

    for tok in unique_toks:
        cooc_sets[tok].update([other for other in unique_toks if other != tok])

cooc_breadth_df = pd.DataFrame({
    "token": list(cooc_sets.keys()),
    "cooc_unique_token_count": [len(v) for v in cooc_sets.values()]
})

token_quality_stats = token_quality_stats.merge(
    cooc_breadth_df,
    on="token",
    how="left"
)

token_quality_stats["cooc_unique_token_count"] = (
    token_quality_stats["cooc_unique_token_count"]
    .fillna(0)
    .astype(int)
)

# ------------------------------------------------------------
# 5) 저정보 후보 점수 생성
# ------------------------------------------------------------
# 점수는 자동 삭제 기준이 아니라 검토 우선순위입니다.

for col in [
    "freq",
    "row_count",
    "file_count",
    "sido_count",
    "jichache_count",
    "cooc_unique_token_count"
]:
    rank_col = col + "_pct_rank"
    token_quality_stats[rank_col] = token_quality_stats[col].rank(pct=True)

token_quality_stats["lowinfo_review_score"] = (
    token_quality_stats["freq_pct_rank"] * 0.25
    + token_quality_stats["row_count_pct_rank"] * 0.20
    + token_quality_stats["file_count_pct_rank"] * 0.20
    + token_quality_stats["sido_count_pct_rank"] * 0.10
    + token_quality_stats["jichache_count_pct_rank"] * 0.10
    + token_quality_stats["cooc_unique_token_count_pct_rank"] * 0.15
).round(4)

# ------------------------------------------------------------
# 6) 1차 판정 라벨
# ------------------------------------------------------------

def classify_lowinfo_review(row):
    token = row["token"]
    freq = row["freq"]
    file_share = row["file_share_percent"]
    sido_share = row["sido_share_percent"]
    cooc_n = row["cooc_unique_token_count"]
    token_len = row["token_len"]

    # 이미 제거한 후보가 혹시 남아 있으면 별도 표시
    if token in ["비", "날", "한마음"]:
        return "이미제거대상_잔존확인"

    # 한 글자 토큰은 별도 검토
    if token_len == 1:
        return "한글자_추가검토"

    # 고빈도·광범위·범용 연결 후보
    if freq >= 1000 and file_share >= 70 and sido_share >= 80:
        return "검토우선_초고빈도광범위"

    if freq >= 500 and file_share >= 50 and sido_share >= 70 and cooc_n >= 300:
        return "검토우선_고빈도범용연결"

    if freq >= 300 and file_share >= 40 and cooc_n >= 200:
        return "검토_중빈도범용연결"

    return "일반"

token_quality_stats["검토판정_초안"] = token_quality_stats.apply(
    classify_lowinfo_review,
    axis=1
)

# ------------------------------------------------------------
# 7) 보기용 정렬
# ------------------------------------------------------------

decision_order = {
    "이미제거대상_잔존확인": 1,
    "검토우선_초고빈도광범위": 2,
    "검토우선_고빈도범용연결": 3,
    "검토_중빈도범용연결": 4,
    "한글자_추가검토": 5,
    "일반": 9
}

token_quality_stats["정렬순서"] = (
    token_quality_stats["검토판정_초안"]
    .map(decision_order)
    .fillna(9)
)

token_quality_review = (
    token_quality_stats
    .sort_values(
        ["정렬순서", "lowinfo_review_score", "freq"],
        ascending=[True, False, False]
    )
    .drop(columns=["정렬순서"])
    .reset_index(drop=True)
)

print("[1] 저정보 토큰 후보 점검표 상위 200개")
display(token_quality_review.head(200))

print("\n[2] 검토판정별 요약")
token_quality_review_summary = (
    token_quality_review
    .groupby("검토판정_초안")
    .agg(
        토큰수=("token", "count"),
        freq합계=("freq", "sum"),
        평균검토점수=("lowinfo_review_score", "mean")
    )
    .reset_index()
    .sort_values("freq합계", ascending=False)
)

display(token_quality_review_summary)

print("\n[3] 고빈도 상위 100개")
display(
    token_quality_stats
    .sort_values("freq", ascending=False)
    .head(100)
)

print("\n03-TOKEN-SUPP-7 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("다음 셀에서 후보별 실제 사업명 사례를 확인합니다.")
```

```python
# 03-TOKEN-SUPP-8. 저정보 후보별 실제 사업명 사례 확인
# 주의:
# - 이 셀은 토큰을 수정하지 않습니다.
# - 검토 우선 후보별 사업명 사례를 출력합니다.

import pandas as pd

if "token_quality_review" not in globals():
    raise NameError("token_quality_review가 없습니다. 03-TOKEN-SUPP-7을 먼저 실행하세요.")

TOKEN_COL_FOR_REVIEW = "tokens_analysis_network_final"

# ------------------------------------------------------------
# 1) 검토 대상 토큰 선정
# ------------------------------------------------------------
# 일반 토큰은 제외하고, 검토판정이 붙은 토큰 위주로 확인합니다.

review_target_tokens = (
    token_quality_review[
        token_quality_review["검토판정_초안"].isin([
            "검토우선_초고빈도광범위",
            "검토우선_고빈도범용연결",
            "검토_중빈도범용연결",
            "한글자_추가검토"
        ])
    ]
    .sort_values(["lowinfo_review_score", "freq"], ascending=[False, False])
    ["token"]
    .head(120)
    .tolist()
)

print("검토 대상 토큰 수:", len(review_target_tokens))
print(review_target_tokens[:50])

# ------------------------------------------------------------
# 2) 토큰별 사례 추출 함수
# ------------------------------------------------------------

def sample_rows_for_review_token(df, token_col, token, n=5):
    mask = df[token_col].apply(
        lambda toks: token in toks if isinstance(toks, list) else False
    )

    sample_cols = [
        "v1_row_id",
        "시도",
        "지자체명",
        "원파일명",
        "사업명",
        "사업명_clean_spacefix",
        token_col,
        "보조사업자",
        "평가결과",
        "평가종류"
    ]
    sample_cols = [col for col in sample_cols if col in df.columns]

    return df.loc[mask, sample_cols].head(n).copy()

# ------------------------------------------------------------
# 3) 사례표 생성
# ------------------------------------------------------------

lowinfo_review_sample_records = []

for tok in review_target_tokens:
    stat_row = token_quality_review[token_quality_review["token"] == tok].iloc[0]

    sample_df = sample_rows_for_review_token(
        df_token_supp,
        TOKEN_COL_FOR_REVIEW,
        tok,
        n=5
    )

    for _, row in sample_df.iterrows():
        lowinfo_review_sample_records.append({
            "token": tok,
            "검토판정_초안": stat_row["검토판정_초안"],
            "freq": int(stat_row["freq"]),
            "row_count": int(stat_row["row_count"]),
            "file_count": int(stat_row["file_count"]),
            "sido_count": int(stat_row["sido_count"]),
            "cooc_unique_token_count": int(stat_row["cooc_unique_token_count"]),
            "lowinfo_review_score": float(stat_row["lowinfo_review_score"]),
            "v1_row_id": row.get("v1_row_id", None),
            "시도": row.get("시도", None),
            "지자체명": row.get("지자체명", None),
            "원파일명": row.get("원파일명", None),
            "사업명": row.get("사업명", None),
            "사업명_clean_spacefix": row.get("사업명_clean_spacefix", None),
            TOKEN_COL_FOR_REVIEW: row.get(TOKEN_COL_FOR_REVIEW, None)
        })

lowinfo_review_samples = pd.DataFrame(lowinfo_review_sample_records)

print("[1] 저정보 후보별 실제 사업명 사례")
display(lowinfo_review_samples)

# ------------------------------------------------------------
# 4) 사람이 판단할 수 있는 후보 판정표 초안
# ------------------------------------------------------------

manual_lowinfo_decision_sheet = token_quality_review[
    token_quality_review["token"].isin(review_target_tokens)
].copy()

manual_lowinfo_decision_sheet["수동판단"] = ""
manual_lowinfo_decision_sheet["처리방향"] = ""
manual_lowinfo_decision_sheet["검토메모"] = ""

print("\n[2] 수동 판단용 저정보 후보 판정표")
display(manual_lowinfo_decision_sheet)

print("\n03-TOKEN-SUPP-8 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("다음 셀에서 검토용 Excel로 저장할 수 있습니다.")
```

```python
# 03-TOKEN-SUPP-9. 저정보 후보 검토표 Excel 저장
# 목적:
# - 글자 수와 무관한 저정보 후보 점검표를 review 폴더에 저장합니다.
# - 이 파일을 보고 사용자가 제거할 토큰을 직접 판단합니다.

from datetime import datetime
from zoneinfo import ZoneInfo
import pandas as pd

SAVE_REVIEW_OUTPUT = True

RUN_TS_REVIEW = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M")

if "TABLE_DIR" in globals():
    REVIEW_DIR = TABLE_DIR / "review"
else:
    REVIEW_DIR = FINAL_DIR.parent / "review"

REVIEW_DIR.mkdir(parents=True, exist_ok=True)

out_lowinfo_review_xlsx = REVIEW_DIR / f"03_token_lowinfo_general_review_v1_{RUN_TS_REVIEW}.xlsx"

if SAVE_REVIEW_OUTPUT:
    with pd.ExcelWriter(out_lowinfo_review_xlsx, engine="openpyxl") as writer:
        token_quality_review.to_excel(
            writer,
            sheet_name="token_quality_review",
            index=False
        )
        token_quality_review_summary.to_excel(
            writer,
            sheet_name="summary",
            index=False
        )
        manual_lowinfo_decision_sheet.to_excel(
            writer,
            sheet_name="manual_decision",
            index=False
        )
        lowinfo_review_samples.to_excel(
            writer,
            sheet_name="token_samples",
            index=False
        )

    print("저정보 후보 검토용 Excel 저장 완료")
    print("- Excel:", out_lowinfo_review_xlsx)
    print("- exists:", out_lowinfo_review_xlsx.exists())
else:
    print("SAVE_REVIEW_OUTPUT = False이므로 저장하지 않았습니다.")

print("\n03-TOKEN-SUPP-9 완료")
print("이 Excel에서 수동판단/처리방향/검토메모를 채운 뒤, 실제 제거 목록을 확정하면 됩니다.")
```

### - 전체 토큰 기준 인접 2-gram·3-gram 재결합 후보표 검토

```python
# 03-TOKEN-SUPP-10. 네트워크 토큰 master 불러오기
# 목적:
# - 03-TOKEN-SUPP-6에서 저장한 네트워크용 토큰 보완 master를 불러옵니다.
# - 이후 전체 인접 2-gram/3-gram 재결합 후보를 검토합니다.
#
# 주의:
# - 이 셀은 파일을 새로 생성하지 않습니다.
# - tokens_analysis_network_final을 기준으로 재결합 후보를 생성합니다.

from pathlib import Path
import pandas as pd
import numpy as np
import ast

# ------------------------------------------------------------
# 1) 경로 설정
# ------------------------------------------------------------

BASE_DIR = Path(
    "/content/drive/MyDrive/파공장 (PA Factory)/2026 KAPA/[guide & output] 분석 협업 매뉴얼 및 Figure 후보"
)

TABLE_DIR = BASE_DIR / "03_중간결과표" / "03_v1_사업명텍스트분석" / "tables"
FINAL_DIR = TABLE_DIR / "final"
REVIEW_DIR = TABLE_DIR / "review"

NETWORK_MASTER_NAME = "03_project_name_text_analysis_network_tokens_v1_20260607_1810.parquet"
NETWORK_MASTER_FILE = FINAL_DIR / NETWORK_MASTER_NAME

print("기본 경로:")
print(NETWORK_MASTER_FILE)
print("exists:", NETWORK_MASTER_FILE.exists())

# ------------------------------------------------------------
# 2) 기본 경로에 없으면 Drive 전체에서 자동 검색
# ------------------------------------------------------------

if not NETWORK_MASTER_FILE.exists():
    print("\n기본 경로에서 파일을 찾지 못했습니다. /content/drive 아래에서 자동 검색합니다.")

    search_root = Path("/content/drive")
    matches = list(search_root.rglob(NETWORK_MASTER_NAME))

    if len(matches) == 0:
        similar = list(search_root.rglob("03_project_name_text_analysis_network_tokens*.parquet"))
        print("\n유사 parquet 후보:")
        for i, p in enumerate(similar[:20], start=1):
            print(f"{i}. {p}")

        raise FileNotFoundError(
            f"{NETWORK_MASTER_NAME} 파일을 찾지 못했습니다. "
            "03-TOKEN-SUPP-6 저장 결과 또는 Drive 마운트 계정을 확인하세요."
        )
    else:
        NETWORK_MASTER_FILE = matches[0]
        print("자동 검색 성공:")
        print(NETWORK_MASTER_FILE)

# ------------------------------------------------------------
# 3) 토큰 리스트 안전 변환 함수
# ------------------------------------------------------------

def ensure_token_list(x):
    if isinstance(x, list):
        return [str(t).strip() for t in x if str(t).strip() != ""]

    if isinstance(x, tuple):
        return [str(t).strip() for t in list(x) if str(t).strip() != ""]

    if isinstance(x, np.ndarray):
        return [str(t).strip() for t in x.tolist() if str(t).strip() != ""]

    if x is None:
        return []

    try:
        if pd.isna(x):
            return []
    except Exception:
        pass

    if isinstance(x, str):
        x_strip = x.strip()

        if x_strip == "":
            return []

        try:
            parsed = ast.literal_eval(x_strip)

            if isinstance(parsed, list):
                return [str(t).strip() for t in parsed if str(t).strip() != ""]

            if isinstance(parsed, tuple):
                return [str(t).strip() for t in list(parsed) if str(t).strip() != ""]

            if isinstance(parsed, np.ndarray):
                return [str(t).strip() for t in parsed.tolist() if str(t).strip() != ""]

        except Exception:
            pass

        return [x_strip]

    return [str(x).strip()] if str(x).strip() != "" else []

# ------------------------------------------------------------
# 4) 파일 불러오기
# ------------------------------------------------------------

df_token_supp = pd.read_parquet(NETWORK_MASTER_FILE)

print("\n불러오기 완료")
print("- 행 수:", len(df_token_supp))
print("- 열 수:", df_token_supp.shape[1])

# ------------------------------------------------------------
# 5) 주요 토큰 열 타입 보정
# ------------------------------------------------------------

token_cols_to_fix = [
    "tokens_analysis_primary",
    "tokens_analysis_second_filter",
    "tokens_analysis_network",
    "tokens_network_removed_lowinfo",
    "tokens_analysis_network_final",
    "tokens_network_merged_pairs"
]

for col in token_cols_to_fix:
    if col in df_token_supp.columns:
        df_token_supp[col] = df_token_supp[col].apply(ensure_token_list)

# ------------------------------------------------------------
# 6) 필수 열 확인
# ------------------------------------------------------------

required_cols = [
    "v1_row_id",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_network_final"
]

missing_cols = [col for col in required_cols if col not in df_token_supp.columns]

if missing_cols:
    raise KeyError(f"필수 열이 없습니다: {missing_cols}")

check_summary = pd.DataFrame({
    "항목": [
        "행 수",
        "열 수",
        "v1_row_id 결측 수",
        "v1_row_id 중복 수",
        "원파일명 결측 수",
        "tokens_analysis_network_final 빈 토큰 행 수"
    ],
    "값": [
        len(df_token_supp),
        df_token_supp.shape[1],
        int(df_token_supp["v1_row_id"].isna().sum()),
        int(df_token_supp["v1_row_id"].duplicated().sum()),
        int(df_token_supp["원파일명"].isna().sum()),
        int(df_token_supp["tokens_analysis_network_final"].apply(len).eq(0).sum())
    ]
})

display(check_summary)

print("\n03-TOKEN-SUPP-10 완료")
print("이제 03-TOKEN-SUPP-11을 실행하세요.")
```

```python
# 03-TOKEN-SUPP-11. 전체 인접 2-gram·3-gram 재결합 후보 생성
# 목적:
# - tokens_analysis_network_final 기준으로 인접 2-gram, 3-gram 후보를 전수 집계합니다.
# - 단순 동시출현이 아니라 "나란히 등장한 토큰 조합"만 봅니다.
# - 실제 사업명 원문에서 붙은 형태가 확인되는지도 함께 계산합니다.
#
# 주의:
# - 이 셀은 토큰을 수정하지 않습니다.
# - 재결합 후보 검토표만 생성합니다.
# - 재결합은 반드시 수동 판단 후 확정합니다.

from collections import Counter, defaultdict
from datetime import datetime
from zoneinfo import ZoneInfo
import pandas as pd
import numpy as np
import re
import unicodedata

# ------------------------------------------------------------
# 1) 설정
# ------------------------------------------------------------

TOKEN_COL_FOR_MERGE_REVIEW = "tokens_analysis_network_final"

# 너무 낮게 잡으면 후보가 과도하게 많아집니다.
# 전체적으로 보려면 3, 우선 검토용이면 5 권장
MIN_ADJACENT_FREQ_FOR_REVIEW = 5

MAX_EXAMPLES_PER_NGRAM = 8

print("TOKEN_COL_FOR_MERGE_REVIEW:", TOKEN_COL_FOR_MERGE_REVIEW)
print("MIN_ADJACENT_FREQ_FOR_REVIEW:", MIN_ADJACENT_FREQ_FOR_REVIEW)
print("MAX_EXAMPLES_PER_NGRAM:", MAX_EXAMPLES_PER_NGRAM)

if "df_token_supp" not in globals():
    raise NameError("df_token_supp가 없습니다. 03-TOKEN-SUPP-10을 먼저 실행하세요.")

if TOKEN_COL_FOR_MERGE_REVIEW not in df_token_supp.columns:
    raise KeyError(f"{TOKEN_COL_FOR_MERGE_REVIEW} 열이 없습니다.")

# ------------------------------------------------------------
# 2) 원문 결합형 확인용 정규화 함수
# ------------------------------------------------------------

def normalize_for_join_check(text):
    """
    원문에서 결합형이 있는지 확인하기 위한 정규화 함수입니다.
    공백과 주요 기호를 제거합니다.
    예: '공동 주택', '공동·주택', '공동-주택' → '공동주택'
    """
    if pd.isna(text):
        return ""

    text = str(text)
    text = unicodedata.normalize("NFKC", text)

    # 공백 제거
    text = re.sub(r"\s+", "", text)

    # 결합 확인에 방해되는 기호 제거
    text = re.sub(r"[·ㆍ\-\_/.,:;()\[\]{}<>「」『』'\"“”‘’]", "", text)

    return text

# ------------------------------------------------------------
# 3) 입력 열 준비
# ------------------------------------------------------------

base_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    TOKEN_COL_FOR_MERGE_REVIEW,
    "보조사업자",
    "평가결과",
    "평가종류"
]

base_cols = [col for col in base_cols if col in df_token_supp.columns]

df_merge_base = df_token_supp[base_cols].copy()

df_merge_base[TOKEN_COL_FOR_MERGE_REVIEW] = df_merge_base[TOKEN_COL_FOR_MERGE_REVIEW].apply(
    ensure_token_list
)

print("\n입력 데이터 확인")
print("- 행 수:", len(df_merge_base))
print("- 기준 토큰 열:", TOKEN_COL_FOR_MERGE_REVIEW)
print("- 빈 토큰 행 수:", int(df_merge_base[TOKEN_COL_FOR_MERGE_REVIEW].apply(len).eq(0).sum()))

# ------------------------------------------------------------
# 4) 인접 n-gram 집계
# ------------------------------------------------------------

ngram_occ_counter = Counter()
ngram_row_ids = defaultdict(set)
ngram_joined_row_ids = defaultdict(set)
ngram_examples = defaultdict(list)

for _, row in df_merge_base.iterrows():
    tokens = row[TOKEN_COL_FOR_MERGE_REVIEW]

    if not isinstance(tokens, list):
        continue

    tokens = [str(t).strip() for t in tokens if str(t).strip() != ""]

    if len(tokens) < 2:
        continue

    name_clean = row.get("사업명_clean_spacefix", "")
    name_raw = row.get("사업명", "")

    name_norm = normalize_for_join_check(name_clean)
    name_raw_norm = normalize_for_join_check(name_raw)

    for n in [2, 3]:
        if len(tokens) < n:
            continue

        for i in range(len(tokens) - n + 1):
            ngram_tokens = tuple(tokens[i:i+n])

            if any(t == "" for t in ngram_tokens):
                continue

            joined_candidate = "".join(ngram_tokens)
            joined_norm = normalize_for_join_check(joined_candidate)

            key = (n, ngram_tokens)

            ngram_occ_counter[key] += 1
            ngram_row_ids[key].add(row["v1_row_id"])

            joined_in_name = (
                joined_norm != ""
                and (
                    joined_norm in name_norm
                    or joined_norm in name_raw_norm
                )
            )

            if joined_in_name:
                ngram_joined_row_ids[key].add(row["v1_row_id"])

            if len(ngram_examples[key]) < MAX_EXAMPLES_PER_NGRAM:
                ngram_examples[key].append({
                    "ngram_type": f"{n}-gram",
                    "tokens": " + ".join(ngram_tokens),
                    "joined_candidate": joined_candidate,
                    "joined_in_name": joined_in_name,
                    "v1_row_id": row.get("v1_row_id", None),
                    "시도": row.get("시도", None),
                    "지자체명": row.get("지자체명", None),
                    "원파일명": row.get("원파일명", None),
                    "사업명": row.get("사업명", None),
                    "사업명_clean_spacefix": row.get("사업명_clean_spacefix", None),
                    "tokens_context": " / ".join(tokens),
                    "보조사업자": row.get("보조사업자", None),
                    "평가결과": row.get("평가결과", None),
                    "평가종류": row.get("평가종류", None)
                })

# ------------------------------------------------------------
# 5) 후보표 생성
# ------------------------------------------------------------

candidate_rows = []

for key, adjacent_freq in ngram_occ_counter.items():
    n, ngram_tokens = key

    if adjacent_freq < MIN_ADJACENT_FREQ_FOR_REVIEW:
        continue

    joined_candidate = "".join(ngram_tokens)

    adjacent_row_count = len(ngram_row_ids[key])
    joined_in_name_row_count = len(ngram_joined_row_ids[key])
    joined_ratio = (
        joined_in_name_row_count / adjacent_row_count
        if adjacent_row_count > 0
        else 0
    )

    example_names = " || ".join([
        str(ex["사업명_clean_spacefix"])
        for ex in ngram_examples[key][:5]
    ])

    token_1 = ngram_tokens[0] if len(ngram_tokens) >= 1 else ""
    token_2 = ngram_tokens[1] if len(ngram_tokens) >= 2 else ""
    token_3 = ngram_tokens[2] if len(ngram_tokens) >= 3 else ""

    token_len_pattern = "-".join([str(len(t)) for t in ngram_tokens])
    includes_one_char = any(len(t) == 1 for t in ngram_tokens)

    # 판정초안은 자동 확정이 아니라 검토 우선순위입니다.
    if adjacent_row_count >= 30 and joined_ratio >= 0.70:
        draft_decision = "재결합후보_강"
    elif adjacent_row_count >= 10 and joined_ratio >= 0.70:
        draft_decision = "재결합후보_중"
    elif adjacent_row_count >= 5 and joined_ratio >= 0.50:
        draft_decision = "재결합후보_약"
    else:
        draft_decision = "보류"

    candidate_rows.append({
        "ngram_type": f"{n}-gram",
        "ngram_n": n,
        "token_1": token_1,
        "token_2": token_2,
        "token_3": token_3,
        "tokens_with_plus": " + ".join(ngram_tokens),
        "joined_candidate": joined_candidate,
        "token_len_pattern": token_len_pattern,
        "includes_one_char": includes_one_char,
        "adjacent_freq": adjacent_freq,
        "adjacent_row_count": adjacent_row_count,
        "joined_in_name_row_count": joined_in_name_row_count,
        "joined_in_name_ratio": round(joined_ratio, 4),
        "판정초안": draft_decision,
        "example_names": example_names,
        "수동판단": "",
        "처리방향": "",
        "검토메모": ""
    })

ngram_merge_candidates = pd.DataFrame(candidate_rows)

decision_order = {
    "재결합후보_강": 1,
    "재결합후보_중": 2,
    "재결합후보_약": 3,
    "보류": 4
}

ngram_merge_candidates["정렬순서"] = (
    ngram_merge_candidates["판정초안"]
    .map(decision_order)
    .fillna(9)
)

ngram_merge_candidates = (
    ngram_merge_candidates
    .sort_values(
        ["정렬순서", "ngram_n", "adjacent_row_count", "joined_in_name_ratio"],
        ascending=[True, True, False, False]
    )
    .drop(columns=["정렬순서"])
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 6) 사례 long table 생성
# ------------------------------------------------------------

example_rows = []

for key, examples in ngram_examples.items():
    n, ngram_tokens = key
    adjacent_freq = ngram_occ_counter[key]

    if adjacent_freq < MIN_ADJACENT_FREQ_FOR_REVIEW:
        continue

    adjacent_row_count = len(ngram_row_ids[key])
    joined_in_name_row_count = len(ngram_joined_row_ids[key])
    joined_ratio = (
        joined_in_name_row_count / adjacent_row_count
        if adjacent_row_count > 0
        else 0
    )

    for ex in examples:
        example_rows.append({
            "ngram_type": f"{n}-gram",
            "ngram_n": n,
            "tokens_with_plus": " + ".join(ngram_tokens),
            "joined_candidate": "".join(ngram_tokens),
            "adjacent_freq": adjacent_freq,
            "adjacent_row_count": adjacent_row_count,
            "joined_in_name_row_count": joined_in_name_row_count,
            "joined_in_name_ratio": round(joined_ratio, 4),
            **ex
        })

ngram_merge_examples = pd.DataFrame(example_rows)

# ------------------------------------------------------------
# 7) 요약표
# ------------------------------------------------------------

ngram_merge_summary = (
    ngram_merge_candidates
    .groupby(["ngram_type", "판정초안"])
    .agg(
        후보수=("joined_candidate", "count"),
        adjacent_freq합계=("adjacent_freq", "sum"),
        adjacent_row_count평균=("adjacent_row_count", "mean"),
        평균결합형확인비율=("joined_in_name_ratio", "mean")
    )
    .reset_index()
    .sort_values(["ngram_type", "adjacent_freq합계"], ascending=[True, False])
)

print("\n[1] n-gram 재결합 후보 요약")
display(ngram_merge_summary)

print("\n[2] 재결합후보_강 상위 100개")
display(
    ngram_merge_candidates[
        ngram_merge_candidates["판정초안"] == "재결합후보_강"
    ].head(100)
)

print("\n[3] 전체 후보 상위 150개")
display(ngram_merge_candidates.head(150))

print("\n[4] 후보별 사례 상위 100개")
display(ngram_merge_examples.head(100))

print("\n03-TOKEN-SUPP-11 완료")
print("이 셀은 파일을 저장하지 않았습니다.")
print("검토표 저장은 03-TOKEN-SUPP-12에서 진행합니다.")
```

```python
# 03-TOKEN-SUPP-12. 전체 인접 2-gram·3-gram 재결합 후보 검토표 저장
# 목적:
# - 03-TOKEN-SUPP-11에서 만든 재결합 후보표를 Excel로 저장합니다.
# - 사용자는 manual_merge_decision 시트에서 수동 판단을 입력합니다.
#
# 저장 여부:
# - SAVE_OUTPUT = True  : review 폴더에 Excel 저장
# - SAVE_OUTPUT = False : 저장하지 않고 화면만 확인

from datetime import datetime
from zoneinfo import ZoneInfo
import pandas as pd

SAVE_OUTPUT = True

RUN_TS_MERGE_REVIEW = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M")

print("SAVE_OUTPUT:", SAVE_OUTPUT)
print("RUN_TS_MERGE_REVIEW:", RUN_TS_MERGE_REVIEW)

if "ngram_merge_candidates" not in globals():
    raise NameError("ngram_merge_candidates가 없습니다. 03-TOKEN-SUPP-11을 먼저 실행하세요.")

if "ngram_merge_examples" not in globals():
    raise NameError("ngram_merge_examples가 없습니다. 03-TOKEN-SUPP-11을 먼저 실행하세요.")

if "ngram_merge_summary" not in globals():
    raise NameError("ngram_merge_summary가 없습니다. 03-TOKEN-SUPP-11을 먼저 실행하세요.")

REVIEW_DIR.mkdir(parents=True, exist_ok=True)

out_ngram_review_xlsx = REVIEW_DIR / f"03_token_adjacent_ngram_merge_review_v1_{RUN_TS_MERGE_REVIEW}.xlsx"

# ------------------------------------------------------------
# 1) 수동 판단용 시트 생성
# ------------------------------------------------------------

manual_merge_decision = ngram_merge_candidates.copy()

# 우선 검토 대상만 상단에 오도록 이미 정렬되어 있음
# 수동판단 / 처리방향 / 검토메모는 사용자가 채우는 칸

# ------------------------------------------------------------
# 2) 판단 기준 안내 시트
# ------------------------------------------------------------

merge_review_guide = pd.DataFrame({
    "구분": [
        "수동판단 값",
        "수동판단 값",
        "수동판단 값",
        "수동판단 값",
        "처리방향 값",
        "처리방향 값",
        "처리방향 값",
        "처리방향 값",
        "재결합 권장 기준",
        "재결합 보류 기준",
        "재결합 금지 기준",
        "적용 순서"
    ],
    "값": [
        "재결합",
        "분리유지",
        "보류",
        "오류후보",
        "merge",
        "keep_separate",
        "review_later",
        "error_candidate",
        "인접 빈도가 충분하고, 원문 결합형 확인 비율이 높으며, 결합 후 의미가 자연스러운 경우",
        "의미는 가능하지만 맥락이 다양하거나 사업명 사례를 더 봐야 하는 경우",
        "우연히 옆에 있거나, 붙이면 어색하거나, 더 긴 복합어의 일부인 경우",
        "최종 적용 시에는 3-gram 재결합을 먼저 적용하고, 그 다음 2-gram 재결합을 적용"
    ]
})

# ------------------------------------------------------------
# 3) 저장 실행
# ------------------------------------------------------------

if SAVE_OUTPUT:
    with pd.ExcelWriter(out_ngram_review_xlsx, engine="openpyxl") as writer:
        ngram_merge_summary.to_excel(
            writer,
            sheet_name="summary",
            index=False
        )

        ngram_merge_candidates.to_excel(
            writer,
            sheet_name="ngram_candidates_all",
            index=False
        )

        ngram_merge_candidates[
            ngram_merge_candidates["판정초안"] == "재결합후보_강"
        ].to_excel(
            writer,
            sheet_name="strong_candidates",
            index=False
        )

        manual_merge_decision.to_excel(
            writer,
            sheet_name="manual_merge_decision",
            index=False
        )

        ngram_merge_examples.to_excel(
            writer,
            sheet_name="candidate_examples",
            index=False
        )

        merge_review_guide.to_excel(
            writer,
            sheet_name="guide",
            index=False
        )

    print("재결합 후보 검토표 저장 완료")
    print("- Excel:", out_ngram_review_xlsx)
    print("- exists:", out_ngram_review_xlsx.exists())
else:
    print("SAVE_OUTPUT=False이므로 저장하지 않았습니다.")

print("\n03-TOKEN-SUPP-12 완료")
```

시각화 결과를 비교하며, 수동 검토를 여러번 걸쳐 작성됨.

```python
# 03-TOKEN-FINAL-1. 최종 처리 기준 설정 및 입력 파일 불러오기
# 목적:
# - 03에서 만든 기존 network token master를 불러옵니다.
# - 최신 수동 재결합 판단 Excel을 불러옵니다.
# - 이후 최종 토큰열(tokens_analysis_network_final_reviewed)을 생성할 준비를 합니다.
#
# 실행 위치:
# - 03 노트북 맨 아래
#
# 입력:
# - 03_project_name_text_analysis_network_tokens_v1_20260607_1810.parquet
# - 03_token_adjacent_ngram_merge_review_v1_20260611_2023.xlsx
#
# 출력:
# - 아직 파일 저장 없음

from pathlib import Path
from datetime import datetime
from zoneinfo import ZoneInfo
import pandas as pd
import numpy as np
import ast

# ------------------------------------------------------------
# 1) 경로 설정
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/파공장 (PA Factory)/2026 KAPA/[guide & output] 분석 협업 매뉴얼 및 Figure 후보"
)

TABLE_DIR = PROJECT_ROOT / "03_중간결과표" / "03_v1_사업명텍스트분석" / "tables"
FINAL_DIR = TABLE_DIR / "final"
REVIEW_DIR = TABLE_DIR / "review"

FINAL_DIR.mkdir(parents=True, exist_ok=True)
REVIEW_DIR.mkdir(parents=True, exist_ok=True)

RUN_VERSION = "v1"
RUN_TS_FINAL = datetime.now(ZoneInfo("Asia/Seoul")).strftime("%Y%m%d_%H%M")

print("PROJECT_ROOT:", PROJECT_ROOT)
print("FINAL_DIR:", FINAL_DIR)
print("REVIEW_DIR:", REVIEW_DIR)
print("RUN_VERSION:", RUN_VERSION)
print("RUN_TS_FINAL:", RUN_TS_FINAL)

# ------------------------------------------------------------
# 2) 입력 파일명 설정
# ------------------------------------------------------------

NETWORK_MASTER_NAME = "03_project_name_text_analysis_network_tokens_v1_20260607_1810.parquet"
MERGE_DECISION_XLSX_NAME = "03_token_adjacent_ngram_merge_review_v1_20260611_2023.xlsx"

NETWORK_MASTER_FILE = FINAL_DIR / NETWORK_MASTER_NAME
MERGE_DECISION_XLSX = REVIEW_DIR / MERGE_DECISION_XLSX_NAME

print("\n[입력 파일 확인]")
print("NETWORK_MASTER_FILE:", NETWORK_MASTER_FILE)
print("exists:", NETWORK_MASTER_FILE.exists())
print("MERGE_DECISION_XLSX:", MERGE_DECISION_XLSX)
print("exists:", MERGE_DECISION_XLSX.exists())

# ------------------------------------------------------------
# 3) 파일이 기본 경로에 없으면 Drive 전체에서 자동 검색
# ------------------------------------------------------------

if not NETWORK_MASTER_FILE.exists():
    print("\nNETWORK_MASTER_FILE을 기본 경로에서 찾지 못했습니다. /content/drive에서 검색합니다.")
    matches = list(Path("/content/drive").rglob(NETWORK_MASTER_NAME))
    if len(matches) == 0:
        raise FileNotFoundError(f"{NETWORK_MASTER_NAME} 파일을 찾지 못했습니다.")
    NETWORK_MASTER_FILE = matches[0]
    print("자동 검색 성공:", NETWORK_MASTER_FILE)

if not MERGE_DECISION_XLSX.exists():
    print("\nMERGE_DECISION_XLSX를 기본 경로에서 찾지 못했습니다. /content/drive에서 검색합니다.")
    matches = list(Path("/content/drive").rglob(MERGE_DECISION_XLSX_NAME))
    if len(matches) == 0:
        raise FileNotFoundError(
            f"{MERGE_DECISION_XLSX_NAME} 파일을 찾지 못했습니다. "
            "해당 Excel 파일을 03_v1_사업명텍스트분석/tables/review 폴더에 업로드하세요."
        )
    MERGE_DECISION_XLSX = matches[0]
    print("자동 검색 성공:", MERGE_DECISION_XLSX)

# ------------------------------------------------------------
# 4) 토큰 리스트 안전 변환 함수
# ------------------------------------------------------------

def ensure_token_list(x):
    if isinstance(x, list):
        return [str(t).strip() for t in x if str(t).strip() != ""]

    if isinstance(x, tuple):
        return [str(t).strip() for t in list(x) if str(t).strip() != ""]

    if isinstance(x, np.ndarray):
        return [str(t).strip() for t in x.tolist() if str(t).strip() != ""]

    if x is None:
        return []

    try:
        if pd.isna(x):
            return []
    except Exception:
        pass

    if isinstance(x, str):
        x_strip = x.strip()
        if x_strip == "":
            return []

        try:
            parsed = ast.literal_eval(x_strip)
            if isinstance(parsed, list):
                return [str(t).strip() for t in parsed if str(t).strip() != ""]
            if isinstance(parsed, tuple):
                return [str(t).strip() for t in list(parsed) if str(t).strip() != ""]
        except Exception:
            pass

        return [x_strip]

    return [str(x).strip()] if str(x).strip() != "" else []

# ------------------------------------------------------------
# 5) 03 network token master 불러오기
# ------------------------------------------------------------

df_03_final = pd.read_parquet(NETWORK_MASTER_FILE)

print("\n[03 network token master 불러오기 완료]")
print("행 수:", len(df_03_final))
print("열 수:", df_03_final.shape[1])

# ------------------------------------------------------------
# 6) 주요 토큰 열 타입 보정
# ------------------------------------------------------------

token_cols_to_fix = [
    "tokens_analysis_primary",
    "tokens_analysis_second_filter",
    "tokens_analysis_network",
    "tokens_network_removed_lowinfo",
    "tokens_analysis_network_final",
    "tokens_network_merged_pairs"
]

for col in token_cols_to_fix:
    if col in df_03_final.columns:
        df_03_final[col] = df_03_final[col].apply(ensure_token_list)

# ------------------------------------------------------------
# 7) 필수 열 확인
# ------------------------------------------------------------

required_cols = [
    "v1_row_id",
    "원파일명",
    "사업명",
    "tokens_analysis_network_final"
]

missing_cols = [col for col in required_cols if col not in df_03_final.columns]

if missing_cols:
    raise KeyError(f"필수 열이 없습니다: {missing_cols}")

input_check = pd.DataFrame({
    "항목": [
        "입력 parquet",
        "행 수",
        "열 수",
        "v1_row_id 결측 수",
        "v1_row_id 중복 수",
        "tokens_analysis_network_final 빈 토큰 행 수"
    ],
    "값": [
        str(NETWORK_MASTER_FILE),
        len(df_03_final),
        df_03_final.shape[1],
        int(df_03_final["v1_row_id"].isna().sum()),
        int(df_03_final["v1_row_id"].duplicated().sum()),
        int(df_03_final["tokens_analysis_network_final"].apply(len).eq(0).sum())
    ]
})

display(input_check)

print("\n03-TOKEN-FINAL-1 완료")
```

```python
# 03-TOKEN-FINAL-2. 재결합·제거·보존/보류 목록 생성
# 목적:
# - 최신 수동 재결합 판단 Excel에서 재결합 규칙을 읽습니다.
# - 최종 제거 토큰 목록을 만듭니다.
# - 보존/보류 토큰 목록도 문서화합니다.
#
# 주의:
# - 이 셀은 목록만 만듭니다.
# - 실제 토큰열 생성은 03-TOKEN-FINAL-3에서 수행합니다.

import pandas as pd
import numpy as np

if "df_03_final" not in globals():
    raise NameError("df_03_final이 없습니다. 03-TOKEN-FINAL-1을 먼저 실행하세요.")

if "MERGE_DECISION_XLSX" not in globals():
    raise NameError("MERGE_DECISION_XLSX가 없습니다. 03-TOKEN-FINAL-1을 먼저 실행하세요.")

# ------------------------------------------------------------
# 1) manual_merge_decision 읽기
# ------------------------------------------------------------

manual_merge_df = pd.read_excel(
    MERGE_DECISION_XLSX,
    sheet_name="manual_merge_decision"
)

manual_merge_df.columns = [str(c).strip() for c in manual_merge_df.columns]

required_cols = [
    "ngram_n",
    "token_1",
    "token_2",
    "token_3",
    "tokens_with_plus",
    "joined_candidate",
    "수동판단",
    "처리방향"
]

missing_cols = [col for col in required_cols if col not in manual_merge_df.columns]

if missing_cols:
    raise KeyError(f"manual_merge_decision 시트에 필요한 열이 없습니다: {missing_cols}")

for col in ["token_1", "token_2", "token_3", "joined_candidate", "수동판단", "처리방향"]:
    manual_merge_df[col] = manual_merge_df[col].fillna("").astype(str).str.strip()

# ------------------------------------------------------------
# 2) 재결합 확정 행 추출
# ------------------------------------------------------------

merge_rule_df = manual_merge_df[
    (manual_merge_df["수동판단"] == "재결합")
    | (manual_merge_df["처리방향"] == "merge")
].copy()

print("[manual_merge_decision 반영 현황]")
print("전체 후보 행 수:", len(manual_merge_df))
print("재결합 반영 행 수:", len(merge_rule_df))

if len(merge_rule_df) == 0:
    raise ValueError("재결합으로 표시된 행이 없습니다.")

# ------------------------------------------------------------
# 3) 재결합 규칙 생성
# ------------------------------------------------------------

merge_rules_rows = []

for _, row in merge_rule_df.iterrows():
    try:
        ngram_n = int(row["ngram_n"])
    except Exception:
        ngram_n = None

    tokens = []

    if row["token_1"] != "":
        tokens.append(row["token_1"])
    if row["token_2"] != "":
        tokens.append(row["token_2"])
    if row["token_3"] != "":
        tokens.append(row["token_3"])

    if ngram_n is not None:
        tokens = tokens[:ngram_n]

    tokens = [t for t in tokens if t != ""]
    joined_candidate = row["joined_candidate"]

    if len(tokens) < 2:
        continue

    if joined_candidate == "":
        joined_candidate = "".join(tokens)

    merge_rules_rows.append({
        "pattern": tuple(tokens),
        "merged": joined_candidate,
        "ngram_n": len(tokens),
        "tokens_with_plus": " + ".join(tokens),
        "수동판단": row["수동판단"],
        "처리방향": row["처리방향"]
    })

merge_rules_final_df = pd.DataFrame(merge_rules_rows)

merge_rules_final_df = (
    merge_rules_final_df
    .drop_duplicates(subset=["pattern", "merged"])
    .sort_values(["ngram_n", "tokens_with_plus"], ascending=[False, True])
    .reset_index(drop=True)
)

merge_rules_final = [
    (tuple(row["pattern"]), row["merged"])
    for _, row in merge_rules_final_df.iterrows()
]

print("\n[최종 재결합 규칙]")
print("규칙 수:", len(merge_rules_final_df))
display(merge_rules_final_df.head(100))

# ------------------------------------------------------------
# 4) 핵심 재결합 규칙 확인
# ------------------------------------------------------------

key_merge_patterns = [
    "시각 + 장애인",
    "지체 + 장애인",
    "평생 + 교육",
    "평생 + 학습",
    "주간 + 보호",
    "주간 + 보호 + 시설",
    "주간 + 보호 + 센터",
    "지역 + 아동 + 센터",
    "복지 + 시설",
    "생활 + 체육",
    "체육 + 대회",
    "자원 + 봉사",
    "역량 + 강화"
]

key_merge_check_df = merge_rules_final_df[
    merge_rules_final_df["tokens_with_plus"].isin(key_merge_patterns)
].copy()

print("\n[핵심 재결합 규칙 확인]")
display(key_merge_check_df)

# ------------------------------------------------------------
# 5) 최종 제거 토큰 목록
# ------------------------------------------------------------
# 주의:
# - '비', '날', '한마음'은 이미 tokens_analysis_network 단계에서 제거되었지만,
#   최종 문서화를 위해 포함합니다.
# - 아래 목록은 최종 tokens_analysis_network_final_reviewed 생성 시 제거 대상으로 다시 적용해도 무방합니다.

initial_removed_lowinfo_tokens = [
    "비",
    "날",
    "한마음",
]

base_lowinfo_remove_tokens = [
    "개선", "전국", "활성", "육성", "협의회", "강화", "지회", "조성", "구입",
    "설치", "우수", "사업비", "연합회", "사랑", "보급", "마당", "정비",
    "협회", "보수", "추가", "참여", "출전", "자체", "기념행사", "규모",
    "기념", "교체", "맞춤", "행복",
]

extra_lowinfo_remove_tokens = [
    "시범",
    "활동",
    "단체",
    "프로그램",
    "단위",
    "포인트",
]

final_remove_tokens = sorted(
    set(initial_removed_lowinfo_tokens)
    | set(base_lowinfo_remove_tokens)
    | set(extra_lowinfo_remove_tokens)
)

removed_token_decision_rows = []

for tok in initial_removed_lowinfo_tokens:
    removed_token_decision_rows.append({
        "token": tok,
        "수동판단": "제거",
        "처리방향": "remove_lowinfo",
        "처리단계": "03 1차 저정보 제거",
        "검토메모": "네트워크에서 정보량이 낮거나 행사/비용 접미 성격이 강해 제거"
    })

for tok in base_lowinfo_remove_tokens:
    removed_token_decision_rows.append({
        "token": tok,
        "수동판단": "제거",
        "처리방향": "remove_lowinfo",
        "처리단계": "기본 저정보 토큰 제거",
        "검토메모": "사업명에서 반복되지만 정책대상·정책분야 식별력은 낮은 일반 표현"
    })

for tok in extra_lowinfo_remove_tokens:
    removed_token_decision_rows.append({
        "token": tok,
        "수동판단": "제거",
        "처리방향": "remove_lowinfo",
        "처리단계": "06A 시각화 검토 후 추가 제거",
        "검토메모": "재결합 후에도 단독 잔여 토큰의 정보량이 낮아 lowinfo2에서 제거"
    })

removed_token_decision_df = (
    pd.DataFrame(removed_token_decision_rows)
    .drop_duplicates(subset=["token", "처리단계"])
    .reset_index(drop=True)
)

print("\n[최종 제거 토큰 목록]")
print("제거 토큰 unique 수:", len(final_remove_tokens))
display(removed_token_decision_df)

# ------------------------------------------------------------
# 6) 보존/보류 토큰 목록
# ------------------------------------------------------------

keep_hold_token_rows = [
    {
        "token": "대회",
        "수동판단": "보류",
        "처리방향": "keep_for_now",
        "검토메모": "체육·장애인체육 영역에서 사업형식을 보여주는 핵심 연결어. 추가 제거 시 네트워크 분절 가능성이 있어 보류"
    },
    {
        "token": "행사",
        "수동판단": "보류",
        "처리방향": "keep_for_now",
        "검토메모": "보훈·문화·복지 영역의 행사성 보조사업을 보여줄 수 있어 보류"
    },
    {
        "token": "종목",
        "수동판단": "보류",
        "처리방향": "keep_for_now",
        "검토메모": "체육·장애인체육 맥락에서 의미가 있어 제거하지 않음"
    },
    {
        "token": "교실",
        "수동판단": "보류",
        "처리방향": "keep_for_now",
        "검토메모": "교육·생활체육 프로그램 유형을 나타낼 수 있어 제거하지 않음"
    },
    {
        "token": "센터",
        "수동판단": "보류",
        "처리방향": "keep_for_now",
        "검토메모": "복지·돌봄·지역 기반 전달체계 의미가 있어 제거하지 않음"
    },
    {
        "token": "역량강화",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "역량+강화 재결합 결과. 교육·주민·농업인 사업유형을 나타내므로 보존"
    },
    {
        "token": "시각장애인",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "시각+장애인 재결합 결과. 장애 유형을 나타내므로 보존"
    },
    {
        "token": "지체장애인",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "지체+장애인 재결합 결과. 장애 유형을 나타내므로 보존"
    },
    {
        "token": "평생교육",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "평생+교육 재결합 결과. 교육 분야 사업유형으로 보존"
    },
    {
        "token": "평생학습",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "평생+학습 재결합 결과. 교육 분야 사업유형으로 보존"
    },
    {
        "token": "주간보호",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "주간+보호 재결합 결과. 복지·돌봄 서비스 유형으로 보존"
    },
    {
        "token": "주간보호시설",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "주간+보호+시설 재결합 결과. 복지시설 유형으로 보존"
    },
    {
        "token": "주간보호센터",
        "수동판단": "보존",
        "처리방향": "keep",
        "검토메모": "주간+보호+센터 재결합 결과. 복지 전달체계로 보존"
    },
]

keep_hold_token_decision_df = pd.DataFrame(keep_hold_token_rows)

print("\n[보존/보류 토큰 목록]")
display(keep_hold_token_decision_df)

# ------------------------------------------------------------
# 7) 목록 요약
# ------------------------------------------------------------

token_decision_summary_df = pd.DataFrame({
    "항목": [
        "재결합 규칙 수",
        "제거 토큰 unique 수",
        "보존/보류 토큰 수",
        "최종 입력 토큰열",
        "최종 출력 토큰열"
    ],
    "값": [
        len(merge_rules_final_df),
        len(final_remove_tokens),
        len(keep_hold_token_decision_df),
        "tokens_analysis_network_final",
        "tokens_analysis_network_final_reviewed"
    ]
})

print("\n[처리 기준 요약]")
display(token_decision_summary_df)

print("\n03-TOKEN-FINAL-2 완료")
```

```python
# 03-TOKEN-FINAL-3. 최종 토큰열 생성
# 목적:
# - tokens_analysis_network_final을 기준으로 최종 재결합 규칙을 먼저 적용합니다.
# - 그 다음 최종 제거 토큰 목록을 제거합니다.
# - 03 이후 단계에서 사용할 공식 토큰열 tokens_analysis_network_final_reviewed를 생성합니다.
#
# 최종 공식 토큰열:
# - tokens_analysis_network_final_reviewed

if "df_03_final" not in globals():
    raise NameError("df_03_final이 없습니다. 03-TOKEN-FINAL-1을 먼저 실행하세요.")

if "merge_rules_final" not in globals():
    raise NameError("merge_rules_final이 없습니다. 03-TOKEN-FINAL-2를 먼저 실행하세요.")

if "final_remove_tokens" not in globals():
    raise NameError("final_remove_tokens가 없습니다. 03-TOKEN-FINAL-2를 먼저 실행하세요.")

SOURCE_TOKEN_COL_FINAL = "tokens_analysis_network_final"
MERGED_TOKEN_COL_FINAL = "tokens_analysis_network_final_manualmerge"
FINAL_REVIEWED_TOKEN_COL = "tokens_analysis_network_final_reviewed"

if SOURCE_TOKEN_COL_FINAL not in df_03_final.columns:
    raise KeyError(f"{SOURCE_TOKEN_COL_FINAL} 열이 없습니다.")

print("[최종 토큰열 생성]")
print("입력 토큰열:", SOURCE_TOKEN_COL_FINAL)
print("재결합 후 토큰열:", MERGED_TOKEN_COL_FINAL)
print("최종 reviewed 토큰열:", FINAL_REVIEWED_TOKEN_COL)

# ------------------------------------------------------------
# 1) 재결합 적용 함수
# ------------------------------------------------------------

def apply_manual_merge_rules_final(tokens, merge_rules):
    tokens = ensure_token_list(tokens)

    out_tokens = []
    applied_rules = []

    i = 0

    while i < len(tokens):
        matched = False

        for pattern, merged in merge_rules:
            n = len(pattern)

            if n == 0:
                continue

            if tuple(tokens[i:i+n]) == pattern:
                out_tokens.append(merged)
                applied_rules.append(f"{' + '.join(pattern)} -> {merged}")
                i += n
                matched = True
                break

        if not matched:
            out_tokens.append(tokens[i])
            i += 1

    return out_tokens, applied_rules

# ------------------------------------------------------------
# 2) 최종 제거 함수
# ------------------------------------------------------------

final_remove_set = set(final_remove_tokens)

def remove_final_lowinfo(tokens):
    tokens = ensure_token_list(tokens)
    return [tok for tok in tokens if tok not in final_remove_set]

def get_final_removed_tokens(tokens):
    tokens = ensure_token_list(tokens)
    return [tok for tok in tokens if tok in final_remove_set]

# ------------------------------------------------------------
# 3) 재결합 먼저 적용
# ------------------------------------------------------------

df_03_final[SOURCE_TOKEN_COL_FINAL] = df_03_final[SOURCE_TOKEN_COL_FINAL].apply(
    ensure_token_list
)

merged_results = df_03_final[SOURCE_TOKEN_COL_FINAL].apply(
    lambda toks: apply_manual_merge_rules_final(toks, merge_rules_final)
)

df_03_final[MERGED_TOKEN_COL_FINAL] = merged_results.apply(lambda x: x[0])
df_03_final["tokens_network_manual_merged_pairs_final"] = merged_results.apply(lambda x: x[1])

# ------------------------------------------------------------
# 4) 최종 저정보 토큰 제거
# ------------------------------------------------------------

df_03_final[FINAL_REVIEWED_TOKEN_COL] = df_03_final[MERGED_TOKEN_COL_FINAL].apply(
    remove_final_lowinfo
)

df_03_final["tokens_network_removed_lowinfo_final"] = df_03_final[
    MERGED_TOKEN_COL_FINAL
].apply(get_final_removed_tokens)

df_03_final[f"{FINAL_REVIEWED_TOKEN_COL}_count"] = df_03_final[
    FINAL_REVIEWED_TOKEN_COL
].apply(len)

df_03_final[f"{FINAL_REVIEWED_TOKEN_COL}_empty"] = df_03_final[
    FINAL_REVIEWED_TOKEN_COL
].apply(lambda x: len(x) == 0)

# ------------------------------------------------------------
# 5) 적용 결과 요약
# ------------------------------------------------------------

merge_rows_changed = df_03_final["tokens_network_manual_merged_pairs_final"].apply(len).gt(0).sum()
merge_total_applied = df_03_final["tokens_network_manual_merged_pairs_final"].apply(len).sum()

removed_rows_changed = df_03_final["tokens_network_removed_lowinfo_final"].apply(len).gt(0).sum()
removed_total_applied = df_03_final["tokens_network_removed_lowinfo_final"].apply(len).sum()

source_empty_rows = df_03_final[SOURCE_TOKEN_COL_FINAL].apply(len).eq(0).sum()
final_empty_rows = df_03_final[f"{FINAL_REVIEWED_TOKEN_COL}_empty"].sum()

final_token_apply_summary_df = pd.DataFrame({
    "항목": [
        "입력 토큰열",
        "재결합 후 토큰열",
        "최종 reviewed 토큰열",
        "재결합 규칙 수",
        "재결합 발생 행 수",
        "재결합 총 적용 횟수",
        "최종 제거 토큰 unique 수",
        "최종 제거 발생 행 수",
        "최종 제거 총 횟수",
        "입력 빈 토큰 행 수",
        "최종 빈 토큰 행 수",
        "추가 빈 토큰 행 수"
    ],
    "값": [
        SOURCE_TOKEN_COL_FINAL,
        MERGED_TOKEN_COL_FINAL,
        FINAL_REVIEWED_TOKEN_COL,
        len(merge_rules_final_df),
        int(merge_rows_changed),
        int(merge_total_applied),
        len(final_remove_tokens),
        int(removed_rows_changed),
        int(removed_total_applied),
        int(source_empty_rows),
        int(final_empty_rows),
        int(final_empty_rows - source_empty_rows)
    ]
})

print("\n[최종 토큰 적용 요약]")
display(final_token_apply_summary_df)

print("\n[재결합 적용 사례 상위 50개]")
merge_changed_cases = df_03_final[
    df_03_final["tokens_network_manual_merged_pairs_final"].apply(len).gt(0)
].copy()

display_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "사업명",
    SOURCE_TOKEN_COL_FINAL,
    MERGED_TOKEN_COL_FINAL,
    FINAL_REVIEWED_TOKEN_COL,
    "tokens_network_manual_merged_pairs_final",
    "tokens_network_removed_lowinfo_final"
]

display_cols = [col for col in display_cols if col in merge_changed_cases.columns]

display(merge_changed_cases[display_cols].head(50))

print("\n[최종 제거 적용 사례 상위 50개]")
removed_changed_cases = df_03_final[
    df_03_final["tokens_network_removed_lowinfo_final"].apply(len).gt(0)
].copy()

display_cols_removed = [
    "v1_row_id",
    "시도",
    "지자체명",
    "사업명",
    MERGED_TOKEN_COL_FINAL,
    FINAL_REVIEWED_TOKEN_COL,
    "tokens_network_removed_lowinfo_final"
]

display_cols_removed = [col for col in display_cols_removed if col in removed_changed_cases.columns]

display(removed_changed_cases[display_cols_removed].head(50))

# ------------------------------------------------------------
# 6) 핵심 토큰 존재 여부 확인
# ------------------------------------------------------------

check_words = [
    "지체장애인",
    "시각장애인",
    "평생교육",
    "평생학습",
    "주간보호",
    "주간보호시설",
    "주간보호센터",
    "단위",
    "포인트",
    "시범",
    "활동",
    "단체",
    "프로그램",
    "대회",
    "행사",
    "종목",
    "교실",
    "센터",
    "역량강화"
]

from collections import Counter

final_word_counter = Counter()

for toks in df_03_final[FINAL_REVIEWED_TOKEN_COL]:
    for tok in set(ensure_token_list(toks)):
        final_word_counter[tok] += 1

check_token_rows = []

for word in check_words:
    check_token_rows.append({
        "word": word,
        "doc_freq_unique_in_final_reviewed": final_word_counter.get(word, 0),
        "최종처리해석": (
            "제거됨" if word in final_remove_set else
            "보존/보류 또는 재결합 결과"
        )
    })

check_token_df = pd.DataFrame(check_token_rows)

print("\n[핵심 토큰 최종 존재 여부 확인]")
display(check_token_df)

print("\n03-TOKEN-FINAL-3 완료")
```

```python
# 03-TOKEN-FINAL-4. 최종 파일 저장 및 06A 전달 메모 생성
# 목적:
# - 03 이후 단계에서 사용할 최종 parquet를 저장합니다.
# - 처리 기준 목록 Excel을 저장합니다.
# - 06A 전달 메모와 save_check 파일을 생성합니다.
#
# 주요 산출:
# - 03_project_name_text_analysis_network_tokens_reviewed_v1_YYYYMMDD_HHMM.parquet
# - 03_token_processing_decision_lists_v1_YYYYMMDD_HHMM.xlsx
# - 03_handoff_memo_for_06A_network_tokens_reviewed_v1_YYYYMMDD_HHMM.txt

if "df_03_final" not in globals():
    raise NameError("df_03_final이 없습니다. 03-TOKEN-FINAL-3을 먼저 실행하세요.")

if "FINAL_REVIEWED_TOKEN_COL" not in globals():
    raise NameError("FINAL_REVIEWED_TOKEN_COL이 없습니다. 03-TOKEN-FINAL-3을 먼저 실행하세요.")

SAVE_OUTPUT = True

print("SAVE_OUTPUT:", SAVE_OUTPUT)

# ------------------------------------------------------------
# 1) 저장 파일 경로
# ------------------------------------------------------------

out_parquet = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_reviewed_{RUN_VERSION}_{RUN_TS_FINAL}.parquet"
out_sample_xlsx = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_reviewed_sample_{RUN_VERSION}_{RUN_TS_FINAL}.xlsx"
out_columns_csv = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_reviewed_columns_{RUN_VERSION}_{RUN_TS_FINAL}.csv"
out_save_check_csv = FINAL_DIR / f"03_project_name_text_analysis_network_tokens_reviewed_save_check_{RUN_VERSION}_{RUN_TS_FINAL}.csv"
out_decision_xlsx = REVIEW_DIR / f"03_token_processing_decision_lists_{RUN_VERSION}_{RUN_TS_FINAL}.xlsx"
out_handoff_txt = FINAL_DIR / f"03_handoff_memo_for_06A_network_tokens_reviewed_{RUN_VERSION}_{RUN_TS_FINAL}.txt"

print("\n[저장 예정 파일]")
print("- parquet:", out_parquet)
print("- sample xlsx:", out_sample_xlsx)
print("- columns csv:", out_columns_csv)
print("- save check csv:", out_save_check_csv)
print("- decision xlsx:", out_decision_xlsx)
print("- handoff txt:", out_handoff_txt)

# ------------------------------------------------------------
# 2) 저장 전 검증
# ------------------------------------------------------------

required_final_cols = [
    "v1_row_id",
    "원파일명",
    "사업명",
    "tokens_analysis_primary",
    "tokens_analysis_network",
    "tokens_analysis_network_final",
    "tokens_analysis_network_final_manualmerge",
    "tokens_analysis_network_final_reviewed",
    "tokens_network_manual_merged_pairs_final",
    "tokens_network_removed_lowinfo_final",
    "tokens_analysis_network_final_reviewed_count",
    "tokens_analysis_network_final_reviewed_empty"
]

required_final_cols = [col for col in required_final_cols if col in df_03_final.columns]

missing_final_cols = [
    col for col in [
        "v1_row_id",
        "원파일명",
        "사업명",
        "tokens_analysis_network_final_reviewed"
    ]
    if col not in df_03_final.columns
]

if missing_final_cols:
    raise KeyError(f"최종 저장 전 필수 열 누락: {missing_final_cols}")

save_check_rows = []

save_check_rows.append({
    "항목": "행 수",
    "값": len(df_03_final)
})

save_check_rows.append({
    "항목": "열 수",
    "값": df_03_final.shape[1]
})

save_check_rows.append({
    "항목": "v1_row_id 결측 수",
    "값": int(df_03_final["v1_row_id"].isna().sum())
})

save_check_rows.append({
    "항목": "v1_row_id 중복 수",
    "값": int(df_03_final["v1_row_id"].duplicated().sum())
})

save_check_rows.append({
    "항목": "최종 토큰열",
    "값": FINAL_REVIEWED_TOKEN_COL
})

save_check_rows.append({
    "항목": "최종 토큰 빈 행 수",
    "값": int(df_03_final[f"{FINAL_REVIEWED_TOKEN_COL}_empty"].sum())
})

save_check_rows.append({
    "항목": "재결합 규칙 수",
    "값": len(merge_rules_final_df)
})

save_check_rows.append({
    "항목": "최종 제거 토큰 unique 수",
    "값": len(final_remove_tokens)
})

save_check_rows.append({
    "항목": "재결합 발생 행 수",
    "값": int(df_03_final["tokens_network_manual_merged_pairs_final"].apply(len).gt(0).sum())
})

save_check_rows.append({
    "항목": "최종 제거 발생 행 수",
    "값": int(df_03_final["tokens_network_removed_lowinfo_final"].apply(len).gt(0).sum())
})

# 제거 토큰 잔존 여부
final_tokens_flat = set()

for toks in df_03_final[FINAL_REVIEWED_TOKEN_COL]:
    final_tokens_flat.update(ensure_token_list(toks))

removed_still_remaining = sorted(set(final_remove_tokens) & final_tokens_flat)

save_check_rows.append({
    "항목": "최종 제거 토큰 잔존 수",
    "값": len(removed_still_remaining)
})

save_check_rows.append({
    "항목": "최종 제거 토큰 잔존 목록",
    "값": ", ".join(removed_still_remaining)
})

save_check_df = pd.DataFrame(save_check_rows)

print("\n[저장 전 검증]")
display(save_check_df)

if len(removed_still_remaining) > 0:
    print("\n주의: 제거 토큰 중 일부가 최종 토큰열에 남아 있습니다.")
    print(removed_still_remaining)
else:
    print("\n최종 제거 토큰은 최종 토큰열에 남아 있지 않습니다.")

# ------------------------------------------------------------
# 3) 샘플 파일 생성
# ------------------------------------------------------------

sample_cols = [
    "v1_row_id",
    "시도",
    "지자체명",
    "원파일명",
    "사업명",
    "사업명_clean_spacefix",
    "tokens_analysis_primary",
    "tokens_analysis_network_final",
    "tokens_analysis_network_final_manualmerge",
    "tokens_analysis_network_final_reviewed",
    "tokens_network_manual_merged_pairs_final",
    "tokens_network_removed_lowinfo_final"
]

sample_cols = [col for col in sample_cols if col in df_03_final.columns]

sample_df = df_03_final[sample_cols].copy()

# 재결합 또는 제거가 발생한 사례를 우선 포함
sample_priority = df_03_final[
    df_03_final["tokens_network_manual_merged_pairs_final"].apply(len).gt(0)
    | df_03_final["tokens_network_removed_lowinfo_final"].apply(len).gt(0)
].copy()

sample_priority = sample_priority[sample_cols].head(300)

sample_random = df_03_final[sample_cols].sample(
    n=min(300, len(df_03_final)),
    random_state=42
)

sample_out_df = pd.concat([sample_priority, sample_random], ignore_index=True).drop_duplicates(
    subset=["v1_row_id"]
)

# ------------------------------------------------------------
# 4) column metadata
# ------------------------------------------------------------

columns_df = pd.DataFrame({
    "column": df_03_final.columns,
    "dtype": [str(df_03_final[col].dtype) for col in df_03_final.columns],
    "non_null_count": [int(df_03_final[col].notna().sum()) for col in df_03_final.columns]
})

# ------------------------------------------------------------
# 5) 처리 기준 Excel 저장용 요약
# ------------------------------------------------------------

processing_summary_df = pd.DataFrame({
    "항목": [
        "최종 입력 파일",
        "최종 출력 parquet",
        "기준 입력 토큰열",
        "최종 reviewed 토큰열",
        "수동 재결합 판단 파일",
        "재결합 규칙 수",
        "최종 제거 토큰 unique 수",
        "보존/보류 토큰 수",
        "06A 권장 입력 토큰열"
    ],
    "값": [
        str(NETWORK_MASTER_FILE),
        str(out_parquet),
        "tokens_analysis_network_final",
        "tokens_analysis_network_final_reviewed",
        str(MERGE_DECISION_XLSX),
        len(merge_rules_final_df),
        len(final_remove_tokens),
        len(keep_hold_token_decision_df),
        "tokens_analysis_network_final_reviewed"
    ]
})

# ------------------------------------------------------------
# 6) 파일 저장
# ------------------------------------------------------------

if SAVE_OUTPUT:
    df_03_final.to_parquet(out_parquet, index=False)

    with pd.ExcelWriter(out_sample_xlsx, engine="openpyxl") as writer:
        sample_out_df.to_excel(writer, sheet_name="sample", index=False)
        final_token_apply_summary_df.to_excel(writer, sheet_name="apply_summary", index=False)
        check_token_df.to_excel(writer, sheet_name="check_tokens", index=False)

    columns_df.to_csv(out_columns_csv, index=False, encoding="utf-8-sig")
    save_check_df.to_csv(out_save_check_csv, index=False, encoding="utf-8-sig")

    with pd.ExcelWriter(out_decision_xlsx, engine="openpyxl") as writer:
        processing_summary_df.to_excel(writer, sheet_name="summary", index=False)
        removed_token_decision_df.to_excel(writer, sheet_name="removed_tokens", index=False)
        merge_rules_final_df.to_excel(writer, sheet_name="merge_rules", index=False)
        keep_hold_token_decision_df.to_excel(writer, sheet_name="keep_hold_tokens", index=False)
        key_merge_check_df.to_excel(writer, sheet_name="key_merge_check", index=False)
        check_token_df.to_excel(writer, sheet_name="final_token_check", index=False)
        final_token_apply_summary_df.to_excel(writer, sheet_name="apply_summary", index=False)

# ------------------------------------------------------------
# 7) 06A 전달 메모 생성
# ------------------------------------------------------------

handoff_text = f"""# 03 최종 네트워크 토큰 처리 결과 전달 메모

생성 시각: {RUN_TS_FINAL}
RUN_VERSION: {RUN_VERSION}

## 1. 입력 파일

- 입력 parquet:
  {NETWORK_MASTER_FILE}

- 수동 재결합 판단 Excel:
  {MERGE_DECISION_XLSX}

## 2. 최종 출력 파일

- 최종 parquet:
  {out_parquet}

- 최종 토큰 처리 기준 목록 Excel:
  {out_decision_xlsx}

- 샘플 검토 파일:
  {out_sample_xlsx}

- 저장 검증 파일:
  {out_save_check_csv}

## 3. 06A에서 사용할 입력

06A는 아래 parquet를 입력으로 사용한다.

{out_parquet}

06A에서 사용할 토큰 열은 아래와 같다.

tokens_analysis_network_final_reviewed

## 4. 처리 흐름

기준 토큰열:
tokens_analysis_network_final

처리 순서:
1. 최신 수동 재결합 판단표의 merge 행을 기준으로 인접 2-gram/3-gram 재결합 적용
2. 최종 저정보 토큰 제거 목록 적용
3. 최종 토큰열 tokens_analysis_network_final_reviewed 생성

## 5. 최종 제거 토큰

{", ".join(final_remove_tokens)}

## 6. 보존/보류 토큰

보존/보류 목록은 아래 파일의 keep_hold_tokens 시트를 확인한다.

{out_decision_xlsx}

주요 보존/보류:
- 대회, 행사, 종목, 교실, 센터, 역량강화
- 시각장애인, 지체장애인, 평생교육, 평생학습, 주간보호, 주간보호시설, 주간보호센터

## 7. 주요 검증 결과

- 행 수: {len(df_03_final)}
- 열 수: {df_03_final.shape[1]}
- 최종 토큰 빈 행 수: {int(df_03_final[f"{FINAL_REVIEWED_TOKEN_COL}_empty"].sum())}
- 재결합 규칙 수: {len(merge_rules_final_df)}
- 최종 제거 토큰 unique 수: {len(final_remove_tokens)}
- 최종 제거 토큰 잔존 수: {len(removed_still_remaining)}

## 8. 비고

기존 tokens_analysis_primary, tokens_analysis_network, tokens_analysis_network_final은 보존하였다.
최종 분석 및 06A 재실행에는 tokens_analysis_network_final_reviewed를 사용한다.
"""

if SAVE_OUTPUT:
    with open(out_handoff_txt, "w", encoding="utf-8") as f:
        f.write(handoff_text)

# ------------------------------------------------------------
# 8) 저장 확인
# ------------------------------------------------------------

saved_files_check = pd.DataFrame({
    "file_type": [
        "final parquet",
        "sample xlsx",
        "columns csv",
        "save check csv",
        "decision xlsx",
        "handoff txt"
    ],
    "path": [
        str(out_parquet),
        str(out_sample_xlsx),
        str(out_columns_csv),
        str(out_save_check_csv),
        str(out_decision_xlsx),
        str(out_handoff_txt)
    ],
    "exists": [
        out_parquet.exists(),
        out_sample_xlsx.exists(),
        out_columns_csv.exists(),
        out_save_check_csv.exists(),
        out_decision_xlsx.exists(),
        out_handoff_txt.exists()
    ]
})

print("\n[저장 파일 확인]")
display(saved_files_check)

print("\n03-TOKEN-FINAL-4 완료")
print("이제 06A에서는 아래 파일과 토큰열을 사용하세요.")
print("입력 parquet:", out_parquet)
print("토큰열:", FINAL_REVIEWED_TOKEN_COL)
```
